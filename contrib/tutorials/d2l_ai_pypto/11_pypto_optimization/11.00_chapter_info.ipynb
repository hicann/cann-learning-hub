# pypto 优化算法
本章内容改编自[《Dive into Deep Learning》第11章「优化算法」](https://zh.d2l.ai/chapter_optimization)。在本章中，我们深入探讨常见的深度学习优化算法，包括了凸优化的入门，以及凸目标函数上非常简单的随机梯度下降算法的证明。

- 了解张量的基本操作（创建、运算、索引及广播）。
