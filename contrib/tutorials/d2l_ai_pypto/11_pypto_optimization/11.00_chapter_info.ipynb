# pypto 优化算法
本章内容改编自[《Dive into Deep Learning》第11章「优化算法」](https://zh.d2l.ai/chapter_optimization)。在本章中，我们深入探讨常见的深度学习优化算法，包括了凸优化的入门，以及凸目标函数上非常简单的随机梯度下降算法的证明。

---

## 前置要求
为了充分掌握本章内容，你应已具备以下能力：
- 掌握 Python 基础语法，能够编写和运行 Python 脚本。
- 了解张量的基本操作（创建、运算、索引及广播）。
- 理解自动微分与反向传播的基本原理。
- 了解线性回归、多层感知机等基本模型的结构与训练流程。
- 了解卷积神经网络的基本结构，能够使用 pypto 在 NPU 上构建和训练模型。


---

## 章节目标
完成本课程后，你将能够：
- 理解优化与深度学习之间的关系，区分优化目标与泛化目标的差异。
- 理解凸性在优化算法分析中的作用及其对实际问题的指导意义。
- 掌握梯度下降、随机梯度下降及小批量随机梯度下降的原理与实现方式。
- 理解动量法（Momentum）加速收敛的机制，能够应用于病态条件目标函数。
- 掌握 AdaGrad、RMSProp、Adadelta 和 Adam 等自适应学习率优化算法的原理与区别。
- 使用学习率调度器（Learning Rate Scheduler）设计学习率衰减与预热策略。
- 使用 pypto 在 NPU 上实现并比较不同优化算法及学习率调度策略的训练效果。
