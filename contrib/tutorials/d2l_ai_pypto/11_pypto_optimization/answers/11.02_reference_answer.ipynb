本节的解答思路参考了 [《动手学深度学习》习题解答](https://datawhalechina.github.io/d2l-ai-solutions-manual/#/ch11/ch11)。

### 练习11.2.1
假设我们想要通过绘制集合内点之间的所有直线并检查这些直线是否包含来验证集合的凸性。

i. 证明只检查边界上的点是充分的。

ii. 证明只检查集合的顶点是充分的。

**解答：**

i. 设闭集 $C$ 中任意两边界点的连线都含于 $C$。任取 $\boldsymbol{a},\boldsymbol{b}\in C$，过它们的直线交边界于 $\boldsymbol{p},\boldsymbol{q}$，则 $[\boldsymbol{a},\boldsymbol{b}]\subseteq[\boldsymbol{p},\boldsymbol{q}]\subseteq C$，故 $C$ 凸。

ii. 多面体是其顶点的凸包 $P=\mathrm{conv}(\boldsymbol{v}_1,\dots,\boldsymbol{v}_k)$。对任意 $\boldsymbol{x},\boldsymbol{y}\in P$，写 $\boldsymbol{x}=\sum_i\alpha_i\boldsymbol{v}_i$、$\boldsymbol{y}=\sum_i\beta_i\boldsymbol{v}_i$（$\alpha_i,\beta_i\ge0$，各自和为 $1$），则

$$t\boldsymbol{x}+(1-t)\boldsymbol{y}=\sum_i\big(t\alpha_i+(1-t)\beta_i\big)\boldsymbol{v}_i\in P,$$

故 $P$ 凸，其凸性由顶点完全决定。

### 练习11.2.2
用$p$-范数表示半径为$r$的球，证明
$$B_p[r]:=\big\{\boldsymbol{x}\mid \boldsymbol{x}\in \mathbb R^d \text{ and } \|\boldsymbol{x}\|_p \le r\big\}$$
对所有 $p\ge 1$ 是凸集。

**解答：**

在这个问题中，我们需要证明，对于所有的p范数，半径为r的球的凸集性。p‑范数形式的说明如下：
$$\|x\|_p = \Big(\sum_{i=1}^n |x_i|^p\Big)^{\frac{1}{p}}$$
其中 $x \in \mathbb R^n$ 是一个n维向量，$p \ge 1$。

要证明一个集合是凸（convex）的，我们需要满足以下条件：对于该集合中的任意两点，将这两点连接起来形成的线段也在该集合中。

对于定义的集合 $\mathcal B_p[r]$，如果 $x,y \in \mathcal B_p[r]$，我们需要证明在 $0 \le t \le 1$ 的任意情况下，$tx + (1-t)y \in \mathcal B_p[r]$。这里，$tx + (1-t)y$ 表示$x$和$y$之间的一条线段。

由于$x$和$y$满足$p$-范数小于或等于$r$，我们有 $\|x\|_p \le r$ 和 $\|y\|_p \le r$。

根据$t$，我们可以得到：
$$
\begin{aligned}
\|tx + (1 - t)y\|_p
&\le t\|x\|_p + (1 - t)\|y\|_p \\
&\le tr + (1 - t)r = r
\end{aligned}
$$

最后一步我们用到的是三角不等式，这里$\|\cdot\|_p$是一个范数，所以满足三角不等式。

因此，我们证明了对于所有的 $p \ge 1$，集合 $\mathcal B_p[r]$ 是凸的。

### 练习11.2.3
已知凸函数$f$和$g$表明$\mathrm{max}(f, g)$也是凸函数。证明$\mathrm{min}(f, g)$是非凸的。

**解答：**

在这个问题中中，我们需要分析最大函数（$\max$）和最小函数（$\min$）的凸性质。

首先，我们来分析最大函数的凸性。假设 $f$ 和 $g$ 是凸函数，我们需要证明 $\max(f,g)$ 也是凸函数。

对于任意的 $x_1,x_2\in\mathbb R$ 和 $0\le t\le 1$，我们有：
$$
\max\big(tf(x_1)+(1-t)f(x_2),\ tg(x_1)+(1-t)g(x_2)\big)
$$

我们可以将其拆分为两个部分：
1. $\max\big(tf(x_1),\ tg(x_1)\big)$
2. $\max\big((1-t)f(x_2),\ (1-t)g(x_2)\big)$

对于第一部分，由于 $f$ 和 $g$ 是凸函数，我们有：
$$
tf(x_1)+(1-t)f(x_2) \le f\big(tx_1+(1-t)x_2\big),\quad
tg(x_1)+(1-t)g(x_2) \le g\big(tx_1+(1-t)x_2\big)
$$

因此，
$$
\max\big(tf(x_1),tg(x_1)\big) \le \max\big(f(tx_1+(1-t)x_2),\,g(tx_1+(1-t)x_2)\big)
$$

对于第二部分，我们可以进行类似的推导。

综上所述，我们有：
$$
\max\big(tf(x_1)+(1-t)f(x_2),\ tg(x_1)+(1-t)g(x_2)\big)
\le \max\big(f(tx_1+(1-t)x_2),\,g(tx_1+(1-t)x_2)\big)
$$

因此，我们证明了最大函数 $\max(f,g)$ 是凸函数。

接下来，我们来证明最小函数 $\min(f,g)$ 是非凸的。

假设 $f(x)=x$ 和 $g(x)=-x$，我们可以看到 $f$ 和 $g$ 都是凸函数，因为它们的二阶导数恒为零。

然而，$\min(f,g)=\min(x,-x)=-|x|$，这是一个非凸函数。

因此，我们证明了最小函数 $\min(f,g)$ 是非凸的。

### 练习11.2.4
证明Softmax函数的规范化是凸的，即$f(x) = \log \sum_i \exp(x_i)$的凸性。

**解答：**

记 $s_i=\mathrm{softmax}(x_i)=\dfrac{e^{x_i}}{\sum_j e^{x_j}}$，则 $\sum_i s_i=1$。

一阶导：$\dfrac{\partial f}{\partial x_i}=s_i$。

Hessian：$\dfrac{\partial^2 f}{\partial x_i\partial x_j}=s_i(\delta_{ij}-s_j)$，即 $H=\mathrm{diag}(\boldsymbol{s})-\boldsymbol{s}\boldsymbol{s}^\top$。对任意 $\boldsymbol{z}$，

$$\boldsymbol{z}^\top H\boldsymbol{z}=\sum_i s_i z_i^2-\Big(\sum_i s_i z_i\Big)^2\ge 0,$$

因由 Cauchy–Schwarz（取 $a_i=\sqrt{s_i}$、$b_i=\sqrt{s_i}\,z_i$）有 $\big(\sum_i s_i z_i\big)^2\le\big(\sum_i s_i\big)\big(\sum_i s_i z_i^2\big)=\sum_i s_i z_i^2$。故 $H\succeq 0$，$f$ 凸。

### 练习11.2.5
证明线性子空间$\mathcal{X} = \{\mathbf{x} | \mathbf{W} \mathbf{x} = \mathbf{b}\}$是凸集。

**解答：**

假设$X$非空，即存在至少一个解$\boldsymbol x$满足$\boldsymbol W\boldsymbol x = \boldsymbol b$。这可以通过选择$\boldsymbol x = \boldsymbol W^{-1}\boldsymbol b$来实现，其中$\boldsymbol W^{-1}$表示$\boldsymbol W$的逆矩阵，我们选择任意两个解$\boldsymbol x_1$和$\boldsymbol x_2$，满足$\boldsymbol W\boldsymbol x_1 = \boldsymbol b$和$\boldsymbol W\boldsymbol x_2 = \boldsymbol b$。

我们选择介于$\boldsymbol x_1$和$\boldsymbol x_2$之间的向量$\boldsymbol y$，可以表示为$\boldsymbol y = \lambda \boldsymbol x_1 + (1-\lambda)\boldsymbol x_2$，其中$0 \le \lambda \le 1$。

证明$\boldsymbol W\boldsymbol y = \boldsymbol b$过程如下：
$$
\boldsymbol W\boldsymbol y = \boldsymbol W(\lambda \boldsymbol x_1 + (1-\lambda)\boldsymbol x_2)
$$

由于$\boldsymbol W\boldsymbol x_1 = \boldsymbol b$和$\boldsymbol W\boldsymbol x_2 = \boldsymbol b$，可以得到：
$$
\boldsymbol W\boldsymbol y = \lambda \boldsymbol W\boldsymbol x_1 + (1-\lambda)\boldsymbol W\boldsymbol x_2 = \lambda \boldsymbol b + (1-\lambda)\boldsymbol b = \boldsymbol b
$$

因此，对于任意介于$\boldsymbol x_1$和$\boldsymbol x_2$之间的向量$\boldsymbol y$，都有$\boldsymbol W\boldsymbol y = \boldsymbol b$。

根据凸集的定义，我们证明了对于任意两个解$\boldsymbol x_1$和$\boldsymbol x_2$，以及介于$\boldsymbol x_1$和$\boldsymbol x_2$之间的向量$\boldsymbol y$，都有$\boldsymbol W\boldsymbol y = \boldsymbol b$。因此，线性子空间$X=\{\boldsymbol x\mid \boldsymbol W\boldsymbol x=\boldsymbol b\}$是凸集。

### 练习11.2.6
证明在线性子空间$\mathbf{b} = \mathbf{0}$的情况下，对于矩阵$\mathbf{M}$的投影$\mathrm {Proj} \mathcal{X}$可以写成$\mathbf{M} \mathbf{X}$。

**解答：**

设 $\mathcal{X}=\{M\boldsymbol{z}\mid\boldsymbol{z}\in\mathbb{R}^k\}$ 为 $M$ 的列空间（$\boldsymbol{b}=\boldsymbol{0}$ 时为过原点的子空间）。$\boldsymbol{v}$ 到 $\mathcal{X}$ 的投影是其中与 $\boldsymbol{v}$ 最近的点：

$$\min_{\boldsymbol{z}}\|\boldsymbol{v}-M\boldsymbol{z}\|^2.$$

对 $\boldsymbol{z}$ 求导令其为零：$M^\top(\boldsymbol{v}-M\boldsymbol{z})=\boldsymbol{0}$，$M$ 列满秩时

$$\boldsymbol{z}^*=(M^\top M)^{-1}M^\top\boldsymbol{v},\qquad \mathrm{Proj}_{\mathcal{X}}(\boldsymbol{v})=M\boldsymbol{z}^*=M(M^\top M)^{-1}M^\top\boldsymbol{v}.$$

记系数 $\boldsymbol{X}=\boldsymbol{z}^*$，则投影可写作 $M\boldsymbol{X}$。

### 练习11.2.7
证明对于凸二次可微函数$f$，对于$\xi \in [0, \epsilon]$，我们可以写成$f(x + \epsilon) = f(x) + \epsilon f'(x) + \frac{1}{2} \epsilon^2 f''(x + \xi)$。

**解答：**

这是带 Lagrange 余项的 Taylor 公式，对任意二次可微函数成立，不依赖凸性。

由带积分余项的展开

$$f(x+\epsilon)=f(x)+\epsilon f'(x)+\int_x^{x+\epsilon}(x+\epsilon-t)\,f''(t)\,\mathrm{d}t.$$

$f''$ 连续，由积分中值定理存在 $\xi\in[0,\epsilon]$ 使

$$\int_x^{x+\epsilon}(x+\epsilon-t)\,f''(t)\,\mathrm{d}t=f''(x+\xi)\int_x^{x+\epsilon}(x+\epsilon-t)\,\mathrm{d}t=\tfrac12\epsilon^2 f''(x+\xi),$$

其中 $\int_x^{x+\epsilon}(x+\epsilon-t)\,\mathrm{d}t=\tfrac12\epsilon^2$。代入即得。

### 练习11.2.8
给定一个凸集$\mathcal{X}$和两个向量$\mathbf{x}$和$\mathbf{y}$证明了投影不会增加距离，即$\|\mathbf{x} - \mathbf{y}\| \geq \|\mathrm{Proj}_\mathcal{X}(\mathbf{x}) - \mathrm{Proj}_\mathcal{X}(\mathbf{y})\|$。

**解答：**

记 $\boldsymbol{p}=\mathrm{Proj}_X(\boldsymbol{x})$、$\boldsymbol{q}=\mathrm{Proj}_X(\boldsymbol{y})$。投影的变分刻画为：$\boldsymbol{p}\in X$ 且对任意 $\boldsymbol{z}\in X$

$$\langle\boldsymbol{x}-\boldsymbol{p},\,\boldsymbol{z}-\boldsymbol{p}\rangle\le0.$$

分别取 $\boldsymbol{z}=\boldsymbol{q}$、$\boldsymbol{z}=\boldsymbol{p}$ 得

$$\langle\boldsymbol{x}-\boldsymbol{p},\,\boldsymbol{q}-\boldsymbol{p}\rangle\le0,\qquad \langle\boldsymbol{y}-\boldsymbol{q},\,\boldsymbol{p}-\boldsymbol{q}\rangle\le0.$$

于是

$$\|\boldsymbol{p}-\boldsymbol{q}\|^2=\langle\boldsymbol{x}-\boldsymbol{y},\,\boldsymbol{p}-\boldsymbol{q}\rangle+\underbrace{\langle\boldsymbol{p}-\boldsymbol{x},\,\boldsymbol{p}-\boldsymbol{q}\rangle}_{\le0}+\underbrace{\langle\boldsymbol{y}-\boldsymbol{q},\,\boldsymbol{p}-\boldsymbol{q}\rangle}_{\le0}\le\langle\boldsymbol{x}-\boldsymbol{y},\,\boldsymbol{p}-\boldsymbol{q}\rangle\le\|\boldsymbol{x}-\boldsymbol{y}\|\,\|\boldsymbol{p}-\boldsymbol{q}\|,$$

两边除以 $\|\boldsymbol{p}-\boldsymbol{q}\|$ 得 $\|\boldsymbol{p}-\boldsymbol{q}\|\le\|\boldsymbol{x}-\boldsymbol{y}\|$。
