# pypto 计算机视觉
本章内容改编自[《Dive into Deep Learning》第12章「计算机视觉」](https://zh.d2l.ai/chapter_computer-vision/index.html)，使用 pypto 框架在昇腾 NPU 上实现。涵盖图像增广与微调、目标检测与边界框、锚框与多尺度检测、SSD 与 R-CNN 系列、语义分割与转置卷积、全卷积网络、神经风格迁移。

---

## 前置要求
为了充分掌握本章内容，你应已具备以下能力：
- 掌握卷积神经网络的基础结构（卷积层、汇聚层、批量归一化等）。
- 理解现代卷积网络（ResNet、GoogLeNet 等）的设计思想。
- 熟悉 PyTorch / pypto 张量运算与自动微分。
- 能够搭建并训练基础的图像分类模型。

---

## 章节目标
完成本课程后，你将能够：
- 使用图像增广和微调提升图像分类模型的泛化能力。
- 理解目标检测中的边界框、锚框、交并比等核心概念。
- 掌握 SSD 与 R-CNN 系列多尺度目标检测模型的实现。
- 理解语义分割与全卷积网络的工作原理。
- 运用转置卷积实现特征图的上采样。
- 实现基于 CNN 的神经风格迁移。
