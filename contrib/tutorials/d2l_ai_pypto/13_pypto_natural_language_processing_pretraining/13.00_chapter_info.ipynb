# pypto 自然语言处理：预训练

本章内容改编自[《Dive into Deep Learning》第13章「自然语言处理：预训练」](https://zh.d2l.ai/chapter_natural-language-processing-pretraining/)，使用 pypto 框架在昇腾 NPU 上实现。涵盖 word2vec、近似训练、预训练词嵌入的数据集、GloVe、子词嵌入、词的相似性和类比任务，以及 BERT 预训练等关键主题。

---

## 章节目标
完成本课程后，你将能够：
- 理解词嵌入的基本概念，以及 word2vec 的跳元模型与连续词袋（CBOW）模型。
- 掌握负采样和分层 softmax 两种近似训练方法。
- 理解预训练词嵌入数据集的构建流程（下采样、负采样、小批量加载）。
- 理解 GloVe 模型与全局语料统计的关系。
- 理解子词嵌入（fastText、字节对编码）的基本思想。
- 理解 BERT 的架构、输入表示与两个预训练任务（掩蔽语言模型、下一句预测）。
- 了解 BERT 预训练数据集的构建与预训练过程。
