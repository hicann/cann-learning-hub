### 练习14.4.1
机器翻译长期以来一直是基于翻译输出和翻译真实值之间的表面$n$元语法匹配来进行评估的。可以设计一种用自然语言推断来评价机器翻译结果的方法吗？

**解答：**

 可以。随着自然语言处理技术的发展，自然语言推断已经成为了一种有效的评价机器翻译结果的方法。自然语言推断可以通过比较翻译输出和翻译真实值之间的逻辑关系来评估翻译质量。例如，可以使用逻辑推理技术来检测翻译输出是否与翻译真实值一致，或者是否存在逻辑矛盾或歧义。此外，还可以使用语义推理技术来比较翻译输出和翻译真实值之间的语义相似性。因此，自然语言推断可以为机器翻译的评估提供更加准确和全面的方法。

### 练习14.4.2
我们如何更改超参数以减小词表大小？

**解答：**

本节构建词表时使用的是 `Vocab(train_tokens, min_freq=5)`（见 `SNLIDataset`），因此最直接相关的超参数是最小词频 `min_freq`：

1. **提高最小词频 `min_freq`**：`min_freq` 越大，只有出现次数足够多的词元才会被保留在词表中，其余低频词元都被映射为 `<unk>`，词表随之变小。例如把 `min_freq` 从 5 提高到 50，可以明显减小词表大小。

2. **改用更粗粒度的词元化方式**：例如把词级（word）词元化换成字符级（char）词元化，字符词表只有几十个词元，远小于词级词表。

3. **限制词表的最大长度**：在构建 `Vocab` 时按词频排序，只保留频率最高的前 N 个词元（`max_size` 截断），其余词元均视为 `<unk>`。

4. **去除停用词或过滤数据**：去掉“的/是/在”等高频停用词、清洗低质量样本，可以减少语料中出现的词元种类，从而减小词表。

注意：词表大小与词向量维度（`embed_size`）无关，减小 `embed_size` 只会改变每个词元的表示维度，并不会减少词表中的词元数量。
