# pypto 自然语言处理：应用
本章内容改编自[《Dive into Deep Learning》第15章「自然语言处理：应用」](https://zh.d2l.ai/chapter_natural-language-processing-applications)。本章将探讨两种流行且具有代表性的下游自然语言处理任务： 情感分析和自然语言推断，它们分别分析单个文本和文本对之间的关系。

---

## 前置要求
为了充分掌握本章内容，你应已具备以下能力：
- 掌握 Python 基础语法，能够编写和运行 Python 脚本。
- 了解张量的基本操作（创建、运算、索引及广播）。
- 了解词嵌入与预训练词向量（如 GloVe）的基本概念。
- 理解循环神经网络（RNN/LSTM）及其双向结构的基本原理。
- 了解卷积神经网络和注意力机制的基本原理。
- 能够使用 pypto 在 NPU 上构建、训练和评估深度学习模型。


---

## 章节目标
完成本课程后，你将能够：
- 理解情感分析任务，能够对 IMDb 影评数据集进行读取、分词和预处理。
- 使用双向 LSTM（BiRNN）实现情感分析模型，并在 NPU 上完成训练与评估。
- 使用一维卷积（textCNN）实现情感分析模型，掌握卷积与池化在文本处理中的应用。
- 理解自然语言推断任务，能够对 SNLI 数据集进行解析和预处理。
- 实现基于可分解注意力机制的自然语言推断模型（Attend、Compare、Aggregate）。
- 理解 BERT 微调范式，掌握针对序列级和词元级下游任务的微调方法。
- 使用微调后的 BERT 模型完成自然语言推断任务。
