# pypto 循环神经网络

本章内容改编自[《Dive into Deep Learning》第8章「循环神经网络」](https://zh.d2l.ai/chapter_recurrent-neural-networks/)，使用 pypto 框架在昇腾 NPU 上实现。涵盖序列模型、文本预处理、语言模型、循环神经网络（RNN）及其从零实现与简洁实现、通过时间反向传播等关键主题。

---

## 前置要求
为了充分掌握本章内容，你应已具备以下能力：
- 掌握 Python 基础语法，能够编写和运行 Python 脚本。
- 了解张量的基本操作（创建、运算、索引及广播）。
- 理解线性回归和 softmax 回归的基本原理。
- 理解多层感知机（MLP）的架构和训练方法。
- 了解自动微分和反向传播的基本概念。

---

## 章节目标
完成本课程后，你将能够：
- 理解序列模型的基本概念和统计工具。
- 掌握文本预处理的常用技术（分词、词表构建）。
- 理解语言模型和 n 元语法的基本原理。
- 理解循环神经网络（RNN）的架构和原理。
- 使用 pypto 在 NPU 上从零实现 RNN。
- 使用 pypto 简洁实现 RNN 并进行训练。
- 理解通过时间反向传播（BPTT）的梯度计算方法。
