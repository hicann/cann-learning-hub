# pypto 深度学习计算
本章内容改编自[《Dive into Deep Learning》第5章「深度学习计算」](https://zh.d2l.ai/chapter_deep-learning-computation)，使用 pypto 框架在昇腾 NPU 上实现。涵盖层与块的构建、参数访问与初始化、延后初始化、自定义层（含 PyPTO 算子）、模型读写以及 NPU 设备的使用。

---

## 前置要求
为了充分掌握本章内容，你应已具备以下能力：
- 掌握 Python 基础语法，能够编写和运行 Python 脚本。
- 了解张量的基本操作（创建、运算、索引及广播）。
- 理解自动微分与反向传播的基本原理。
- 了解线性回归、多层感知机等基本模型的结构与训练流程。

---

## 章节目标
完成本课程后，你将能够：
- 理解层与块（`nn.Module`）的嵌套结构，能够自定义网络块并管理其中的子模块。
- 掌握模型参数的访问、初始化与共享方式。
- 理解延后初始化的机制，能够使用 Lazy 系列模块按需推断参数形状。
- 能够设计并实现自定义层，并结合 PyPTO 在 NPU 上编写算子内核。
- 掌握模型参数与网络结构的保存与加载方法。
- 能够在 CPU 与 NPU 之间迁移张量与模型，理解设备间数据搬运的开销。
