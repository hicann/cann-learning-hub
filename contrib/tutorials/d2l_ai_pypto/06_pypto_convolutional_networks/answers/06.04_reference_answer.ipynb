# 多输入多输出通道

## 练习 6.4.1

假设我们有两个卷积核，大小分别为 $k_1$ 和 $k_2$（中间没有非线性激活函数）。
1. 证明运算可以用单次卷积来表示。
2. 这个等效的单个卷积核的维数是多少呢？
3. 反之亦然吗？

### 解答

**第1问：**

&emsp;&emsp;假设输入为 $x$ ，先用大小为 $k_1$ 的卷积核进行卷积，得到中间结果 $z = conv(x, k_1)$ ，再用大小为 $k_2$ 的卷积核进行卷积，得到最终输出 $y = conv(z, k_2)$ ，两个卷积之间没有非线性激活函数。两次顺序卷积可以合并为一次卷积：

$$y = conv(conv(x, k_1), k_2) = conv(x, k_1 * k_2)$$

&emsp;&emsp;其中 $k_1 * k_2$ 表示对两个卷积核做互相关（即核与核的卷积），其元素为 $(k_1 * k_2)[i][j] = \sum_{i',j'} k_1[i'][j'] \times k_2[i-i'][j-j']$ 。

**第2问：**

&emsp;&emsp;可以使用一个大小为 $(k_1+k_2-1)×(k_1+k_2-1)$ 的卷积核来表示这两个卷积核的运算。

**第3问：**

&emsp;&emsp;反之亦然。任意大小为 $(k_1+k_2-1)\times(k_1+k_2-1)$ 的卷积核都可以分解为两个大小分别为 $k_1$ 和 $k_2$ 的卷积核的互相关（例如取 $k_1$ 为 $1\times 1$ 的核，则 $k_1 * k_2 = k_2$ 即原核，分解退化为平凡情形）。因此合并后的运算也可以反过来拆分为两次顺序卷积。

## 练习 6.4.2

假设输入为 $c_i\times h\times w$ ，卷积核大小为 $c_o\times c_i\times k_h\times k_w$ ，填充为 $(p_h, p_w)$，步幅为$(s_h, s_w)$。
1. 前向传播的计算成本（乘法和加法）是多少？
1. 内存占用是多少？
1. 反向传播的内存占用是多少？
1. 反向传播的计算成本是多少？

### 解答

**第1问：**

&emsp;&emsp;前向计算成本为

$$flops_{forward} = c_i \times c_o \times k_h \times k_w \times m_h \times m_w$$

&emsp;&emsp;其中 $m_h=\lfloor \frac{h+2p_h-k_h}{s_h}+1 \rfloor$ , $m_w=\lfloor \frac{w+2p_w-k_w}{s_w}+1 \rfloor$ 。


**第2问：**

&emsp;&emsp;输入张量需要 $c_i \times h \times w$ 个浮点数，卷积核需要 $c_o \times c_i \times k_h \times k_w$ 个浮点数，输出张量需要 $c_o \times m_h \times m_w$ 个浮点数（$m_h$、$m_w$ 定义同上）。因此，前向内存占用为 
$$memory_{forward}=c_i*h*w+c_o*c_i*k_h*k_w+c_o*m_h*m_w$$

**第3问：**

&emsp;&emsp;反向传播的内存占用与前向传播相同（还需要额外存储各层梯度，量级相同），总内存占用为
$$memory_{backward}=c_i*h*w+c_o*c_i*k_h*k_w+c_o*m_h*m_w$$

**第4问：**

&emsp;&emsp;反向计算成本为

$$flops_{backward} = c_i \times c_o \times k_h \times k_w \times m_h \times m_w$$

&emsp;&emsp;其中 $m_h$ 和 $m_w$ 的定义同上。

## 练习 6.4.3

如果我们将输入通道 $c_i$ 和输出通道 $c_o$ 的数量加倍，计算数量会增加多少？如果我们把填充数量翻一番会怎么样？

### 解答

&emsp;&emsp;如果我们将输入通道 $c_i$ 和输出通道 $c_o$ 的数量加倍，计算数量会增加 $4$ 倍。如果我们把填充数量翻一番，计算数量会增加 $2$ 倍。

## 练习 6.4.4

如果卷积核的高度和宽度是 $k_h=k_w=1$ ，前向传播的计算复杂度是多少？

### 解答

&emsp;&emsp;当 $k_h=k_w=1$ 时，每次输出元素需要 $c_i$ 次乘加，输出元素数为 $c_o \times m_h \times m_w$ ，因此

$$flops = c_i \times c_o \times m_h \times m_w$$

&emsp;&emsp;其中 $m_h=\lfloor \frac{h+2p_h-1}{s_h}+1 \rfloor$ , $m_w=\lfloor \frac{w+2p_w-1}{s_w}+1 \rfloor$ 。

## 练习 6.4.5

本节最后一个示例中的变量 $Y1$ 和 $Y2$ 是否完全相同？为什么？

### 解答

&emsp;&emsp;理论上相同，但由于浮点数计算有误差，因而两者不完全相同（有微小差异）。

## 练习 6.4.6

当卷积窗口不是 $1\times 1$ 时，如何使用矩阵乘法实现卷积？

### 解答

&emsp;&emsp;可以将输入张量和卷积核张量分别展开为二维矩阵，然后对这两个矩阵进行乘法运算，得到的结果再变换为输出张量。
