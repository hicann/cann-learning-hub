# pypto 卷积神经网络
本章内容改编自[《Dive into Deep Learning》第6章「卷积神经网络」](https://zh.d2l.ai/chapter_convolutional-neural-networks/index.html)，使用 pypto 框架在昇腾 NPU 上实现。涵盖从全连接层到卷积的动机、图像卷积、填充和步幅、多输入多输出通道、汇聚层以及经典卷积网络 LeNet。

---

## 前置要求
为了充分掌握本章内容，你应已具备以下能力：
- 掌握 Python 基础语法，能够编写和运行 Python 脚本。
- 熟悉 pypto 张量运算与自动微分。
- 理解线性回归、softmax 回归与多层感知机等基础模型。
- 能够搭建并训练基础的图像分类模型。

---

## 章节目标
完成本课程后，你将能够：
- 理解从全连接层到卷积层的动机：局部连接与权值共享。
- 掌握卷积层的工作原理（互相关运算）。
- 掌握填充与步幅对输出特征图形状的影响。
- 理解多输入多输出通道及 1×1 卷积的作用。
- 理解汇聚层的作用及其与卷积层的区别。
- 使用 pypto 在 NPU 上构建、训练和评估 LeNet。
