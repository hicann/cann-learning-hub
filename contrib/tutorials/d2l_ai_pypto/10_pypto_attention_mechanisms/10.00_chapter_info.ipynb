# pypto 注意力机制

本章内容改编自[《Dive into Deep Learning》第10章「注意力机制」](https://zh.d2l.ai/chapter_attention-mechanisms/)，使用 pypto 框架在昇腾 NPU 上实现。涵盖注意力提示、Nadaraya-Watson 核回归、注意力评分函数、Bahdanau 注意力、多头注意力、自注意力和位置编码以及Transformer等关键主题。

---

## 前置要求
为了充分掌握本章内容，你应已具备以下能力：
- 掌握 Python 基础语法，能够编写和运行 Python 脚本。
- 了解张量的基本操作（创建、运算、索引及广播）。
- 理解线性回归和 softmax 回归的基本原理。
- 理解多层感知机（MLP）的架构和训练方法。
- 理解循环神经网络（RNN）和序列到序列（seq2seq）模型。

---

## 章节目标
完成本课程后，你将能够：
- 理解注意力机制的基本概念和注意力提示。
- 掌握 Nadaraya-Watson 核回归作为注意力汇聚的示例。
- 理解注意力评分函数的设计和实现。
- 理解 Bahdanau 注意力的原理和结构。
- 掌握多头注意力的设计和实现。
- 理解自注意力和位置编码的核心思想。
- 理解 Transformer 架构并能够实现。
- 理解视觉 Transformer（ViT）的架构与实现。
- 了解大规模预训练 Transformer（BERT / T5 / GPT）与可扩展性。
