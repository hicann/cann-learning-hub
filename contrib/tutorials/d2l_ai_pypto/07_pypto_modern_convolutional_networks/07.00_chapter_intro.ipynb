# pypto 现代卷积神经网络
本章内容改编自[《Dive into Deep Learning》第7章「现代卷积神经网络」](https://zh.d2l.ai/chapter_convolutional-modern/index.html)，使用 pypto 框架在昇腾 NPU 上实现。涵盖 AlexNet、VGG、NiN、GoogLeNet、批量规范化、ResNet、DenseNet 等经典现代卷积网络架构，按历史时间顺序介绍。

---

## 前置要求
为了充分掌握本章内容，你应已具备以下能力：
- 掌握卷积神经网络的基础结构（卷积层、汇聚层、批量归一化等）。
- 理解经典卷积网络（LeNet）的设计与实现。
- 能够搭建并训练基础的图像分类模型。
- 熟悉 PyTorch / pypto 张量运算与自动微分。

---

## 章节目标
完成本课程后，你将能够：
- 理解 AlexNet 的网络结构及其在 ImageNet 上的历史意义。
- 掌握 VGG 的「块」设计思想。
- 理解 NiN 用 1×1 卷积替代全连接层的思想。
- 掌握 GoogLeNet 的多尺度并行 Inception 块结构。
- 运用批量规范化提升模型训练稳定性与收敛速度。
- 理解 ResNet 的残差连接及其对深层网络训练的关键作用。
- 理解 DenseNet 的稠密连接思想。
