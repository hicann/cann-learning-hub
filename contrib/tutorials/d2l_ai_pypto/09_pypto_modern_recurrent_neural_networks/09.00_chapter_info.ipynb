# pypto 现代循环神经网络

本章内容改编自[《Dive into Deep Learning》第9章「现代循环神经网络」](https://zh.d2l.ai/chapter_recurrent-modern/)，使用 pypto 框架在昇腾 NPU 上实现。涵盖门控循环单元（GRU）、长短期记忆网络（LSTM）、深度循环神经网络、双向循环神经网络、机器翻译与数据集、编码器-解码器架构、序列到序列学习（seq2seq）及束搜索等关键主题。

---

## 前置要求
为了充分掌握本章内容，你应已具备以下能力：
- 掌握 Python 基础语法，能够编写和运行 Python 脚本。
- 了解张量的基本操作（创建、运算、索引及广播）。
- 理解循环神经网络（RNN）的基本架构和原理。
- 了解通过时间反向传播（BPTT）的梯度计算方法。
- 掌握文本预处理和语言模型的基本概念。

---

## 章节目标
完成本课程后，你将能够：
- 理解门控循环单元（GRU）的架构与门控机制。
- 理解长短期记忆网络（LSTM）的架构与记忆单元。
- 掌握深度循环神经网络的构建方法。
- 掌握双向循环神经网络的原理与应用。
- 了解机器翻译的基本流程与数据集处理方法。
- 理解编码器-解码器架构的设计思想。
- 实现序列到序列学习（seq2seq）模型进行机器翻译。
- 掌握束搜索（beam search）等序列生成策略。
