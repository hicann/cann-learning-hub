## 09.08 束搜索 参考答案


### 练习 9.8.1

**题目：** 我们可以把穷举搜索看作一种特殊的束搜索吗？为什么？

**解答：** 可以。束搜索在每一步保留束宽 $K$ 个最优候选；当束宽 $K$ 不小于词表大小 $\left|\mathcal{Y}\right|$（即每一步保留全部候选）时，束搜索不再剪枝，退化为对全部输出序列的穷举搜索。因此穷举搜索可以看作束宽足够大的束搜索。

一般场景下，穷举搜索计算量 $\mathcal{O}(\left|\mathcal{Y}\right|^{T'})$ 巨大，而束搜索通过限制束宽 $K$ 在精度与计算量之间权衡。


### 练习 9.8.2

**题目：** 在 9.7 节的机器翻译问题中应用束搜索。束宽如何影响预测速度和结果？

**解答：** 束宽是控制候选项数量的参数。较小束宽减少搜索空间，预测速度更快；较大束宽保留更多候选项，可带来更准确结果但增加计算复杂性和预测时间。需根据实际需求权衡速度和精度。



### 练习 9.8.3

**题目：** 在 8.5 节中，我们基于前缀通过语言模型生成文本，使用了哪种搜索策略？可以改进吗？

**解答：** 8.5 节使用 `y.argmax(dim=1)` 选择最高概率字符，属于**贪心算法**。可通过束搜索改进：在每一步选择多个候选字符，根据概率得分排序，保留得分最高的候选继续生成，提高文本质量和连贯性，但会增加计算复杂度。
