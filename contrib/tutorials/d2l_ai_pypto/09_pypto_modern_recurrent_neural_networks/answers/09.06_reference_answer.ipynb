## 09.06 编码器-解码器架构 参考答案


### 练习 9.6.1

假设我们使用神经网络来实现"编码器－解码器"架构，那么编码器和解码器必须是同一类型的神经网络吗？


**答：** 不一定。编码器和解码器可以使用不同类型的神经网络。

例如：
- 在图像描述生成（Image Captioning）中，常用 CNN 作为编码器提取图像特征，RNN/LSTM 作为解码器生成文本描述；
- 在语音识别中，可能使用 CNN/BiRNN 编码音频信号，再用 RNN/Transformer 解码为文本；
- 编码器和解码器可以是任何适合输入/输出数据形式的网络架构，只需保证编码器输出的编码状态能被解码器正确接收即可。


### 练习 9.6.2

除了机器翻译，还有其它可以适用于"编码器－解码器"架构的应用吗？


**答：** 编码器-解码器架构广泛应用于各类序列转换（sequence transduction）任务，包括但不限于：

- **图像描述生成**（Image Captioning）：编码器处理图像，解码器生成文本描述；
- **语音识别**（ASR）：编码器处理音频信号，解码器输出文本；
- **文本摘要**（Text Summarization）：编码器读取原文，解码器生成摘要；
- **对话系统/聊天机器人**：编码器理解输入语句，解码器生成回复；
- **代码生成**：编码器处理自然语言描述，解码器生成代码；
- **图像分割/目标检测**：某些架构中编码器提取视觉特征，解码器生成像素级分割掩码。
