### 2.1 系统硬件信息

#### 2.1.1 CPU 信息

`lscpu` 命令可以显示 CPU 架构、核心数、主频等详细信息。

> 📁 **对应代码文件**：`code/cpu_info.py`  
> **运行方式**：`python3 code/cpu_info.py`

**`lscpu` 命令原理：**

- `lscpu` 从 `/sys/devices/system/cpu/` 和 `/proc/cpuinfo` 中读取 CPU 信息并格式化输出
- 在香橙派上，你会看到：
  - `Architecture: aarch64`（ARM 架构）
  - `CPU(s): 4`（4 个逻辑核心）
  - `Model name`（CPU 型号）
  - `CPU max MHz`（最大主频）

**获取 CPU 核心数的三种方式对比：**

<table style="text-align: left; margin-left: 0;">
<tr>
<th style="text-align: left;">方式</th>
<th style="text-align: left;">原理</th>
<th style="text-align: left;">返回值</th>
</tr>
<tr>
<td style="text-align: left;"><code>os.cpu_count()</code></td>
<td style="text-align: left;">Python 标准库，读取系统信息</td>

<td style="text-align: left;"><code>multiprocessing.cpu_count()</code></td>

</tr>
<tr>
<td style="text-align: left;"><code>nproc</code></td>

<td style="text-align: left;">逻辑核心数</td>
</tr>
</table>

> **知识点**：`/proc` 是 Linux 的**虚拟文件系统**，内容存在于内存中。`/proc/cpuinfo` 包含每个 CPU 核心的详细信息，与 `lscpu` 相比信息更原始、更详细。

**`code/cpu_info.py` 核心代码片段：**

```python
import os, subprocess, multiprocessing

# lscpu 获取 CPU 详细信息
result = subprocess.run(['lscpu'], capture_output=True, text=True)
print(result.stdout)

# 三种方式获取 CPU 核心数
print(f"  os.cpu_count()              = {os.cpu_count()}")
print(f"  multiprocessing.cpu_count() = {multiprocessing.cpu_count()}")

# /proc/cpuinfo 原始信息
result = subprocess.run(['bash', '-c', 'cat /proc/cpuinfo | head -30'], capture_output=True, text=True)
print(result.stdout)
```

**香橙派上的预期输出示例：**

#### 2.1.3 内存信息

使用 `free` 和 `/proc/meminfo` 查看系统内存使用情况。

> 📁 **对应代码文件**：`code/memory_info.py`  
> **运行方式**：`python3 code/memory_info.py`

**命令原理：**

- `free -h`：以人类可读格式（GB/MB）显示内存总量、已用、可用、缓存和交换分区
- `/proc/meminfo`：更详细的内存信息，每行一个指标
  - `MemTotal`：总内存
  - `MemAvailable`：可用内存（含可回收缓存）
  - `SwapTotal`：交换分区总量

**`code/memory_info.py` 核心代码片段：**

```python
import subprocess

# free -h 显示内存使用情况
result = subprocess.run(['free', '-h'], capture_output=True, text=True)
print(result.stdout)

# /proc/meminfo 详细内存信息
result = subprocess.run(['bash', '-c', 'cat /proc/meminfo | head -10'], capture_output=True, text=True)
print(result.stdout)
```

> **知识点**：Linux 会把空闲内存用作磁盘缓存（Cache/Buffers），`MemAvailable` 才是真正可用的内存。

#### 2.1.4 磁盘信息

> 📁 **对应代码文件**：`code/disk_info.py`  
> **运行方式**：`python3 code/disk_info.py`

**命令原理：**

- `df -h`：显示各挂载点的磁盘使用情况（总量/已用/可用/使用率）
- `lsblk`：列出所有块设备（硬盘、分区）的树状结构

**`code/disk_info.py` 核心代码片段：**

```python
import subprocess

# 磁盘使用情况
result = subprocess.run(['df', '-h'], capture_output=True, text=True)
print(result.stdout)

# 块设备信息
result = subprocess.run(['lsblk'], capture_output=True, text=True)
print(result.stdout)
```

#### 2.1.5 CPU 与 NPU 负载监控

前面我们查看了 CPU 和 NPU 的**硬件信息**（型号、核心数等）。在实际开发中，还需要**实时监控**它们的运行负载，判断是否在满负荷运转。

> 📁 **对应代码文件**：`code/load_monitor.py`  
> **运行方式**：`python3 code/load_monitor.py`

**`top` 命令原理：**

- `top` 是 Linux 下最常用的系统监控工具，实时显示 CPU、内存使用情况和进程列表
- `-b`：批处理模式（不交互，直接输出）
- `-n1`：只输出 1 次快照
- `head -20`：只看前 20 行

**输出中关注的内容：**

- `%Cpu(s)` 行：显示 CPU 整体利用率（us=用户态, sy=内核态, id=空闲）
- `load average`：1/5/15 分钟的平均负载

**NPU 监控指标说明：**

<td style="text-align: left;">Memory</td>
<td style="text-align: left;">显存使用量 / 总量</td>

**`code/load_monitor.py` 核心代码片段：**

```python
import subprocess

# CPU 负载监控 (top 快照)
result = subprocess.run(['bash', '-c', 'top -bn1 | head -20'], capture_output=True, text=True)
print(result.stdout)

# NPU 负载监控
result = subprocess.run(['npu-smi', 'info', '-t', 'usages', '-i', '0'], capture_output=True, text=True)
print(result.stdout)
```

> **知识点**：CPU 利用率 = 100% - 空闲率(id)。如果 id 接近 0，说明 CPU 几乎满负荷。

### 3.2 多核 CPU 并行程序

香橙派有 **4 个 CPU 核心**。如果任务只在一个核心上跑，就浪费了其余 3 个核心。Python 的 `multiprocessing` 模块可以让我们把任务分给多个核心**同时**执行。

> 📁 **对应代码文件**：`code/multiprocessing_demo.py`  
> **运行方式**：`python3 code/multiprocessing_demo.py`

#### A. 最简单的多进程示例

用 4 个进程同时打印消息，感受"并行"的概念：

```python
import multiprocessing
import os

num_cores = os.cpu_count()
print(f"本机 CPU 核心数: {num_cores}")

def hello_worker(worker_id):
    return f"Worker {worker_id}: Hello! (来自进程)"

num = min(4, num_cores)
with multiprocessing.Pool(num) as pool:
    results = pool.map(hello_worker, range(num))

for r in results:
    print(f"  {r}")
```

#### B. 串行 vs 并行对比

把一个计算任务分成几份，分给多个核心同时算，看看快了多少：

```python
import multiprocessing
import time

def simple_sum(n):
    return sum(range(n))

N = 5_000_000
num = min(4, multiprocessing.cpu_count())

# 串行：一个核心做 num 次
start = time.time()
results_serial = [simple_sum(N) for _ in range(num)]
t_serial = time.time() - start

# 并行：num 个核心各做 1 次
start = time.time()
with multiprocessing.Pool(num) as pool:
    results_parallel = pool.map(simple_sum, [N] * num)
t_parallel = time.time() - start

print(f"串行: {t_serial:.3f} 秒")
print(f"并行: {t_parallel:.3f} 秒")
print(f"加速比: {t_serial / t_parallel:.1f}x")
```

**关键概念：**

- **加速比（Speedup）**：$S_p = T_{串行} / T_{并行}$，值越大越好
- **并行效率** = 加速比 / 核心数 × 100%，理想情况接近 100%
- 实际中并行效率 < 100%，因为进程创建和数据传输有额外开销

> **为什么加速比 < 核心数？** 因为启动进程、分配任务、汇总结果都需要时间，这部分开销是串行没有的。

### 3.3 CPU 与 NPU 通信

香橙派中有 **CPU**（通用处理器）和 **NPU**（AI 专用处理器，型号 Ascend 310B4）。它们是两个独立的处理器，各有各的内存。

**核心概念：**

- **张量（Tensor）**：PyTorch 中的数据容器，类似数组
- **设备（Device）**：张量存放在哪里。默认在 CPU 上，用 `.npu()` 可以搬到 NPU 上
- **CPU 和 NPU 的内存是独立的**：NPU 不能直接读取 CPU 内存中的数据，必须先"搬"过去

> 📁 **对应代码文件**：`code/cpu_npu_comm.py`  
> **运行方式**：`python3 code/cpu_npu_comm.py`

**依赖包**：`torch`, `torch_npu`  
若未安装，建议执行：
```bash
pip install torch
pip install torch_npu
```

**本例演示最基本操作：在 CPU 上创建数据 → 搬到 NPU → 搬回 CPU**

```python
import torch
import torch_npu

# 第1步：在 CPU 上创建一个张量
x = torch.tensor([1.0, 2.0, 3.0])
print(f"第1步 - CPU 上的张量: {x}, 设备: {x.device}")

# 第2步：把数据从 CPU 搬到 NPU
x_npu = x.npu()
print(f"第2步 - 已搬到 NPU: {x_npu}, 设备: {x_npu.device}")

# 第3步：把数据从 NPU 搬回 CPU
x_back = x_npu.cpu()
print(f"第3步 - 搬回 CPU: {x_back}, 设备: {x_back.device}")
```

**预期输出：**

```text
NPU 是否可用: True

第1步 - CPU 上的张量: tensor([1., 2., 3.]), 设备: cpu
第2步 - 已搬到 NPU: tensor([1., 2., 3.], device='npu:0'), 设备: npu:0
第3步 - 搬回 CPU: tensor([1., 2., 3.]), 设备: cpu

小结: 数据成功从 CPU -> NPU -> CPU 搬了一个来回！
```

   提示：`lscpu` 是格式化输出，信息更易读；`/proc/cpuinfo` 是原始信息，更详细。

2. **`os.cpu_count()`、`multiprocessing.cpu_count()` 和 `nproc` 三种方式获取的 CPU 核心数是否一定相同？**

   提示：在容器或虚拟化环境中，可能因为资源限制导致结果不同。

3. **若 NPU 负载始终为 0% 而 CPU 满载，应如何排查？**

   提示：检查环境变量是否正确配置、模型是否成功加载到 NPU、是否存在 CPU fallback。

4. **串行 vs 并行的加速比为什么通常小于 CPU 核心数？**

   提示：进程创建、任务分配、结果汇总都有额外开销。

5. **`.npu()` 和 `.cpu()` 分别完成什么操作？为什么需要它们？**

   提示：CPU 和 NPU 有独立的内存，数据不能直接互访，需要显式搬运。

6. **猫狗分类案例中，为什么使用预训练模型而不自己训练模型？**

   提示：预训练模型已在 ImageNet 上训练过，无需额外训练数据和训练时间，直接推理即可获得准确结果。
