# 第2章 智能硬件系统

本章节系统介绍智能硬件系统的核心架构与关键技术，涵盖 SoC 片上系统、CPU、协处理加速单元（NPU/GPU/DSP/FPGA）、存储器单元、常用接口，并重点解析华为昇腾 AI 处理器的硬件系统、达芬奇架构与开发应用平台。

> **学习目标**：理解智能硬件 SoC 架构；掌握异构计算思想；了解昇腾 NPU 与达芬奇架构；了解昇腾开发应用平台与软件生态。

## 1. 智能硬件片上系统芯片（SoC）

### 1.1 SoC 定义与核心特点

**SoC（System on Chip，片上系统）** 是将处理器、存储器、外设接口以及各种专用计算单元高度集成在一块芯片上的系统级解决方案。

上表总结了 SoC 的三大核心特点。**高性能**意味着 SoC 内部的各计算单元（CPU、GPU、NPU 等）通过片上高速总线互联，通信延迟远低于跨芯片方案；**低功耗**得益于集成度提升后信号路径缩短、外部驱动减少，同时可针对不同任务动态开启/关闭相应计算单元；**小尺寸**则使 SoC 能够嵌入手机、摄像头、无人机等空间受限的设备中，是端侧 AI 部署的物理前提。

**广泛应用领域**：嵌入式系统与智能终端、物联网设备与边缘计算、智能摄像头与安防监控、智能家居与工业控制、无人驾驶与机器人、TinyML 与端侧 AI。

**全链路计算能力**：SoC 芯片能够支撑从 **边缘感知 → 数据处理 → 本地决策** 的全链路计算，为智能摄像头、智能家居、无人驾驶及机器人等场景提供可靠算力基础。

## 2. 中央处理单元（CPU）

### 2.1 通用 CPU 芯片架构

嵌入式 SoC 中的 CPU 往往采用**精简指令集架构（RISC）**，如 ARM 和 RISC-V，通过简化指令实现高能效比，并支持对称多处理（SMP）和非对称多处理（AMP）模式。

**ARM big.LITTLE 技术**：低功耗核心处理轻量任务，高性能核心负责复杂计算任务，实现性能与能效的平衡。

**SIMD 向量处理技术**：现代嵌入式 CPU 内核普遍集成单指令多数据（SIMD）引擎，能够一次执行多组数据操作，大幅提升矩阵运算、图像处理、音视频处理的吞吐能力。

> 实现从算法开发到边缘部署的完整闭环，极大提升开发效率和落地能力。

```python
import ctypes, ctypes.util, os
try:
    _g = ctypes.util.find_library('gomp')
    if _g: ctypes.CDLL(_g, mode=ctypes.RTLD_GLOBAL)
    else:
        for _p in ['/usr/lib/x86_64-linux-gnu/libgomp.so.1','/usr/lib/aarch64-linux-gnu/libgomp.so.1','/lib64/libgomp.so.1']:
            if os.path.exists(_p): ctypes.CDLL(_p, mode=ctypes.RTLD_GLOBAL); break
except Exception: pass

import torch
import torch_npu
import time

N = 4096

# CPU 矩阵乘法
A_cpu = torch.randn(N, N)
B_cpu = torch.randn(N, N)
t0 = time.time()
C_cpu = torch.mm(A_cpu, B_cpu)
t1 = time.time()
print(f'[CPU]  {N}x{N} 矩阵乘法耗时: {t1 - t0:.3f} 秒')

# NPU 矩阵乘法
A_npu = A_cpu.npu()
B_npu = B_cpu.npu()
torch.npu.synchronize()
t0 = time.time()
C_npu = torch.mm(A_npu, B_npu)
torch.npu.synchronize()
t1 = time.time()
print(f'[NPU]  {N}x{N} 矩阵乘法耗时: {t1 - t0:.3f} 秒')
print(f'加速比: {(time.time()):.1f}')  # 占位，实际加速比由上面两次计时决定
```
