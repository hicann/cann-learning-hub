### 步骤 1：注册 Jupyter 内核

在终端中依次执行以下命令，注册 **Python 3.11.4 (CANN)** 内核。

> ⚠️ 以下命令需在 **WebIDE 终端**（而非本 Notebook 的 Cell）中执行。此处仅作记录与演示，实际操作请打开终端运行。


**代码功能说明：**

这段代码在 WebIDE 终端中注册 Jupyter 内核，分两步：

1. **安装 ipykernel**：ipykernel 是 Jupyter 与 Python 之间的通信桥梁。有了它，Jupyter Notebook 才能逐个 Cell 运行 Python 代码。如果已安装会提示 lready satisfied，忽略即可。

2. **注册内核**：将 \/opt/buildtools/Python-3.11.4/bin/python3.11\ 这个 Python 解释器注册为 Jupyter 可选内核，显示名为 \Python 3.11.4 (CANN)\。注册后，在 Notebook 右上角就能选择这个内核来运行代码。

```python
%%bash
cd ~
# ============================================================
# 步骤 1：注册 Jupyter 内核（在 WebIDE 终端中执行）
# ============================================================

# 1) 安装 ipykernel（如已安装会提示 already satisfied，忽略即可）
/opt/buildtools/Python-3.11.4/bin/python3.11 -m pip install ipykernel

# 2) 注册内核
/opt/buildtools/Python-3.11.4/bin/python3.11 -m ipykernel install --user \
  --name cann_py311 \
  --display-name "Python 3.11.4 (CANN)"
```

### 步骤 2：设置环境变量自动加载

创建 `usercustomize.py`，让 Python 启动时**自动设置 CANN 环境变量**，避免每次手动 `source set_env.sh`。

原理：Python 在启动时会自动导入 `site-packages` 下的 `usercustomize.py`，利用此机制可将 CANN 环境变量注入到每个 Python 进程中。


**代码功能说明：**

这段代码创建 `usercustomize.py` 文件，实现 CANN 环境变量的**自动加载**。

**原理：**

- Python 启动时会自动执行 `site-packages` 目录下的 `usercustomize.py` 文件
- 我们在这个文件中用 `os.environ.setdefault()` 设置 4 个 CANN 核心环境变量：
  - `ASCEND_TOOLKIT_HOME`：CANN 工具包安装路径
  - `ASCEND_OPP_PATH`：算子库路径（OPP = Operator Package）
  - `ASCEND_HOME_PATH`：CANN 主路径
  - `ASCEND_AICPU_PATH`：AI CPU 路径
- `setdefault` 的好处是：如果环境变量已设置，不会覆盖原有值

> **为什么需要这一步？** 在 Jupyter 中运行代码时，终端的环境变量不会自动传递到 Python 进程。通过 `usercustomize.py` 可以确保每次启动 Python 都自动设置好 CANN 路径，避免找不到算子库等错误。


```python
%%bash
cd ~
# ============================================================
# 步骤 2：创建 usercustomize.py 自动加载 CANN 环境变量
# ============================================================

# 获取 CANN 路径
CANN_PATH=$(echo $ASCEND_TOOLKIT_HOME)
echo "CANN_PATH = ${CANN_PATH}"

# 创建 usercustomize.py
mkdir -p ~/.local/lib/python3.11/site-packages
cat > ~/.local/lib/python3.11/site-packages/usercustomize.py << EOF
import os
os.environ.setdefault('ASCEND_OPP_PATH', '${CANN_PATH}/opp')
os.environ.setdefault('ASCEND_TOOLKIT_HOME', '${CANN_PATH}')
os.environ.setdefault('ASCEND_HOME_PATH', '${CANN_PATH}')
os.environ.setdefault('ASCEND_AICPU_PATH', '${CANN_PATH}')
EOF

echo "--- usercustomize.py 内容 ---"
cat ~/.local/lib/python3.11/site-packages/usercustomize.py
```

## 2.2 猫狗分类深度学习案例

本例使用一个**预训练的深度学习模型**（ResNet18）对 4 张图片进行分类，判断每张图片是猫还是狗。

### 什么是深度学习分类？

- **分类**：给计算机看一张图片，让它判断图片里是什么（猫？狗？）
- **深度学习模型**：一种通过大量数据训练出来的神经网络，能自动从图片中提取特征
- **预训练模型**：已经在 ImageNet（120 万张图片，1000 个类别）上训练好的模型，我们直接使用，不需要自己训练

### 为什么用预训练模型？

- **快**：不需要训练，直接推理，几秒钟出结果
- **简单**：加载模型 → 处理图片 → 得到结果，三步完成
- **准确**：ResNet18 在 ImageNet 上训练过，已经认识各种猫和狗

### 第 1 步：查看图片

先用 `matplotlib` 显示 4 张图片，看看我们要分类的内容。

**代码功能说明：**

- `PIL.Image.open()`：打开图片文件
- `plt.subplots()`：创建画布，把 4 张图片并排显示
- `ax.imshow()`：在子图中显示图片


```python
# 显示 4 张图片
import os

# 查找图片所在的根目录（兼容不同运行环境）
base_dir = '.'
for candidate in [os.path.join('..', '..'), '.', os.path.expanduser('~'), os.path.dirname(os.path.abspath('.'))]:
    if os.path.exists(os.path.join(candidate, 'images', 'cat1.jpg')):
        base_dir = candidate
        break
print(f"Base directory: {os.path.abspath(base_dir)}")

import matplotlib.pyplot as plt
from PIL import Image

image_files = [
    os.path.join(base_dir, 'images', 'cat1.jpg'),
    os.path.join(base_dir, 'images', 'cat2.jpg'),
    os.path.join(base_dir, 'images', 'dog1.jpg'),
    os.path.join(base_dir, 'images', 'dog2.jpg'),
]
titles = ['cat1', 'cat2', 'dog1', 'dog2']

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, f, t in zip(axes, image_files, titles):
    img = Image.open(f)
    ax.imshow(img)
    ax.set_title(t, fontsize=14)
    ax.axis('off')
plt.suptitle('Cat and Dog Images', fontsize=16)
plt.tight_layout()
plt.show()

print("4 images loaded successfully!")
```

### 第 2 步：加载预训练模型并进行分类

**代码功能说明：**

这段代码完成深度学习推理的完整流程：

1. **加载模型**：`models.resnet18(weights=...)` 加载在 ImageNet 上预训练好的 ResNet18 模型
   - ResNet18 是一个 18 层的残差网络，结构简单、运行速度快
   - `model.eval()` 设为推理模式（不计算梯度）

2. **图片预处理**：神经网络要求输入格式统一，需要 3 步变换：
   - `Resize(256)` + `CenterCrop(224)`：缩放并裁剪为 224×224 像素
   - `ToTensor()`：转为 PyTorch 张量（0~1 范围）
   - `Normalize()`：标准化（减均值、除标准差），与训练时的处理一致

3. **推理**：`model(input)` 将图片输入模型，输出 1000 个类别的得分

4. **判断猫/狗**：
   - ImageNet 中猫的类别编号为 281~285（5 种猫）
   - ImageNet 中狗的类别编号为 151~268（118 种狗）
   - 分别累加猫和狗类别的概率，比较大小即可判断

> **关键概念**：`torch.softmax()` 将模型输出转为概率（0~1，总和为 1），`torch.no_grad()` 表示推理时不记录梯度，节省内存。


```python
# 猫狗分类：加载预训练 ResNet18 并推理
import os

# 使用上一步找到的根目录
try:
    base_dir
except NameError:
    base_dir = '.'
    for candidate in [os.path.join('..', '..'), '.', os.path.expanduser('~'), os.path.dirname(os.path.abspath('.'))]:
        if os.path.exists(os.path.join(candidate, 'images', 'cat1.jpg')):
            base_dir = candidate
            break

import torch
import torchvision.models as models
import torchvision.transforms as transforms
from PIL import Image

# ---- 1. 加载预训练模型 ----
print("Loading pre-trained ResNet18 model...")
try:
    model = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
except Exception:
    model = models.resnet18(pretrained=True)  # 兼容旧版本
model.eval()
print("Model loaded!\n")

# ---- 2. 定义图片预处理 ----
transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

# ---- 3. 对每张图片进行分类 ----
image_files = [
    os.path.join(base_dir, 'images', 'cat1.jpg'),
    os.path.join(base_dir, 'images', 'cat2.jpg'),
    os.path.join(base_dir, 'images', 'dog1.jpg'),
    os.path.join(base_dir, 'images', 'dog2.jpg'),
]
image_names = ['cat1', 'cat2', 'dog1', 'dog2']

results = []
for name, f in zip(image_names, image_files):
    # 打开并预处理图片
    img = Image.open(f).convert('RGB')
    input_tensor = transform(img).unsqueeze(0)  # 增加 batch 维度

    # 模型推理
    with torch.no_grad():
        output = model(input_tensor)
    prob = torch.softmax(output, dim=1)[0]

    # 计算猫和狗的置信度
    cat_conf = prob[281:286].sum().item()   # ImageNet 猫类别: 281-285
    dog_conf = prob[151:269].sum().item()   # ImageNet 狗类别: 151-268
    prediction = 'cat' if cat_conf > dog_conf else 'dog'

    results.append({
        'name': name, 'file': f,
        'prediction': prediction,
        'cat_conf': cat_conf,
        'dog_conf': dog_conf
    })

    print(f"{name}: predicted = {prediction:3s}  |  cat_conf = {cat_conf:.4f}  |  dog_conf = {dog_conf:.4f}")

print("\nClassification complete!")
```

### 第 3 步：可视化分类结果

将每张图片的「猫置信度」和「狗置信度」画成柱状图，直观展示模型的判断。

**代码功能说明：**

- `plt.bar()`：绘制柱状图，橙色表示猫的置信度，蓝色表示狗的置信度
- 每张图片有两个柱子，哪个高就说明模型认为该图片属于哪一类
- 图表标题和标签使用**英文**，避免中文乱码问题


```python
# 绘制分类结果柱状图
import os

try:
    base_dir
except NameError:
    base_dir = '.'

import matplotlib.pyplot as plt

names = [r['name'] for r in results]
cat_confs = [r['cat_conf'] for r in results]
dog_confs = [r['dog_conf'] for r in results]
preds = [r['prediction'] for r in results]

fig, ax = plt.subplots(figsize=(10, 5))

x = range(len(names))
bar_width = 0.35

bars1 = ax.bar([i - bar_width/2 for i in x], cat_confs, bar_width,
               label='Cat Confidence', color='orange')
bars2 = ax.bar([i + bar_width/2 for i in x], dog_confs, bar_width,
               label='Dog Confidence', color='steelblue')

# 在柱子上方标注预测结果
for i, pred in enumerate(preds):
    y_pos = max(cat_confs[i], dog_confs[i]) + 0.02
    ax.text(i, y_pos, f'-> {pred}', ha='center', fontsize=12, fontweight='bold')

ax.set_xticks(list(x))
ax.set_xticklabels(names, fontsize=12)
ax.set_ylabel('Confidence', fontsize=12)
ax.set_title('Cat vs Dog Classification Results', fontsize=14)
ax.legend(fontsize=11)
ax.set_ylim(0, 1.15)

plt.tight_layout()

output_dir = os.path.join(base_dir, 'output')
os.makedirs(output_dir, exist_ok=True)
plt.savefig(os.path.join(output_dir, 'classification_result.png'), dpi=150, bbox_inches='tight')
plt.show()
print(f"Chart saved to {os.path.join(output_dir, 'classification_result.png')}")
```

### 第 4 步：在图片上标注预测结果

将每张图片连同模型的预测结果一起显示，更直观地看到分类效果。

**代码功能说明：**

- 用 plt.subplots() 创建 1 行 4 列的子图
- 每个子图显示一张原图，标题标注预测类别和置信度
- 猫的标题用橙色，狗的标题用蓝色，方便区分
- plt.savefig() 将结果保存为图片文件


```python
# 在图片上显示预测结果
import os

try:
    base_dir
except NameError:
    base_dir = '.'

import matplotlib.pyplot as plt
from PIL import Image

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
colors = {'cat': 'orange', 'dog': 'steelblue'}

for ax, r in zip(axes, results):
    img = Image.open(r['file'])
    ax.imshow(img)
    pred = r['prediction']
    conf = max(r['cat_conf'], r['dog_conf'])
    ax.set_title(f"{r['name']} -> {pred}\n(conf: {conf:.2%})",
                 fontsize=12, color=colors[pred], fontweight='bold')
    ax.axis('off')

plt.suptitle('Classification Results with Pre-trained ResNet18', fontsize=15)
plt.tight_layout()

output_dir = os.path.join(base_dir, 'output')
os.makedirs(output_dir, exist_ok=True)
plt.savefig(os.path.join(output_dir, 'prediction_summary.png'), dpi=150, bbox_inches='tight')
plt.show()
print(f"Summary saved to {os.path.join(output_dir, 'prediction_summary.png')}")
```
