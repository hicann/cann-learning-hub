# 01 智能系统介绍

深入理解智能系统的定义、意义与应用场景。

我们将从三个维度展开：**什么是智能系统**（本质定义与核心特征）、**智能系统的四大重要意义**（为什么重要）、**智能系统的应用场景**（用在哪里）。


## 1.1 什么是智能系统

### 核心定义

智能系统是指能够通过 **软硬件协同**，在 **感知、认知、决策与执行** 等环节表现出类人智能特征的综合性系统。

其本质是 **计算机科学、电子信息工程、自动控制、人工智能算法以及网络通信** 等多学科的交叉融合成果。这意味着智能系统不是单一技术的产物，而是需要硬件、软件、算法、通信等多个领域协同设计的复杂工程。

### 核心目标：赋予机器类人能力

智能系统的核心目标是 **赋予机器感知环境、分析数据、进行决策并执行动作的能力**。这四个环节构成一个完整的闭环：

## 1.2 智能系统的四大重要意义

智能系统之所以重要，是因为它正在从四个维度深刻改变我们的世界：

### ① 推动生产力升级

智能制造车间、自动化物流系统、无人驾驶汽车等应用 **显著提高生产效率和资源利用率**。通过 AI 算法优化生产流程，实现预测性维护和柔性生产。

- **典型应用**：工业 4.0、智能工厂、自动化仓储
- **实际效果**：效率提升 30%+，故障率降低 50%+，成本降低 20-40%

### ② 改善人类生活质量

从智能家居、可穿戴设备到服务机器人，智能系统 **进入千家万户**，为人们带来便捷与舒适。语音交互、健康监测、智能推荐等功能深刻改变生活方式。

- **典型应用**：智能音箱、扫地机器人、健康监测
- **用户规模**：数亿家庭，年增长率 25%+

### ③ 促进社会治理现代化

智慧城市、智能交通、公共安全监控等应用使 **社会运行更加高效与可控**。通过大数据分析和 AI 决策支持，提升城市管理水平和公共服务质量。

- **典型应用**：智慧交通、城市大脑、安防监控
- **实际效果**：通行效率提升 20%+，事故率降低 30%+

### ④ 支撑国家战略发展

人工智能与智能系统技术已成为 **全球科技竞争的重要高地**，与国防工业、国家安全、战略性新兴产业紧密相关。自主可控的智能系统技术是国家核心竞争力的重要组成部分。

- **典型应用**：国防工业、航天探索、关键基础设施
- **战略地位**：核心支撑，技术领先

> 💡 **总结**：智能系统不仅是技术发展的前沿领域，更是推动社会进步和产业升级的核心动力。这正是我们学习昇腾芯片和 CANN 的根本原因——它们是构建智能系统的国产算力底座。

### 核心特征

无论处于哪个层级，智能系统都具备以下核心特征：

- **软硬件协同**：硬件平台与软件算法深度集成
- **感知能力**：通过传感器获取环境信息
- **认知推理**：运用 AI 算法进行智能分析
- **决策执行**：基于分析结果采取行动

> 💡 **演进趋势**：随着硬件平台算力提升、通信网络演进以及 AI 算法的不断成熟，未来的智能系统将更加普及，并渗透到人类生活的方方面面。本课程关注的昇腾平台，主要服务于 L3（边缘计算）和 L4（自主无人系统）层级。

#### Transformer 为什么重要？

Transformer 的核心创新是 **自注意力机制（Self-Attention）**：

$$\text{Attention}(Q, K, V) = \text{softmax}\!\left(\frac{QK^\top}{\sqrt{d_k}}\right) V$$

它让每个位置都能直接"看到"序列中所有其他位置的信息，从而：
- **全局感受野**：一步即可关联任意距离的token，不存在RNN的长程遗忘问题
- **高度并行**：所有位置同时计算，充分利用GPU/NPU的并行算力
- **可扩展性**：堆叠更多层、更多参数就能持续提升效果（Scaling Law）

Transformer 已成为当今大模型的基础架构：GPT、BERT、Qwen、DeepSeek、LLaMA 等全部基于 Transformer。

> 💡 **架构选择**：图像任务选CNN，序列任务选RNN/Transformer，复杂任务可组合使用。

## 3.3 Host 与 Device 协作：矩阵乘法

<img src="../../images/host_and_device.png" alt="Host和Device" style="display: block; margin-left: 0;" />

昇腾 NPU 采用 **Host/Device 异构计算** 模型：Host（CPU）负责发任务、传数据、收结果；Device（NPU）负责拼命算、算完交差。下面用一个矩阵乘法完整体验这个协作流程。

数据可以直接在 NPU 上创建（推荐），也可以先在 CPU 创建再搬运到 NPU（适用于加载已有数据）。

````python
N = 1024

# ---- 方式一：数据直接在 NPU 上创建（推荐，省去搬运开销） ----
A_npu = torch.randn(N, N, device='npu:0')
B_npu = torch.randn(N, N, device='npu:0')
print(f'[Device] 直接在 NPU 上创建矩阵，A.device={A_npu.device}')

C_npu = torch.mm(A_npu, B_npu)
print(f'[Device] NPU 完成矩阵乘法，结果形状: {tuple(C_npu.shape)}')

# ---- 方式二：数据先在 CPU 创建，再搬运到 NPU（适用于加载已有数据/权重） ----
A_cpu = torch.randn(N, N)
A_on_npu = A_cpu.npu()  # Host -> Device 搬运
print(f'[Host -> Device] 数据搬运到 NPU，A.device={A_on_npu.device}')

# ---- Device -> Host：结果搬回 CPU ----
C = C_npu.cpu()
print(f'[Device -> Host] 结果搬回 CPU，C.device={C.device}')
````

## 3.4 NPU 加速体验：CPU vs NPU 性能对比

3×3 的小张量无法体现 NPU 的优势——数据量太小，计算时间远小于数据搬运时间。让我们用更大的张量（10000×10000，约 400MB）来感受 NPU 的并行计算能力。

注意：NPU 执行是异步的，需要调用 `torch.npu.synchronize()` 等待计算完成才能准确计时。这和 GPU 一样——`torch.cuda.synchronize()` 用于 GPU，`torch.npu.synchronize()` 用于 NPU。

````python
import time

N = 10000
a_large = torch.randn(N, N, dtype=torch.float32)
b_large = torch.randn(N, N, dtype=torch.float32)
print(f"张量大小: {N}x{N}，约 {a_large.numel() * 4 / 1024**2:.0f} MB")

# CPU 计时
start = time.time()
c_cpu_large = a_large + b_large
cpu_time = time.time() - start
print(f"CPU 加法耗时: {cpu_time*1000:.2f} ms")

# NPU 计时
a_npu_large = a_large.npu()
b_npu_large = b_large.npu()
torch.npu.synchronize()
start = time.time()
c_npu_large = a_npu_large + b_npu_large
torch.npu.synchronize()
npu_time = time.time() - start
print(f"NPU 加法耗时: {npu_time*1000:.2f} ms")

# 验证与对比
match = torch.allclose(c_npu_large.cpu(), c_cpu_large, rtol=1e-5)
print(f"\n{'='*50}")
print(f"NPU 与 CPU 结果一致: {match}")
print(f"CPU: {cpu_time*1000:.2f} ms | NPU: {npu_time*1000:.2f} ms")
if npu_time > 0:
    print(f"加速比: {cpu_time/npu_time:.1f}x")
````

### 为什么 NPU 更快？

- **计算单元多**：CPU 只有少量通用 ALU，NPU 有数十~数百个 Vector + Cube 单元
- **并行度高**：CPU 串行或少量并行，NPU 大规模数据并行
- **低精度优化**：CPU 以 FP64 为主，NPU 针对 FP16/BF16 优化

### 什么时候 NPU 有优势？

- **计算密集型任务**：矩阵乘法、卷积、大规模逐元素运算
- **批量并行**：同一操作对大量数据重复执行
- **低精度可接受**：FP16/BF16/FP8 甚至是 Int8/FP4 可满足需求

### 什么时候 CPU 更合适？

- **小数据量**：数据搬运开销 > 计算节省
- **复杂控制流**：大量条件分支、动态 shape
- **高精度需求**：必须 FP64


## 3.5 在 NPU 上执行一个微型神经网络

把一个简单的全连接网络搬到 NPU 上，感受"模型迁移只需一行 `.npu()`"的零门槛体验。这是 PyTorch + torch_npu 的核心优势——几乎不改代码就能从 CPU 迁移到 NPU。

````python
import torch.nn as nn

# 定义一个简单的多层感知机
class SimpleMLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(128, 64)
        self.fc2 = nn.Linear(64, 10)
    def forward(self, x):
        return self.fc2(torch.relu(self.fc1(x)))

model = SimpleMLP()
x = torch.randn(32, 128)  # batch=32, 特征=128

# CPU 推理
out_cpu = model(x)
print(f"CPU 推理输出形状: {tuple(out_cpu.shape)}, 设备: {out_cpu.device}")

# 迁移到 NPU：模型和数据各加一个 .npu() 即可
model_npu = model.npu()
x_npu = x.npu()
out_npu = model_npu(x_npu)
print(f"NPU 推理输出形状: {tuple(out_npu.shape)}, 设备: {out_npu.device}")
print(f"结果一致: {torch.allclose(out_npu.cpu(), out_cpu, atol=1e-5)}")
````

运行上面的代码后，你可以看到：**只需 `.npu()` 一行代码，模型和数据就从 CPU 走到了 NPU**，其余代码完全不变——这就是昇腾平台"零门槛迁移"的设计理念。

回顾关键步骤：

```python
import torch
import torch_npu                    # 1. 导入 NPU 适配层

a = torch.ones(3, 3).npu()        # 2. 创建张量并搬到 NPU
b = torch.full((3,3), 2.0).npu()  # 3. 同上
c = a + b                          # 4. NPU 自动执行加法
print(c.cpu())                     # 5. 搬回 CPU 查看结果
```

**5 行代码，从 CPU 走到 NPU！**

> 在底层，PyTorch + torch_npu 会自动调用 CANN 内置的高性能算子，由 AI Core 的 Cube/Vector 单元执行计算，全程无需手写硬件代码。

## 5.7 边缘AI与端云协同架构

### 边缘AI

边缘AI将人工智能模型部署在边缘设备（如手机、传感器、嵌入式设备）上，**实现本地实时处理**，减少云端传输延迟，保护数据隐私。

**案例1：智能家居安防**
- 边缘端：轻量级YOLOv5模型实时检测入侵者
- 云端：上传可疑图像进行高精度分析
- 优势：实时响应、隐私保护、节省带宽

### 端云协同

端云协同通过边缘与云计算的协作，平衡计算负载、隐私与延迟。边缘处理实时任务，云端处理复杂分析和模型训练。
