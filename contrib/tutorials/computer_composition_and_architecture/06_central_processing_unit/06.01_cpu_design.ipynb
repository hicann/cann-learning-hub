# 实验8：单周期/多周期CPU设计


## 一、实验说明

### 1.1 实验背景

CPU（中央处理器）是计算机系统的核心部件，其设计是计算机组成原理课程的重要内容。单周期CPU和多周期CPU是两种经典的处理器架构实现方式，它们体现了指令执行过程中控制与数据通路的两种不同设计哲学。

单周期CPU是指每条指令在一个时钟周期内完成取指、译码、执行、访存和写回全部五个阶段。其特点是控制逻辑简单、设计直观，但时钟周期必须足够长以保证最慢的指令（如lw访存指令）也能完成，因此整体性能受到限制。

多周期CPU则将一条指令的执行拆分为多个相对固定的时钟阶段（状态），每个阶段在一个时钟周期内完成。这样时钟周期可以大幅缩短，不同指令可以占用不同数量的周期，实现了资源的按需使用和共享。

本实验基于Logisim仿真设计环境，引导学生从功能部件出发，逐步构建完整的单周期/多周期CPU数据通路，并通过深入分析指令执行过程来设计和实现控制器逻辑。

## 二、实验任务

### 2.1 任务描述

本实验要求学生基于Logisim平台，设计并实现一个支持MIPS指令子集的单周期CPU和一个多周期CPU，并通过功能验证测试确保设计正确性。具体包括：

1. **单周期CPU设计**：采用硬布线控制器，在一个时钟周期内完成指令的全部执行流程
2. **多周期CPU设计**：采用状态机控制的硬布线控制器或微程序控制器，将指令执行划分为多个时钟周期
3. **功能验证**：通过测试程序（如冒泡排序）验证CPU设计的正确性

### 2.2 学习目标

通过本实验，学生应达到以下学习目标：

1. 掌握MIPS指令集格式、寻址方式及指令编码规则
2. 理解单周期与多周期CPU的架构差异与设计思想
3. 掌握CPU数据通路的构建方法和各功能部件的作用
4. 掌握硬布线控制器和状态机/微程序控制器的设计原理
5. 具备利用Logisim进行数字电路设计和仿真的能力
6. 培养CPU架构设计、调试及系统验证的能力


## 三、任务准备

### 3.1 前置知识

在进行本实验之前，学生需要掌握以下基础知识：

**（1）MIPS指令集基础**

MIPS指令集规整简洁，分为R型、I型、J型三种指令格式。本实验通常要求支持的核心指令包括：

| 指令类型 | 指令 | 功能描述 |
|---------|------|---------|
| R型 | add | 加法：GPR[rd] ← GPR[rs] + GPR[rt] |
| R型 | sub | 减法：GPR[rd] ← GPR[rs] - GPR[rt] |
| R型 | and | 按位与 |
| R型 | or | 按位或 |
| I型 | lw | 取数：GPR[rt] ← Memory[GPR[base] + offset] |
| I型 | sw | 存数：Memory[GPR[base] + offset] ← GPR[rt] |
| I型 | beq | 相等分支：if GPR[rs] == GPR[rt] then PC ← PC + 4 + (offset << 2) |
| J型 | j | 无条件跳转：PC ← (PC & 0xF0000000) | (address << 2) |

**（2）CPU基本组成**

CPU由数据通路和控制器两大部分组成：
- **数据通路**：包括程序计数器（PC）、指令存储器、寄存器堆（Register File）、ALU、数据存储器、符号扩展器、多路选择器（MUX）等
- **控制器**：负责产生控制信号，驱动数据通路中各组件的使能端、读写端等

**（3）Logisim基本操作**

熟悉Logisim的界面布局、组件工具箱的使用、子电路的创建与调用、时钟仿真等基本操作。

## 五、任务拓展

### 5.1 指令集扩展

在完成基本指令集（8条）的基础上，可进一步扩展支持更多指令：
- **算术指令**：addi、slt、slti等
- **逻辑指令**：andi、ori、xori、lui等
- **分支指令**：bne、blez、bgtz等
- **跳转指令**：jal（带链接跳转）、jr（寄存器跳转）

### 5.2 流水线CPU设计

在掌握单周期和多周期CPU设计的基础上，可进一步学习五级流水线CPU的设计：
- 将指令执行分为IF、ID、EXE、MEM、WB五个流水级
- 处理流水线冒险（数据冒险、控制冒险、结构冒险）
- 实现转发（Forwarding）和分支预测等优化技术

### 5.3 中断与异常处理

为CPU添加中断和异常处理机制，支持外部设备中断和内部异常（如溢出、非法指令等）。

### 5.4 Verilog硬件描述语言实现

将Logisim中的电路设计用Verilog硬件描述语言重新实现，并在FPGA开发板上进行验证。

## 六、实验总结

本实验通过Logisim仿真平台，引导学生完成了从单周期CPU到多周期CPU的设计与实现全过程。通过本实验，学生应能够：

1. **深入理解CPU工作原理**：通过亲手搭建数据通路和控制器，直观理解指令在CPU中的完整执行流程

2. **掌握两种CPU架构的设计方法**：单周期CPU控制逻辑简单但性能受限；多周期CPU通过状态机实现资源复用和时钟频率提升

3. **建立系统设计思维**：从功能部件到数据通路再到控制器，采用模块化、自底向上的设计方法

4. **培养调试与验证能力**：通过测试程序验证设计正确性，学会利用仿真工具进行系统调试

单周期与多周期CPU的核心区别可总结如下：

| 特性维度 | 单周期CPU | 多周期CPU |
|---------|----------|----------|
| 每指令周期数 | 1个周期 | 3-5个周期（因指令而异） |
| 时钟频率 | 较低（受最慢指令限制） | 较高 |
| 硬件资源 | 指令/数据存储器分离 | 可复用同一存储器 |
| 控制器类型 | 硬布线（组合逻辑） | 硬布线+状态机 或 微程序 |
| 控制信号 | 执行前一次性确定 | 随状态变化 |
| 设计复杂度 | 较低 | 较高 |

通过本实验的学习和实践，学生将为后续计算机系统设计、处理器架构等更深入的学习奠定坚实基础。
