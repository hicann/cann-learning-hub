# 实验1：通用运算器设计


## 一、实验目的

本实验基于Logisim仿真设计环境，设计一个通用加、减、乘、除法功能运算器（ALU）。通过本实验，学生应达到以下目标：

1. **理解算术逻辑运算单元（ALU）的基本构成**，掌握其内部各功能模块的工作原理与相互联系。

2. **掌握补码运算及其实现方法**，深刻理解补码表示将加法和减法统一为加法运算的核心思想，具备有符号数运算电路的设计能力。

3. **熟悉Logisim中各种运算组件的使用方法**，包括加法器、减法器、乘法器、除法器、多路选择器等。

4. **培养数字电路设计与仿真验证能力**，能够独立完成从原理分析、电路设计到功能测试的完整实验流程。



## 二、实验说明

### 2.1 实验背景

运算器是计算机CPU的核心部件之一，负责执行所有的算术运算和逻辑运算。在计算机组成原理课程中，ALU的设计是理解和掌握计算机数据通路的关键环节。本实验以Logisim为仿真平台，从数字逻辑层面构建一个完整的通用运算器，使学生能够直观地理解运算器内部各部件的工作机制。

在计算机系统中，有符号数通常采用补码表示。补码表示法的优势在于**减法可以转化为加法运算**（减去一个数等于加上这个数的补码），从而简化了硬件电路的设计。理解并实现基于补码的加减乘除运算，是计算机组成原理教学中的核心内容。

## 三、实验任务

### 3.1 任务描述

利用Logisim平台，设计并实现一个支持**加法、减法、乘法、除法**四种算术运算的通用运算器。具体要求如下：

1. 设计并封装**32位（或8位/16位）加法器**，作为运算器的核心运算部件。
2. 基于补码原理实现**可控加减法电路**，通过控制信号切换加/减运算。
3. 设计**乘法器**和**除法器**模块，支持有符号数的乘除运算。
4. 通过**多路选择器（MUX）** 将各运算单元的结果汇总输出。
5. 实现**程序状态标志**输出，包括有符号溢出标志（OF）、无符号溢出标志（UOF）、零标志（ZF）、符号标志（SF）等。

## 四、任务准备

### 4.1 前置知识

#### 4.1.1 补码表示法

补码是计算机中有符号数的标准表示方式。当真值为正数时，补码符号位为0，数据位与真值相同；当真值为负数时，补码符号位为1，数据位部分通过对真值取反加一得到。

补码系统的核心优势在于：**加法和减法可以统一为加法运算**——减去一个数等于加上这个数的补码。这极大地简化了硬件设计。

#### 4.1.2 全加器与可控加减法器

- **一位全加器**：实现两个1位二进制数及低位进位相加，产生和与进位输出。
- **多位串行加法器**：通过级联多个全加器实现多位二进制数的加法运算。
- **可控加减法器**：在加法器基础上增加控制端Sub，当Sub=0时做加法，Sub=1时做减法（将减数取反加一）。

#### 4.1.3 乘法与除法原理

- **阵列乘法器**：通过移位相加的方式实现二进制乘法。
- **补码一位乘法**：在补码体系下，通过控制电路和数据通路自动完成补码乘法运算。
- **阵列除法器**：通过移位相减的方式实现二进制除法。

#### 4.1.4 溢出判断

- **无符号数溢出（UOF）** ：加法时最高位产生进位。
- **有符号数溢出（OF）** ：最高位进位与次高位进位异或。

## 六、实验总结

### 6.1 实验收获

通过本次实验，深入理解了ALU的内部结构和工作原理。补码表示法将减法转化为加法的设计思想极大地简化了硬件实现，这一核心概念在实验中得到了充分验证。从全加器到可控加减法器，再到完整的ALU集成，逐步构建的过程让我对数字电路设计的层次化方法有了切身体会。

### 6.2 设计体会

1. **模块化设计的重要性**：将复杂电路分解为全加器、加法器、乘法器等独立模块，便于设计、调试和复用。

2. **补码运算的优雅性**：补码系统使得加法和减法可以共享同一套硬件，这是计算机体系结构中最巧妙的设计之一。

3. **测试验证的必要性**：边界条件和异常情况的测试往往能发现隐藏的设计缺陷。

### 6.3 展望

本次实验构建的通用运算器是CPU数据通路的核心部件。未来可在此基础上进一步扩展，与寄存器堆、控制器等部件集成，构成完整的处理器原型。通过Logisim这一可视化仿真工具，计算机组成原理中抽象的概念得以直观呈现，为后续深入学习计算机体系结构奠定了坚实基础。
