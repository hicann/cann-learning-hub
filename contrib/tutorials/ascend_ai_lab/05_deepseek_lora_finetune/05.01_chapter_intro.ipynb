# 05.01 章节概述：DeepSeek-LLM-7B LoRA 微调

## 学习前置要求

学习本章节前，开发者需要已经具备：

- Python 与 PyTorch 基础：张量、nn.Module、DataLoader
- Transformer 与注意力机制：Q/K/V 投影、注意力分数、Softmax
- 大语言模型基础：Token、Prompt、Temperature、top_p
- Ascend NPU / CANN 基础：了解 torch_npu 与 SDPA 的作用

## 章节目标

完成本章节后，你应该能够：

1. 掌握 CANNLab 环境配置与 Ascend NPU 模型加载
2. 解决 DeepSeek 分词器兼容性问题
3. 完成多轮对话数据格式化与 9:1 切分
4. 配置 LoRA 并使用 SFTTrainer 完成微调训练
5. 使用 SwanLab 监控训练，完成单轮与多轮对话测试
