# 02.01 章节概述：MobileNetV3 图像分类实战

## 学习前置要求

学习本章节前，开发者需要已经具备：

- Python 基础：变量、函数、类、模块导入
- PyTorch 基础：张量操作、nn.Module、自动求导、DataLoader
- 深度学习基础：CNN 的卷积、池化、激活函数、反向传播
- Ascend NPU / CANN 基础：了解 torch_npu 的作用

## 章节目标

完成本章节后，你应该能够：

1. 配置 PyTorch + torch_npu 环境并检查 NPU 可用性
2. 完成 Tiny ImageNet 数据集下载、验证集整理与数据增强
3. 使用 PyTorch 从零搭建 MobileNetV3 模型
4. 配置余弦退火、标签平滑、EMA 等训练技巧并完成训练
5. 计算 Top-1 / Top-5 准确率，完成单图推理
6. 完成 GPU 到 Ascend NPU 的迁移与性能对比
