### 问题12（简答题）

**题目：** 为什么本实验先跑通单卡流程，再把多卡 DDP 放到扩展部分？

**解答：** 单卡流程能验证数据、模型、训练脚本、NPU 环境和日志输出是否闭环；多卡 DDP 会额外引入 rank、world_size、HCCL 通信和多进程同步，变量更多。先稳定单卡可以降低排错难度。

**解析：** 这是工程实验常见的递进方式：先保证最小闭环，再扩展并行能力。


### 问题13（简答题）

**题目：** 实验报告中描述 NPU 训练是否成功时，应至少包含哪些证据？

**解答：** 应包含环境检查结果、训练启动命令、关键日志、loss 是否正常变化、throughput 或耗时、checkpoint 是否生成，以及必要的截图或输出片段。

**解析：** 这些证据能说明实验不是只运行了命令，而是完成了可复现的训练过程。


### 问题14（简答题）

**题目：** 如果 notebook 中命令在终端能跑、但在 Jupyter 中失败，排查思路是什么？

**解答：** 先确认 Jupyter kernel 使用的 Python 是否一致，再检查环境变量是否在单元格中重新 source/export，随后确认当前工作目录和相对路径是否正确，最后查看完整错误日志。

**解析：** Jupyter 单元格与交互式终端不是完全同一个 shell 状态，经常需要显式设置环境和路径。


### 问题15（代码设计题）

**题目：** 写一段最小 Python 检查代码，打印 PyTorch、torch_npu 是否可导入，并输出 NPU 是否可用。

**解答：** ```python
import torch
print("torch:", torch.__version__)
try:
    import torch_npu
    print("torch_npu: imported")
    print("npu available:", torch.npu.is_available())
except Exception as exc:
    print("torch_npu import failed:", exc)
```

**解析：** 这段代码能快速判断 Python 训练环境是否具备调用 Ascend NPU 的基础能力。
