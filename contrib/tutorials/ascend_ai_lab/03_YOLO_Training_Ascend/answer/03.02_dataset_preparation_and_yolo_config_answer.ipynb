# 课后练习解答（03.02_dataset_preparation_and_yolo_config）

本解答对应《数据集准备与 YOLO 配置》课后练习，共 15 题。


### 问题1（单选题）

**题目：** PASCAL VOC 原始目标检测标注主要采用哪种格式？

A. JSON
B. XML
C. CSV
D. NPZ

**解答：** B

**解析：** VOC 原始标注通常位于 Annotations 目录，单张图片对应一个 XML 文件。


### 问题2（单选题）

**题目：** YOLO txt 每一行通常表示什么？

A. 一张图片
B. 一个目标框
C. 一个 batch
D. 一个模型权重

**解答：** B

**解析：** YOLO txt 每行一般表示一个目标，包含类别和归一化后的框坐标。


### 问题3（单选题）

**题目：** 把 VOC XML 转为 YOLO txt 的核心原因是？

A. 减小图片尺寸
B. 让训练脚本读取统一的 YOLO 标注格式
C. 提高 Git 上传速度
D. 替代模型结构

**解答：** B

**解析：** YOLO 训练代码通常期望读取 class_id、x_center、y_center、width、height 这种格式。


### 问题4（单选题）

**题目：** `prepare_voc_yolo.py` 的输出不包括哪一项？

A. images/train2007、images/train2012、images/val2007
B. labels/.../*.txt
C. splits/train.txt 与 splits/val.txt
D. CANN .om 模型

**解答：** D

**解析：** 数据准备脚本负责数据组织和标注转换，不负责 ATC 模型转换。


### 问题5（多选题）

**题目：** 关于 VOC2007 与 VOC2012，下列说法正确的是？

A. 它们是 PASCAL VOC 系列中的两个年份版本
B. 训练时可以组合 VOC2007+VOC2012 的 trainval 数据
C. VOC2007 test 常用于评估
D. VOC2012 天然包含 VOC2007 的所有图片

**解答：** A、B、C

**解析：** VOC2007 和 VOC2012 是两个年份版本；常见训练配置会组合 trainval，但二者不是简单包含关系。


### 问题6（多选题）

**题目：** YOLO txt 中的坐标为什么要归一化到 0 到 1？

A. 便于适配不同图片尺寸
B. 避免模型只适用于固定原图宽高
C. 让标注和 resize 后的输入更容易统一处理
D. 为了让类别数变少

**解答：** A、B、C

**解析：** 归一化坐标表达的是相对位置和大小，与类别数无关。


### 问题7（多选题）

**题目：** 数据准备后，应重点检查哪些文件或目录？

A. images/ 是否有图片
B. labels/ 是否有同名 txt 标注
C. splits/train.txt 和 splits/val.txt 是否存在
D. 模型 checkpoint 是否一定已经生成

**解答：** A、B、C

**解析：** checkpoint 是训练后产物，数据准备阶段不一定存在。


### 问题8（判断题）

**题目：** XML 转 YOLO txt 属于训练前的数据预处理步骤。

**解答：** 正确

**解析：** 它把原始标注转换成训练脚本可读取的格式，是典型的数据预处理。


### 问题9（判断题）

**题目：** YOLO txt 中 width 和 height 表示像素宽高，通常不需要除以图片宽高。

**解答：** 错误

**解析：** 标准 YOLO txt 中坐标和宽高通常都是归一化比例值。


### 问题10（填空题）

**题目：** VOC XML 中的 `xmin,ymin,xmax,ymax` 会被转换为 YOLO txt 中的 `____, ____, ____, ____`。

**解答：** x_center, y_center, width, height

**解析：** YOLO 使用中心点加宽高的框表示方式。


### 问题11（填空题）

**题目：** 类别名称列表通常写入 `____`，训练配置中的 `num_classes` 应与类别数量保持一致。

**解答：** voc.names

**解析：** VOC 有 20 个目标类别，类别列表和模型类别数必须对应。


### 问题12（简答题）

**题目：** 为什么训练配置中的数据路径必须和实际生成目录保持一致？

**解答：** 训练脚本会根据配置文件定位图片、标签和 split 列表。如果路径不一致，轻则读不到样本，重则使用错误数据训练，导致实验结果不可复现。

**解析：** 路径配置是数据流进入训练脚本的入口。


### 问题13（简答题）

**题目：** 如果转换后某些图片没有生成对应 txt，可能有哪些原因？

**解答：** 可能是图片没有目标框、类别不在 VOC_CLASSES 中、XML 文件缺失或解析失败，也可能是数据划分列表没有包含该图片。

**解析：** 需要结合 XML 内容、类别表和 split 列表一起检查。


### 问题14（简答题）

**题目：** 为什么数据准备阶段不建议直接修改原始 VOCdevkit 目录？

**解答：** 原始目录应作为数据来源保留，转换后的 YOLO 格式数据应输出到实验工作目录。这样可以避免污染原始数据，也方便重新生成、排错和复现实验。

**解析：** 保留原始数据和派生数据的边界，是实验工程化的好习惯。


### 问题15（代码设计题）

**题目：** 写一段 Python 代码，检查 `splits/train.txt` 中前 5 个样本是否都有图片和标签文件。

**解答：** ```python
from pathlib import Path
root = Path("/mnt/workspace/datasets/voc_yolo")
items = (root / "splits/train.txt").read_text().splitlines()[:5]
for item in items:
    img = root / f"images/{item}.jpg"
    lab = root / f"labels/{item}.txt"
    print(item, "image:", img.exists(), "label:", lab.exists())
```

**解析：** 通过同时检查图片和标签，可以快速发现转换或路径配置问题。
