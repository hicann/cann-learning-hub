# 课后练习解答（03.07_msprof_profiling_and_chapter_test）

本解答对应《MSPROF 性能 Profiling 与实验验收》课后练习，共 15 题。


### 问题1（单选题）

**题目：** MSPROF 在本实验中的主要用途是？

A. 采集训练过程性能数据并辅助定位瓶颈
B. 转换 XML 标注
C. 生成 VOC 数据集
D. 替代优化器

**解答：** A

**解析：** MSPROF 是 Ascend 生态中的性能采集和分析工具。


### 问题2（单选题）

**题目：** Profiling 结果中，`Task Duration(us)` 更接近表示什么？

A. 算子或任务耗时
B. 图片数量
C. 类别编号
D. Git 提交数量

**解答：** A

**解析：** Task Duration(us) 通常用于观察算子/任务耗时。


### 问题3（单选题）

**题目：** 如果 runner 端到端耗时远大于单个算子耗时，最合理的解释是？

A. 端到端时间包含数据生成、编译、初始化、文件 I/O 等开销
B. NPU 一定比 CPU 慢
C. 模型没有类别
D. VOC XML 无法读取

**解答：** A

**解析：** 端到端耗时和纯 kernel/算子耗时不是同一个概念。


### 问题4（单选题）

**题目：** 生成 Profiling 数据后，实验报告中应优先记录什么？

A. 数据目录、关键 csv、目标算子耗时和结论
B. 屏幕亮度
C. 鼠标型号
D. 浏览器缩放比例

**解答：** A

**解析：** 报告需要说明数据来源和分析依据。


### 问题5（多选题）

**题目：** MSPROF 采集后可能关注哪些文件或信息？

A. op_summary_*.csv
B. task_time_*.csv
C. api_statistic_*.csv
D. mindstudio_profiler_output 目录

**解答：** A、B、C、D

**解析：** 这些都是常见的 Profiling 输出或摘要文件。


### 问题6（多选题）

**题目：** 分析训练瓶颈时，应该同时考虑哪些方面？

A. Host 侧数据加载
B. NPU 计算耗时
C. 内存拷贝和同步
D. 日志和 checkpoint I/O

**解答：** A、B、C、D

**解析：** 性能瓶颈可能出现在端到端链路的多个位置。


### 问题7（多选题）

**题目：** 一个严谨的性能结论应避免哪些说法？

A. 只凭一次运行下结论
B. 混淆端到端耗时和算子耗时
C. 只报最快值不说明环境
D. 说明采集命令和数据位置

**解答：** A、B、C

**解析：** D 是推荐做法，不是应避免的说法。


### 问题8（判断题）

**题目：** Profiling 的目标是证明程序一定最快，而不是定位瓶颈。

**解答：** 错误

**解析：** Profiling 的核心价值是量化耗时并定位瓶颈。


### 问题9（判断题）

**题目：** 同一实验中 CPU baseline、训练日志和 MSPROF 数据可以互相补充。

**解答：** 正确

**解析：** CPU baseline 提供参考，训练日志反映端到端过程，MSPROF 细化到任务/算子层面。


### 问题10（填空题）

**题目：** MSPROF 生成的数据目录常以 `____` 开头。

**解答：** PROF_

**解析：** Ascend Profiling 输出目录通常包含 PROF_ 前缀和时间戳。


### 问题11（填空题）

**题目：** 分析算子耗时时，常在 `op_summary_*.csv` 中查找目标算子的 `____` 字段。

**解答：** Task Duration(us)

**解析：** 该字段用于观察任务耗时，单位通常是微秒。


### 问题12（简答题）

**题目：** 为什么性能分析要区分“端到端耗时”和“算子耗时”？

**解答：** 端到端耗时包含脚本启动、数据加载、编译、ACL 初始化、文件 I/O 和同步等开销；算子耗时只关注 NPU 上某个任务的执行时间。混淆二者会导致错误结论。

**解析：** 这也是前面自定义算子实验中反复强调的点。


### 问题13（简答题）

**题目：** 如果 Profiling 显示 NPU 计算很快但整体训练慢，下一步应该看什么？

**解答：** 应检查 DataLoader workers、数据预处理耗时、Host 到 Device 拷贝、同步点、日志打印和 checkpoint I/O。

**解析：** 这类情况通常说明瓶颈不在 NPU kernel 本身。


### 问题14（简答题）

**题目：** 实验验收时如何描述 MSPROF 的分析结果更严谨？

**解答：** 写清采集命令、输出目录、分析的 csv 文件、目标字段、关键耗时数值，并解释这些数值支持什么结论，同时说明端到端开销是否被包含。

**解析：** 这样可以避免只贴截图但没有可复核依据。


### 问题15（代码设计题）

**题目：** 写一条示例命令，用 MSPROF 采集 2 个 epoch 的单卡训练性能数据。

**解答：** ```bash
msprof --application="python src/scripts/train_yolo_single_npu_amp.py --config src/configs/yolo_ascend.yaml --epochs 2 --profile"   --output=profiles/yolo_single_npu
```

**解析：** 实际参数可按 CANN/MSPROF 版本调整，核心是用 --application 指定训练命令，用 --output 指定采集目录。
