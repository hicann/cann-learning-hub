# 课后练习解答（03.03_single_npu_launch_and_ddp_extension）

本解答对应《单卡 NPU 启动与 DDP 扩展》课后练习，共 15 题。


### 问题1（单选题）

**题目：** 本实验默认推荐首先运行哪个训练方式？

A. 单卡 NPU 训练
B. 8 卡 DDP 训练
C. CPU-only 训练
D. 只运行数据转换

**解答：** A

**解析：** 当前实验室环境默认以单卡流程为主，多卡 DDP 作为扩展。


### 问题2（单选题）

**题目：** `launch_1npu.sh` 的核心作用是？

A. 删除数据集
B. 设置环境并启动单卡训练脚本
C. 把 XML 转成 JSON
D. 上传 Git LFS 文件

**解答：** B

**解析：** 启动脚本负责准备运行环境并调用训练入口。


### 问题3（单选题）

**题目：** DDP 中 `WORLD_SIZE` 通常表示什么？

A. 图片宽度
B. 总进程数或总设备数
C. 类别数量
D. 学习率

**解答：** B

**解析：** 分布式训练中 WORLD_SIZE 表示参与训练的进程总数。


### 问题4（单选题）

**题目：** 单卡训练中最常见的本地 rank 是？

A. 0
B. 1
C. 8
D. 不固定且必须大于 1

**解答：** A

**解析：** 单卡只有一个训练进程，local rank 通常为 0。


### 问题5（多选题）

**题目：** 多卡 DDP 扩展时，需要额外关注哪些内容？

A. rank/local_rank/world_size
B. HCCL 通信配置
C. DistributedSampler
D. 每卡 batch size 与总 batch size 的关系

**解答：** A、B、C、D

**解析：** 这些都是多卡训练从单卡扩展时的关键点。


### 问题6（多选题）

**题目：** 单卡启动失败时，可以优先检查哪些问题？

A. CANN 环境变量是否 source
B. torch_npu 是否可导入
C. 数据路径是否存在
D. PR 标题是否填写

**解答：** A、B、C

**解析：** PR 标题与本地训练启动无关。


### 问题7（多选题）

**题目：** 下列哪些环境变量或概念和 Ascend 分布式训练关系更密切？

A. RANK
B. WORLD_SIZE
C. LOCAL_RANK
D. HCCL

**解答：** A、B、C、D

**解析：** DDP/HCCL 训练需要这些进程编号和通信信息。


### 问题8（判断题）

**题目：** 单卡训练跑通后，就可以完全忽略多卡 DDP 的概念。

**解答：** 错误

**解析：** 单卡是主流程，但理解 DDP 的扩展入口有助于后续迁移到多卡环境。


### 问题9（判断题）

**题目：** 在 notebook 中运行 shell 单元格时，每个单元格都应显式设置关键环境变量更稳妥。

**解答：** 正确

**解析：** Jupyter 的单元格状态容易和终端不一致，显式设置能减少偶发错误。


### 问题10（填空题）

**题目：** Ascend 多卡通信通常使用 `____` 后端。

**解答：** HCCL

**解析：** HCCL 是 Ascend 分布式通信的常用后端。


### 问题11（填空题）

**题目：** 单卡训练启动脚本一般会把模型放到 `____` 设备上运行。

**解答：** npu:0

**解析：** 单卡默认使用第 0 张 NPU。


### 问题12（简答题）

**题目：** 为什么单卡脚本和 DDP 脚本不应该混在一个不可区分的命令里？

**解答：** 单卡和多卡在进程数、环境变量、采样器、通信初始化和日志汇总上不同。清晰区分启动入口，有利于教学、复现和排错。

**解析：** 实验文档强调默认单卡，多卡作为扩展，可以降低学习成本。


### 问题13（简答题）

**题目：** DDP 中为什么常常只让 rank 0 保存 checkpoint？

**解答：** 多卡训练每个进程都有模型副本，如果所有 rank 同时写同一个文件，容易发生冲突或重复写入。通常只让 rank 0 负责日志和 checkpoint。

**解析：** 这是分布式训练中常见的 I/O 管理策略。


### 问题14（简答题）

**题目：** 从单卡扩展到多卡时，学习率和 batch size 为什么需要重新考虑？

**解答：** 多卡会改变全局 batch size，梯度统计和优化步长随之变化。如果直接沿用单卡超参，可能导致收敛速度或稳定性变化。

**解析：** 多卡不是简单复制进程，还会改变训练动力学。


### 问题15（代码设计题）

**题目：** 写一段 shell 命令，展示单卡训练前如何设置 CANN 环境并启动 `launch_1npu.sh`。

**解答：** ```bash
source /usr/local/Ascend/ascend-toolkit/latest/aarch64-linux/script/set_env.sh
export ASCEND_GLOBAL_LOG_LEVEL=3
bash launch_1npu.sh
```

**解析：** 真实路径可按环境调整，核心是先 source CANN 环境，再运行单卡启动脚本。
