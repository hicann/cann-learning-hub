# 课后练习解答（03.06_single_npu_training_practice）

本解答对应《单卡 NPU 训练实战》课后练习，共 15 题。


### 问题1（单选题）

**题目：** 单卡训练实践中，启动训练前最不应该忽略的是？

A. 数据路径和环境变量
B. 浏览器主题颜色
C. PPT 字体
D. GitHub 星标数

**解答：** A

**解析：** 路径和环境变量错误是训练失败的高频原因。


### 问题2（单选题）

**题目：** 训练日志中的 `loss` 主要用于观察什么？

A. 模型优化是否在正常进行
B. Git LFS 是否上传成功
C. 图片是否压缩
D. CANN 是否卸载

**解答：** A

**解析：** loss 是判断训练收敛趋势的重要指标。


### 问题3（单选题）

**题目：** `checkpoint` 的主要作用是？

A. 保存训练状态，便于恢复或评估
B. 替代数据集
C. 删除旧日志
D. 提高 XML 解析速度

**解答：** A

**解析：** checkpoint 通常保存模型权重、优化器状态和 epoch 信息。


### 问题4（单选题）

**题目：** 如果训练脚本提示找不到 `splits/train.txt`，最可能的问题是？

A. 数据准备没有完成或配置路径不正确
B. NPU 温度太低
C. 学习率太小
D. PPT 没有保存

**解答：** A

**解析：** split 文件来自数据准备阶段，路径或生成步骤错误都会导致找不到。


### 问题5（多选题）

**题目：** 判断一次单卡训练跑通，可以参考哪些输出？

A. 训练命令正常退出
B. loss 正常打印
C. throughput 或 step time 有输出
D. checkpoint 文件生成

**解答：** A、B、C、D

**解析：** 这些信息共同说明训练流程已经闭环。


### 问题6（多选题）

**题目：** 训练速度慢可能来自哪些因素？

A. DataLoader workers 太少或 I/O 慢
B. batch size 太小导致设备利用率低
C. 频繁保存或打印日志
D. 数据预处理在 CPU 侧成为瓶颈

**解答：** A、B、C、D

**解析：** 训练性能受计算、数据加载、I/O 和日志频率共同影响。


### 问题7（多选题）

**题目：** 断点恢复训练时通常需要哪些信息？

A. checkpoint 路径
B. 模型结构一致
C. 优化器/epoch 状态
D. 数据配置一致

**解答：** A、B、C、D

**解析：** 恢复训练不仅需要权重，还要保证训练上下文一致。


### 问题8（判断题）

**题目：** 只要生成了 checkpoint，就一定说明训练效果很好。

**解答：** 错误

**解析：** checkpoint 只说明保存成功，不等于模型精度或收敛质量好。


### 问题9（判断题）

**题目：** 单卡训练实践中，保留关键日志有助于后续性能分析和实验验收。

**解答：** 正确

**解析：** 日志是复现和解释训练结果的重要证据。


### 问题10（填空题）

**题目：** 训练脚本中常用 `--batch-size` 控制每次迭代输入的 `____` 数量。

**解答：** 样本或图片

**解析：** batch size 表示一个 step 中参与训练的样本数量。


### 问题11（填空题）

**题目：** 断点恢复通常通过参数 `--resume` 指向已有的 `____` 文件。

**解答：** checkpoint

**解析：** checkpoint 保存训练状态，是恢复训练的基础。


### 问题12（简答题）

**题目：** 为什么训练实践中要先跑少量 epoch 验证流程？

**解答：** 少量 epoch 可以快速验证环境、数据、模型、loss、保存和日志链路是否正常，避免长时间训练后才发现基础错误。

**解析：** 这是节省排错时间的实用策略。


### 问题13（简答题）

**题目：** 如果 loss 长时间不下降，可以从哪些方面分析？

**解答：** 检查标注格式和类别数是否匹配，确认学习率和 warmup 是否合理，查看 AMP 是否导致数值异常，观察 batch size 是否过小或数据是否异常。

**解析：** loss 问题通常需要同时看数据、模型和优化器。


### 问题14（简答题）

**题目：** 为什么实验报告不建议只写“训练成功”？

**解答：** “训练成功”缺少可验证证据。应写清楚启动命令、数据规模、关键参数、loss/throughput、checkpoint 路径和截图输出，才能支撑结论。

**解析：** 工程实验强调可复现和可检查。


### 问题15（代码设计题）

**题目：** 写一条命令，使用单卡脚本训练 2 个 epoch，并显式设置 batch size 和 workers。

**解答：** ```bash
python src/scripts/train_yolo_single_npu_amp.py   --config src/configs/yolo_ascend.yaml   --epochs 2   --batch-size 8   --workers 4
```

**解析：** 具体脚本名以当前实验目录为准，命令体现了单卡训练和关键调参入口。
