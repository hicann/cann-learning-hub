### 问题12（简答题）

**题目：** 为什么部署验收要同时检查文件产物、安装位置和运行输出？

**解答：** 文件产物说明编译打包完成，安装位置说明 OPP 已进入 CANN 搜索路径，运行输出说明 ACLNN 调用链路和算法结果正确。三者缺一都不能完整证明部署成功。

**解析：** 验收要覆盖构建、部署和执行三层。


### 问题13（简答题）

**题目：** 遇到 `ModuleNotFoundError: yolo_nms_custom` 是否一定说明实验失败？

**解答：** 不一定。当前实验主验证路径是 ACLNN C++ runner，不依赖 Python wrapper。如果 notebook 某段试图 import Python 封装而未实现，就会报错，但 ACLNN 路径仍可完成验证。

**解析：** 要区分 Python 封装路径和 ACLNN 调用路径。


### 问题14（简答题）

**题目：** 如何把本实验结果写成严谨的最终结论？

**解答：** 应说明 YOLOv5s OM 能在 PyACL 中加载推理，CPU 后处理 baseline 已建立，YoloNmsCustom 自定义算子已编译安装，通过 ACLNN 接入 NPU 后处理，并与 CPU NMS 的 keep/count 对齐；Profiling 已采集并定位到自定义算子耗时。

**解析：** 结论要覆盖端侧推理、自定义算子、正确性和性能分析。


### 问题15（代码设计题）

**题目：** 写一组命令，快速检查自定义 OPP 安装和 ACLNN 头文件是否存在。

**解答：**

```bash
find /usr/local/Ascend/ascend-toolkit/8.0.RC1/opp/vendors/customize \
  -name '*YoloNmsCustom*' -o -name '*yolo_nms_custom*' | head -20

sed -n '1,120p' \
  /usr/local/Ascend/ascend-toolkit/8.0.RC1/opp/vendors/customize/op_api/include/aclnn_yolo_nms_custom.h
```

**解析：** 路径可按实际 CANN 版本调整，重点是检查 custom OPP 文件和 ACLNN 接口签名。
