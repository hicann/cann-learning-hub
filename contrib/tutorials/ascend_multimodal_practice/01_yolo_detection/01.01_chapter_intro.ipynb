# 01.01 章节概述：YOLO 目标检测

## 本章节面向谁

本章带你从零掌握 YOLO 目标检测：先用 OpenCV 理解图像本质，再用 YOLOv10 做推理，最后在华为云昇腾环境完成迁移训练——让模型学会识别自定义的 5 类水果。

## 章节目标

学完本章后，你将能够：

1. 理解**图像在计算机中就是数字矩阵**，掌握 OpenCV 读写、绘图等基础操作；
2. 用 YOLOv10 预训练模型做目标检测，解读边界框、类别、置信度；
3. 在华为云昇腾环境完成**迁移训练**，让模型识别自定义类别；
4. 解读训练指标（loss、mAP、混淆矩阵），评估模型质量。
