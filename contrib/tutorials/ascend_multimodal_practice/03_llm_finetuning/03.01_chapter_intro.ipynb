# 03.01 章节概述：大模型 LoRA 微调

<img src="./images/lora_overview.png" width="600">

## 本章节面向谁

本章带你从零在华为云昇腾 NPU 上完成 DeepSeek-R1-Distill-Qwen-1.5B 的 LoRA 高效微调（基于 PyTorch + torch_npu），得到一个"会扮演甄嬛"的对话模型。

LoRA 通过**冻结原始权重、只训练少量低秩矩阵**，把可训练参数压缩到全量的 0.5%，让普通开发者也能在单卡上微调十亿级模型。

## 章节目标

学完本章后，你将能够：

1. 理解 **LoRA 低秩注入**原理，说清它为何省显存；
2. 在昇腾 NPU 环境完成依赖安装、模型下载、数据预处理；
3. 独立编写对话模板构造、分词、掩码生成的数据处理代码；
4. 配置 `LoraConfig` 并用 `Trainer` 完成训练；
5. 加载 LoRA 权重推理，对比微调前后效果。

## 环境要求

本章必须使用**昇腾 NPU**（CPU 无法训练 1.5B 模型）。

- Python 内核：**Python 3.11.4 (CANN)**
