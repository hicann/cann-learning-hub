# 02.01 章节概述：华为云语音与视觉 SDK

<img src="./images/speech_overview.png" width="600">

## 本章节面向谁

本章带你使用华为云语音交互服务（SIS）和文字识别服务（OCR），结合大语言模型（LLM），构建"能听、能看、能想、能说"的多模态智能应用。

## 章节目标

学完本章后，你将能够：

1. 掌握**音频数字化**基础（采样率、位深），完成音频播放与录音；
2. 用华为云 SIS 做 **ASR 语音识别**、**TTS 语音合成**、**声音复刻**；
3. 用华为云 **OCR** 识别图片文字，集成 **LLM** 做对话与工具调用；
4. 把语音+视觉+LLM 组合成完整的智能应用（拍照读字、语音机器人）。
