## 共享地址空间

OpenMP 线程共享矩阵和 x，每个循环迭代只写自己的 y 元素。`parallel for` 创建线程组（team）并分配行迭代。

## 调度策略

`schedule(static)` 预先分配连续迭代，开销低；当每行 nnz 差异很大时可能负载不均。dynamic 可改善不均衡，但引入调度开销。

## 课程级适配

原工程固定调用 `omp_set_num_threads(16)` 并使用 `schedule(static)`。为支持不改源码的线程数和调度实验，课程副本改用 `schedule(runtime)`：未设置环境变量时仍默认 16 线程和 static；设置 `OMP_NUM_THREADS`、`OMP_SCHEDULE` 后由 OpenMP 运行时读取实验配置。该改动不改变 SpMV 数值计算。
