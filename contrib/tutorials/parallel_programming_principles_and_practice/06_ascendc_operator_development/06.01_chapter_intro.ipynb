## 共同证据标准

两条子路径共用同一套证据标准：

- 构建证据：kernel 与工程能完成真实编译，记录编译命令与输出；
- 调用证据：程序实际提交并执行算子，能通过日志或接口返回定位真实执行位置；
- 正确性证据：NPU 输出与参考实现（CPU/NumPy）一致；
- 结论边界：只有证据齐备的路径才能写 PASS；无通过证据的路径写「未验证/不得 PASS」；
- 只有课程为该目标定义了具体状态码时才使用该状态码：本实验中 `DEFERRED_UNSUPPORTED_TARGET` 仅对应 RoPE SIMT 的 Ascend 950 模板，不适用于其他未验证路径。

## 执行路径与硬件边界

- SpMV 的正式可执行路径是 **SIMD/Vector Core**：已有真实 Ascend C SIMD/Vector Core kernel 与 A3 真机证据；当前**没有**可声称通过的 SpMV SIMT 真机路径。
- RoPE 的 **SIMD/RTC 路径在 A3 通过**（本轮验证硬件为 Atlas A3，SoC Ascend910_9362；`dav-2201` 是编译目标说明，910B3 是历史实现/验证背景，均不等同于本轮 A3）。
- RoPE 的 SIMT 文件仅是 **Ascend 950 模板**：当前 A3 不支持，状态为 `DEFERRED_UNSUPPORTED_TARGET`，只用于理解线程级映射，不得写成已在真机通过。

本轮验证硬件统一为 **Atlas A3（SoC Ascend910_9362）**；章节中出现的 `dav-2201` 仅为编译目标说明、910B3 仅为历史实现/验证背景，均不得等同或冒充本轮 A3。

本实验所有结论以当前运行证据为准，历史记录不能替代当前真机验证。
