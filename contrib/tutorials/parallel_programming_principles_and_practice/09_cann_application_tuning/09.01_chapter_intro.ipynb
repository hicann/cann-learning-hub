## 统一的调优方法

两条子路径共用同一套调优方法，共四步：

1. **基线**：先记录未修改版本的运行时间与正确性结果，作为比较基准；
2. **受控/公平比较**：Dis-GMRES 采用单变量实验（一次只改一个变量，其余保持不变）；MuduoXinyu 是固定模型、输入、解码参数、环境和统计口径下的复合实现包 A/B（Path A 与 Path B 同时改变算子组织与 dtype），结论只能归因于整体实现包；
3. **正确性门禁**：性能数据只有在功能与数据校验通过后才有效；
4. **性能判定**：用同等条件下的多次测量（如 1 次预热 + 3 次统计）判定收益；负收益如实记录。

## 实验条件

- **Dis-GMRES**：GMRES 外层控制在 **Host CPU** 执行；SpMV/Dot/AXPY 等热点在 **AI Core** 执行；world1 已通过，**HCCL 2/4/8 卡未验证**，不得声称多卡扩展性结论。
- **MuduoXinyu**：功能与数据正确性已通过；本轮 Atlas A3 上 Path A mean `166.35666666666665` tokens/s、Path B mean `170.086` tokens/s，Path B 相对 Path A 为 `+2.24176969162644%`（`performance_valid=true`、`performance_beneficial=true`）；该结论只适用于本次 Atlas A3、当前模型/参数与运行顺序，不泛化；其他环境必须重跑公平 A/B。若未来新实测为负值，必须如实报告，不得包装为优化成功。
- 范围仅为上述两条子路径：不新增 Vadd，不迁移 5.1 的故障 Add 调试载体。
