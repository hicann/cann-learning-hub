## 执行环境诊断

从当前章节目录执行下面的 Cell，并对照随后给出的检查点阅读输出。

```python
import os, shutil
for name in ['ASCEND_TOOLKIT_HOME', 'ASCEND_HOME_PATH', 'ASCEND_OPP_PATH']:
    print(name + '=' + os.environ.get(name, '<not set>'))
for tool in ['npu-smi', 'clang', 'clang++', 'bisheng', 'cmake']:
    print(tool + ': ' + (shutil.which(tool) or 'not found'))
```

## 预期现象与结果分析
