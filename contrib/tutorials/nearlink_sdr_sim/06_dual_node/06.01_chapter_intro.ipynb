# 第六章 双节点端到端通信

前五章我们分别学习了物理层波形、信道编码、MAC 帧结构、链路保障机制、跳频和功率自适应技术。本章将上述模块全部串联，构建 G 节点与 T 节点的完整端到端通信系统，涵盖建链、数据交换、安全通信和 MCS 自适应跟踪。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 理解 SleNode 双节点模型的广播->扫描->连接->数据交换全流程
- 掌握 MAC-PHY 转换（mac_to_iq / iq_to_mac）的基本原理
- 理解 FER 和 Byte BER 作为链路性能指标的含义及区别
- 了解 MCS 自适应和 LinkQualityTracker 的滑动窗口机制
- 已完成第一至五章的学习

---

## 章节目标

完成本章学习后，你将能够：

- 掌握双 SleNode 端到端通信的完整建链与数据交换流程
- 理解 FER 与 Byte BER 的区别与各自的适用场景
- 掌握 ECDH 密钥协商与 AES-CCM 加密在数据帧保护中的应用
- 能够分析 MCS 自适应在 SNR 固定场景下的跟踪行为
