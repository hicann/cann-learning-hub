# 第二章 Polar 编码与 BPSK 调制性能分析

信道编码是数字通信系统提升链路可靠性的核心技术。Polar 码作为星闪 SLE 标准采用的编码方案，基于信道极化思想实现逼近香农极限的传输性能。本章从无编码 BPSK 的基线性能出发，逐步引入 Polar 编码的编解码原理，通过 BER 曲线对比定量标定编码增益，并探究码率与码长对链路性能的影响规律。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 掌握 Python 基础语法及 NumPy 数组操作
- 理解 dB 与线性值的换算关系
- 了解 BPSK 调制的基本概念（0 -> +1, 1 -> -1）
- 掌握 BER（误码率）的蒙特卡洛统计方法
- 已完成第一章 GFSK 调制解调的学习

无须具备信道编码的理论基础，也不要求提前掌握 Polar 编解码算法。

---

## 章节目标

完成本章学习后，你将能够：

- 理解信道编码的基本思想及 Polar 码在星闪系统中的地位
- 掌握 Polar 码的信道极化原理及 Arikan 生成矩阵的递归构造
- 掌握 SC 逐次消除解码算法的 LLR 域递推过程
- 能够绘制 Polar 编码与无编码链路的 BER 对比曲线
- 能够定量标定 Polar 码的编码增益
- 理解码率与码长对 Polar 码性能的影响规律
- 理解低 SNR 下编码性能劣于无编码的物理原因及编码门限的含义
