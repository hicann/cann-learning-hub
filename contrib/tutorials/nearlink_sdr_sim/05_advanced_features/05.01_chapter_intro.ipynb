# 第五章 跳频、接入与功率自适应

前四章我们学习了从物理层波形到 MAC 层帧交换的完整链路。但实际无线通信环境远比 AWGN 复杂——多径衰落、同道干扰和动态信道质量变化都需要专门的对抗机制。本章聚焦于星闪 SLE 标准中的抗干扰与链路管理技术：跳频扩频、接入建链流程和发射功率自适应控制。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 理解 AWGN 信道与 Rayleigh 衰落信道的基本区别
- 掌握 Polar 编码与 BPSK/QPSK 调制的流程
- 理解 FER（误帧率）作为链路性能指标的含义
- 熟悉 MAC 帧结构与 MAC-PHY 转换流程
- 已完成第一至四章的学习

---

## 章节目标

完成本章学习后，你将能够：

- 理解跳频 PRNG 的算法原理与跳频分集增益的来源
- 掌握星闪 SLE 六阶段接入建链流程与 G/T 角色协商机制
- 理解闭环功率控制的三类信令与四种应用场景
- 能够分析跳频在 Rayleigh 衰落信道下的 FER 改善效果
- 能够评估不同 SNR 下接入成功率的门限行为
- 理解功率自适应中失败阈值与功率步长对收敛行为的影响
