# 第一章 GFSK 调制解调基础

GFSK（高斯频移键控）是星闪 SLE 标准中帧类型 1 采用的基础调制方式。本章从数字调制的基本概念出发，逐步深入 GFSK 的调制解调原理，通过完整的仿真链路掌握从随机比特到 IQ 信号再到误码率统计的全过程，并探究关键参数对链路性能的影响规律。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 掌握 Python 基础语法（变量、循环、函数定义与调用）
- 理解 NumPy 数组的基本操作（创建、索引、数学运算）
- 了解复数及其基本运算（实部、虚部、模、辐角）
- 了解 dB（分贝）的基本概念及与线性值的转换关系
- 了解概率统计基础概念（高斯分布、均值、方差）

无须具备任何通信系统或星闪标准的前置知识，也不要求提前掌握 GFSK 调制原理。

---

## 章节目标

完成本章学习后，你将能够：

- 理解数字调制的基本概念及 GFSK 在星闪系统中的地位
- 掌握 GFSK 从比特到 IQ 信号的四步调制流程
- 理解 AWGN 信道模型及 SNR 的物理含义
- 掌握 GFSK 频率鉴别器非相干解调原理
- 能够编写完整可运行的 GFSK 链路仿真代码
- 能够绘制 BER-SNR 曲线并与理论值对比验证
- 理解 sps 与 mod_index 参数对 BER 的影响规律
- 能够从零手写 GFSK 调制解调，不依赖库函数
