# 第三章 MAC 帧类型传输与 FER 性能对比

MAC 层是连接物理层与上层应用的关键桥梁。星闪 SLE 标准定义了三种 MAC 帧类型——信令帧、异步数据帧和复用帧。本章从帧结构的基础知识出发，通过帧级端到端仿真对比三种帧在 AWGN 信道下的误帧率差异，并分析数据帧的有效吞吐量。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 理解 Polar 编码与 BPSK 调制的基本流程
- 了解 AWGN 信道模型及 SNR 的概念
- 掌握 FER（误帧率）与 BER（误码率）的区别
- 已完成第一、二章的学习

无须预先了解 MAC 层协议或帧结构的具体实现。
