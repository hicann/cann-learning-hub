# 第四章 HARQ 重传、AMC 自适应与流控机制

链路可靠性不仅依赖物理层的编码调制，还需要 MAC 层的保障机制来应对信道波动。本章聚焦于星闪 SLE MAC 层的三项核心保障技术——HARQ 重传、AMC 自适应 MCS 跟踪和 QoS 流控背压，通过帧级仿真验证其对链路质量的改善效果。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 理解 Polar 编码与 QPSK 调制的流程
- 理解 AWGN 信道模型及 SNR 的概念
- 掌握 FER（误帧率）与吞吐量的计算方法
- 已完成第一至三章的学习

---

## 章节目标

完成本章学习后，你将能够：

- 理解 HARQ Chase Combining 重传机制的原理与吞吐量权衡
- 掌握 AMC 滑动窗口 FER 驱动的 MCS 调整策略
- 理解 QoS 流控背压的高/低水位调控机制
- 能够分析 HARQ 在低/中/高 SNR 区间的 FER 与吞吐量行为差异
- 能够评估 AMC 的跟踪精度与滑动窗口大小的影响
