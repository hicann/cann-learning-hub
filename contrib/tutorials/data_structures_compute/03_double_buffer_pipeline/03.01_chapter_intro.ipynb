# 03.01 章节介绍：双缓冲数据流水线

本章在实验 1 的 VectorAdd 基础上，引入 <code>TQue</code> 队列所有权、物理 Buffer 槽位和预取/排空调度。你将先区分 queue depth 与 Buffer 数量，再比较单缓冲和双缓冲，最后独立完成一个可验证的双缓冲 VectorAdd。


## 章节学习前置要求

开始本章前，你应当：

- 已完成 [实验 1：基础实操](../01_basic_operations/01.05_chapter_test.ipynb)，能解释 <code>N → Block → Tile</code> 的数据切分；
- 能按 <code>Init → Process → CopyIn/Compute/CopyOut</code> 顺序阅读基础 VectorAdd；
- 知道 GM、UB、<code>GlobalTensor</code>、<code>LocalTensor</code> 和 <code>AscendC::Add</code> 的作用；
- 已进入带有 CANN 9.0.0 与 Ascend 910B3 的 CANNLab 环境，并使用 <code>cann_py311</code> 内核。

本章不要求你使用手写 Event，也不讨论非对齐尾块；实验仍使用 <code>float32</code>、均匀切分和 32 Byte 对齐的连续 DataCopy。


## 章节目标

完成本章后，你应能够：

1. 写出输入、输出 Tensor 在 <code>AllocTensor/EnQue/DeQue/FreeTensor</code> 间的生命周期；
2. 区分 <code>TQue&lt;position, depth&gt;</code> 的 depth 与 <code>InitBuffer(queue, num, size)</code> 的 num；
3. 计算单缓冲、双缓冲的每核队列有效载荷，并在 Kernel 启动前判断 UB 预算；
4. 写出预装第一个 Tile、预取下一个 Tile、写回上一个结果、排空最后结果的调度；
5. 证明 <code>tileCount</code> 不随 <code>bufferNum</code> 改变，并用 MockPipeline 与 NPU 精度用例验证实现。
