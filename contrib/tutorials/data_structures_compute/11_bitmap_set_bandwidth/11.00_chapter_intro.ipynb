# 11.00 章节介绍：位图压缩与集合交运算的带宽优化

本章对应<strong>实验11：位图压缩与集合交运算的带宽优化</strong>。你将从集合抽象数据类型出发，依次完成位图成员映射、插入/删除/查询、集合交参考模型，再把逐字集合交映射为 Ascend C 算子，最后用真实物理字节数与 NPU 计时判断带宽优化是否成立。学习主线是“集合抽象 → 位图结构 → 基本操作 → 集合交 → Ascend C 算子 → 访存优化”，而不是只练习一次按位与。

---

## 章节学习前置要求

开始本章前，你应具备以下基础：

- 理解集合的成员关系、交集、并集和差集；
- 掌握数组下标与 C/C++ 位运算 <code>&amp;</code>、<code>|</code>、<code>&lt;&lt;</code>；
- 理解一维连续存储、左闭右开区间和 32 Byte 搬运对齐；
- 了解 <code>GlobalTensor</code>、<code>LocalTensor</code>、<code>TPipe</code>、<code>TQue</code> 与 <code>DataCopy</code> 的基本作用；
- 会使用 Linux、CMake 和 Jupyter Notebook。

不要求提前掌握哈希表实现、双缓冲调度或归约算子。本章使用单缓冲，使性能归因集中在集合的物理表示与 GM 流量。

---

## 章节目标

完成本章后，你应能够：

1. 区分集合抽象与稠密数组、哈希集合、位图三种物理表示；
2. 使用 <code>wordIndex=value/32</code> 和 <code>bitIndex=value%32</code> 实现位图成员操作；
3. 区分逻辑全集大小、有效位图字数和 32 Byte 对齐后的物理行跨度；
4. 为 DenseAnd 与 BitmapAnd 推导连续多核切分、UB Tile 和三队列空间；
5. 正确覆盖 <code>U=31/32/33</code>、尾 Block、尾 Tile、无效高位、填充区和输出 Guard；
6. 同时报告 Kernel 时间、实际物理流量与逻辑有效带宽，解释压缩比和加速比为什么不等价。
