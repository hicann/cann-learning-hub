### 实践题：手写后缀表达式求值

请使用 Python 实现一个后缀表达式求值函数，验证你对后缀表达式求值算法的理解。

**要求**：
- 输入：一个列表，元素为非负整数（操作数）或字符串 `'+'`、`'-'`、`'*'`、`'/'`（运算符）。
- 输出：计算结果（float）。
- 使用栈结构实现。

**测试用例**：
- `[3, 2, '*', 5, '+']` → 11.0（即 3×2+5）
- `[4, 2, 2, '*', '+']` → 8.0（即 4+2×2）
- `[5, 6, '*']` → 30.0（即 5×6）

```python
# 请在此处编写你的后缀表达式求值函数
def suffix_eval(tokens):
    """
    后缀表达式求值
    
    Args:
        tokens: list, 元素为 int（操作数）或 str（'+', '-', '*', '/'）
    Returns:
        float: 计算结果
    """
    stack = []
    for token in tokens:
        if isinstance(token, int):
            stack.append(float(token))
        else:
            b = stack.pop()
            a = stack.pop()
            if token == '+':
                stack.append(a + b)
            elif token == '-':
                stack.append(a - b)
            elif token == '*':
                stack.append(a * b)
            elif token == '/':
                stack.append(a / b if b != 0 else 0.0)
    return stack[0] if stack else 0.0

# 测试
assert suffix_eval([3, 2, '*', 5, '+']) == 11.0, '测试失败: 3*2+5'
assert suffix_eval([4, 2, 2, '*', '+']) == 8.0, '测试失败: 4+2*2'
assert suffix_eval([5, 6, '*']) == 30.0, '测试失败: 5*6'
print('所有测试通过！')
```
