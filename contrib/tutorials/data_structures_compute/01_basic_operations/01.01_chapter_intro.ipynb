# 01.01 章节介绍：基础实操

本章对应课程的<strong>实验 1：基础实操</strong>。你将从可复现的 CANNLab 环境检查开始，一键启动 Ascend C 工程；随后阅读并运行 VectorAdd，最后借助交互式 Tiling 图把一维连续数组映射到多 Block、多 Tile 的数据区间，并独立完成综合编程实践。

---

## 章节学习前置要求

开始本章前，你应具备以下基础：

- 掌握 C/C++ 数组、函数、宏和整数除法；
- 理解一维数组的连续存储和左闭右开区间；
- 会使用 Linux 命令、CMake 和 Jupyter Notebook；
- 已进入分配了昇腾 NPU 的 CANNLab 环境，镜像模板为 <code>cann_9.0.0 py3.11-A2-arm</code>，并选择 <code>Python 3.11.4 (CANN)</code> 内核。

不要求提前掌握 Ascend C 自定义算子开发、动态 Shape 或性能调优。

---

## 章节目标

完成本章后，你应能够：

1. 检查 CANN、ASC 编译工具和 NPU 是否就绪，并一键完成工程配置、编译和运行；
2. 说明 SIMD Membase 路径中 GM、Local Memory、队列和 LocalTensor 的关系；
3. 沿 <code>Init → Process → CopyIn/Compute/CopyOut</code> 主线读懂并运行 VectorAdd；
4. 交互调整 <code>N</code>、<code>blockDim</code> 与 <code>tileCount</code>，判断整除和 32 Byte 对齐约束；
5. 补全核间偏移、核内偏移和 <code>AscendC::Add</code>，通过构建、非法参数与 CPU Golden 精度自检。
