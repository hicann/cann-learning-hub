# Git指令汇总

前面的课程将git指令分类介绍了一遍。本节将 Git 指令汇总，方便查找。




## 1. 环境配置与初始化 (Setup & Init)
| 指令 | 说明 | 使用场景 / 示例 |
| :--- | :--- | :--- |
| `git config --global user.name "<Name>"` | 配置全局用户名 | `git config --global user.name "Your Name"` |
| `git config --global user.email "<Email>"` | 配置全局邮箱 | `git config --global user.email "you@example.com"` |
| `git config --list` | 查看当前生效的所有配置 | 检查用户名、别名及远程设置 |
| `git init` | 在当前目录初始化本地仓库 | 新项目起步 |
| `git init -b <branch>` | 初始化仓库并指定默认分支名 | `git init -b main` |
| `git clone <url>` | 完整克隆远程仓库及历史 | `git clone git@gitcode.com:.../repo.git` |

## 2. 本地三区核心流转 (Working -> Staging -> Repository)
| 指令 | 说明 | 使用场景 / 示例 |
| :--- | :--- | :--- |
| `git status` | 查看工作区与暂存区状态 | 确认当前改动与分支状态 |
| `git add <file>` | 将单个或指定文件加入暂存区 | `git add app.py` |
| `git add .` / `git add -A` | 添加所有改动至暂存区（`.`当前目录，`-A`全仓库） | 批量暂存所有变动 |
| `git add -p` | 交互式选择代码块（Hunk）暂存 | 拆分同一文件中的多个独立修改 |
| `git commit -m "<msg>"` | 提交暂存区改动并记录描述 | `git commit -m "feat: add user login"` |
| `git commit -am "<msg>"` | 针对已跟踪文件的暂存+提交一步到位 | 快速提交小修改（不包含新建未跟踪文件） |
| `git commit --amend` | 追加改动到上一提交或修改提交信息 | 漏加了文件或 Commit Message 拼写错误 |
| `git rm --cached <file>` | 移出版本控制，但保留本地物理文件 | 误提交了 `.env` 或编译输出文件 |
| `git clean -fd` | 强制清除工作区所有未跟踪的文件和目录 | 彻底清理本地编译临时文件与残留 |

## 3. 差异对比与历史追溯 (Inspect & History)
| 指令 | 说明 | 使用场景 / 示例 |
| :--- | :--- | :--- |
| `git diff` | 工作区 vs 暂存区（未 add 的改动） | 检查刚刚写了什么 |
| `git diff --staged` | 暂存区 vs 最新提交（已 add 未 commit） | 提交前的最终代码审查 |
| `git diff HEAD` | 工作区 vs 最新提交（所有未 commit 改动） | 统览所有未落库的代码 |
| `git diff <c1> <c2>` / `<b1> <b2>` | 对比两个提交或两个分支间的差异 | `git diff main feature/login` |
| `git log --oneline --graph --all`| 以 ASCII 树状图显示所有分支演进 | 最推荐的日常历史概览方式 |
| `git log -p -<N>` | 查看最近 N 次提交的具体差异详情 | `git log -p -2` |
| `git log --stat` | 统计每次提交涉及的文件行数变动 | 查看提交规模 |
| `git log --author="<Name>" --since="<时间>"` | 按作者 / 时间过滤提交历史 | 定位某人在某段时间内的提交 |
| `git blame <file>` | 逐行追溯代码作者与提交记录 | 排查线上 bug 引入源头 |
| `git reflog` | 查看 HEAD 指针完整变动流水（终极时光机） | 误删分支或 hard 回滚后找回丢失提交 |

## 4. 撤销、变动修复与回滚 (Undo & Reset)
> 注：Git 2.23+ 推荐使用解耦后的 `restore` 替代容易歧义的 `checkout`。

| 指令 | 说明 | 风险级别与注意事项 |
| :--- | :--- | :--- |
| `git restore <file>` | 丢弃工作区修改，恢复至暂存区状态 | ⚠️ **有损**：工作区未暂存修改将丢失 |
| `git restore --staged <file>` | 将文件撤出暂存区（保留工作区修改） | 🟢 **安全**：等同于旧版 `git reset HEAD <file>` |
| `git reset --soft HEAD~1` | 撤销上一条 commit，改动保留在暂存区 | 🟢 **安全**：用于重新合并或调整提交结构 |
| `git reset --mixed HEAD~1` | 默认模式：撤销 commit 并撤出暂存区，保留工作区 | 🟢 **安全**：代码还在，只是变为未暂存状态 |
| `git reset --hard HEAD~1` | 彻底抹除上一条 commit，重置工作区与暂存区 | 🔴 **危险**：本地修改彻底丢失（需靠 `reflog` 救援） |
| `git revert <commit>` | 产生一个反向修改的新 commit 抵消历史 | 🟢 **安全**：公共分支回滚的标准做法，不重写历史 |

## 5. 现场暂存与状态隔离 (Stash) 
| 指令 | 说明 | 使用场景 / 示例 |
| :--- | :--- | :--- |
| `git stash` / `git stash save "<msg>"` | 暂存当前未提交的工作区与暂存区修改 | 做到一半突然要切分支修线上 emergency bug |
| `git stash list` | 查看所有暂存的工作现场记录 | `git stash list` |
| `git stash pop` | 恢复最近一次暂存，并从 stash 列表中删除 | 修完 bug 回来继续开发 |
| `git stash apply [stash@{n}]` | 恢复指定暂存记录，但不从列表中删除 | 将同一改动复用到多个分支测试 |
| `git stash drop [stash@{n}]` | 丢弃指定的一条暂存记录 | 清理无用暂存 |
| `git stash clear` | 清空所有 stash 暂存 | 彻底清空现场记录 |

## 6. 分支协作与集成 (Branching, Merge, Rebase)
| 指令 | 说明 | 使用场景 / 示例 |
| :--- | :--- | :--- |
| `git branch` / `git branch -r` / `git branch -a` | 列出本地分支 / 远程分支 / 全部分支 | 查看当前所在位置与全部分支 |
| `git switch <branch>` | 切换分支（现代推荐语法） | `git switch main` |
| `git switch -c <branch>` | 创建并切换到新分支 | `git switch -c feature/pay`（等同于 `checkout -b`） |
| `git branch -d <branch>` | 安全删除已合并的分支 | `git branch -d feature/pay` |
| `git branch -D <branch>` | 强制删除未合并的分支 | 放弃废弃的实验性功能分支 |
| `git merge <branch>` | 将目标分支合并到当前分支 | 产生 Merge Commit 或 Fast-Forward |
| `git merge --abort` | 放弃当前冲突的合并，回到合并前状态 | 解决冲突混乱时一键重来 |
| `git rebase <base>` | 变基操作：将当前分支的一组提交移至基底顶端 | 消除冗余合并节点，维持线性干净历史 |
| `git rebase -i HEAD~<N>` | 交互式变基：合并/拆分/重排过去 N 个提交 | 提交 PR 前整理凌乱的本地小 commits（Squash） |
| `git cherry-pick <commit>` | 精准挑拣某个特定提交应用到当前分支 | 将某分支上的单个修复补丁同步至发布分支 |

## 7. 版本发布与里程碑 (Tags) 
| 指令 | 说明 | 使用场景 / 示例 |
| :--- | :--- | :--- |
| `git tag` | 列出所有已存在的标签 | 查看历史版本号列表 |
| `git tag -a <tag> -m "<msg>"` | 创建附注标签（记录打标者、日期与信息） | `git tag -a v1.0.0 -m "Release version 1.0.0"` |
| `git tag <tag> <commit>` | 针对历史上的某次特定 commit 打标签 | 为遗漏的历史发布补打 tag |
| `git push origin <tag>` | 推送指定本地标签到远程仓库 | `git push origin v1.0.0` |
| `git push origin --tags` | 一次性推送所有未推送的本地标签 | 批量发布版本标签 |
| `git tag -d <tag>` | 删除本地标签 | `git tag -d v1.0.0` |

## 8. 远程同步与多端协作 (Remote & Collaboration)
| 指令 | 说明 | 使用场景 / 示例 |
| :--- | :--- | :--- |
| `git remote -v` | 查看已配置的远程源及对应 URL | 确认 origin / upstream 配置 |
| `git remote add <name> <url>` | 添加远程仓库源 | Fork 协作流中关联上游：`git remote add upstream <url>` |
| `git fetch <remote>` | 同步远程所有分支与对象的更新（不改工作区） | 配合 `git diff` 安全查看远程更新 |
| `git pull <remote> <branch>` | 拉取远程更新并合并（`fetch` + `merge`） | `git pull origin main` |
| `git pull --rebase` | 拉取远程更新并通过变基方式合并 | 避免拉取更新时产生“Merge branch...”交叉提交 |
| `git push <remote> <branch>` | 推送本地分支到远程仓库 | `git push origin feature/login` |
| `git push -u <remote> <branch>` | 首次推送并绑定追踪分支（upstream） | `git push -u origin main` |
| `git push --force-with-lease` | 带租约保护的强制推送 | 本地 rebase 后覆盖远程个人功能分支（比 `--force` 安全） |
