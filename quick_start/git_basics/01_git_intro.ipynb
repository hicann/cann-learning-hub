## 1. 什么是 Git？

**Git** 是一个**分布式版本控制系统**（Distributed Version Control System, DVCS），由 Linus Torvalds 于 2005 年创建，用于追踪计算机文件的变化，并协调多人协作开发。

简单来说，Git 解决的是这样一个问题：**「如何高效地记录、回溯和协同修改代码？」**

### Git 能做什么？

| 场景 | 没有 Git | 有 Git |
|------|---------|--------|
| 修改了代码想撤销 | 靠记忆手动改回去 | 一条命令回到任意历史版本 |
| 多人同时改同一份代码 | 互相覆盖，靠发邮件合并 | 自动合并，冲突可视化解决 |
| 想知道某行代码是谁写的 | 无从考证 | `git blame` 直接定位 |
| 想备份项目历史 | 手动复制整个文件夹 | 完整历史自动留存，秒级克隆 |

### Git 的核心特点

1. **分布式**：每个开发者本地都有完整的仓库历史，不依赖中央服务器即可工作
2. **快**：几乎所有操作（提交、分支、日志查询）都在本地完成，无需联网
3. **安全**：使用 SHA-1 校验和为每一次提交生成唯一指纹，任何篡改都会被立即发现
4. **分支友好**：分支本质上只是一个指针，创建、切换、合并成本极低

> 💡 Git 最初是为 Linux 内核开发而设计的，如今已成为软件行业事实上的版本控制标准，广泛用于 GitHub、GitLab、GitCode 等代码托管平台。

## 2. Git 的核心概念：三个区域

要理解 Git，先记住它把你的项目分成 **三个区域**。你后续几乎所有的 Git 操作，本质上都是在把内容从一个区域搬到另一个区域。

可以用「档案管理」来类比整个过程：

```
┌──────────────────────┐ 把要归档的放进文件篮 ┌──────────────────────┐     编号登记入柜     ┌──────────────────────┐
│ 工作区 Working       │ ───────────────────▶ │ 暂存区 Staging       │ ───────────────────▶ │ 仓库 Repository      │
│ 办公桌上正改的文件   │                      │ 待归档文件篮：可增减 │                      │ 档案柜：已归档的卷宗 │
└──────────────────────┘                      └──────────────────────┘                      └──────────────────────┘
```

#### 三个区域分别是什么？

| 区域 | 生活类比 | 在 Git 里是什么 |
|------|---------|----------------|
| **工作区** | 办公桌上摊开、正在翻改的文件 | 你在文件系统里直接看到、正在编辑的目录和文件 |
| **暂存区** | 「待归档」文件篮：先放进去准备归档，还能再加、能取出 | 一份「待存档清单」，记录下一次要正式存档哪些改动 |
| **仓库** | 档案柜里已编号归档的卷宗，永久保存、随时调阅任意一份 | 已正式存档的全部历史版本，永久保留、可随时回溯任一版本 |

#### 为什么不直接从工作区跳到仓库？

设想你正在改代码：A 文件改完了，B 文件改到一半还不想归档，又顺手在 C 文件里修了个错别字。如果「一改就直接进档案柜」，这三处改动会一股脑全归档进去，你没法只归档 A 和 C、把半成品的 B 留在外面。

**暂存区就是解决这个问题的缓冲地带**：你先把想归档的改动放进「待归档文件篮」（可以分多次放、也能取出来），等整理成一组完整、干净的改动后，再一次性编号登记、正式归入档案柜。这样存进档案柜的每一份，都是你深思熟虑后的版本，而不是手忙脚乱的中间状态。

> 💡 对应到 Git 命令（这些命令后续课程会详细讲，**现在只需先建立「三个区域」的概念**）：
> - `git add` —— 把工作区的改动放进「待归档文件篮」
> - `git commit` —— 把文件篮编号登记、正式归入档案柜
> - `git status` —— 查看当前三个区域各有什么内容，这是你最常用的「看一眼」命令

## 3. 在 Linux 上安装并配置 Git

### 3.1 安装 Git

大多数 Linux 发行版已预装 Git。如果没有，按发行版安装即可。

````python
# 检查 Git 是否已安装及其版本
!git --version
````

如果未安装，按发行版执行以下命令：

```bash
# Debian / Ubuntu
sudo apt-get update && sudo apt-get install -y git

# CentOS / RHEL / openEuler
sudo yum install -y git   # 或 sudo dnf install -y git

# Arch Linux
sudo pacman -S git
```

### 3.2 首次配置 Git

安装后，第一步是设置你的**用户名**和**邮箱**。这两个信息会写入每一次提交，用于标识“谁做了这次修改”。

> ⚠️ 请将下面的用户名和邮箱替换为你自己的真实信息。如果你计划向 GitCode 平台提交 PR，建议使用与 GitCode 账号一致的邮箱。

````python
# 配置全局用户名和邮箱（--global 表示对当前用户的所有仓库生效）
!git config --global user.name "Your Name"
!git config --global user.email "you@example.com"

# 查看当前配置
!git config --list
````
