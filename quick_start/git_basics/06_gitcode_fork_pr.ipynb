## 1. 什么是 Fork？为什么贡献要从 Fork 开始？

### 1.1 Fork 的概念

**Fork**（派生）= 在平台上把别人的仓库**服务端复制**一份到你自己的账号下。

它不是简单的下载：

| | `git clone`（本地复制） | **Fork**（平台上的复制） |
|---|---|---|
| 复制到哪 | 你的电脑 | 你的 GitCode 账号下（也在云端） |
| 得到什么 | 本地仓库，远程仍指向**别人的**仓库 | 一个**属于你自己的**云端仓库 |
| 能否直接 push | ❌ 不能（无权限） | ✅ 能（你是所有者） |
| 与原仓库的关系 | 无 | 平台会记住它派生自哪个仓库 |

一句话：**Fork 之后，你在云端有了一个完全归自己管的仓库副本**——可以随意修改、推送，而不必征得原作者同意。

```
① Fork 操作（一次性）：
   cann/repo ──复制──▶ yourname/repo        （两个都在 GitCode 云端）

② 之后，你的修改走过的路：

   你的电脑            你的 fork (origin)         原仓库 (upstream)
   （本地仓库）          yourname/repo              cann/repo
       │                     │                        │
       │ git push            │ Pull Request           │
       │ （你随时可推）        │ （请求维护者合入）        │
       └──────────▶──────────┴──────────▶──────────────┘
                        ↑ 你可随意推送        ↑ 只有维护者能合入
```

- Fork 复制的是**云端**仓库，和你的电脑无关；之后你随时 `git clone` 你的 fork 到本地
- 你 `push` 的目标永远是**你的 fork**（origin），不是原仓库
- 原仓库（upstream）你只有读权限，修改只能通过 **Pull Request** 请求维护者合入

### 1.2 为什么不直接 clone 原仓库开发？

设想你想给 `cann/cann-learning-hub` 修一个错别字：

- **直接 clone**：修改只能存在你的电脑上——你没有原仓库的推送权限，改动永远到不了云端，别人也看不见
- **先 Fork 再 clone**：推送到**自己的 fork**（当然有权限），然后发 Pull Request **请求**维护者把你的修改合入原仓库

| 情况 | 直接 clone 原仓库 | Fork 后 clone |
|------|-------------------|----------------|
| 你有写权限（自己是维护者） | ✅ 可以直接 push | 不需要 Fork |
| 你没有写权限 | ❌ 无法 push，改动出不了你的电脑 | ✅ push 到自己的 fork |
| 贡献修改 | ❌ 无途径 | ✅ 通过 PR 请求合入 |

> 绝大多数开源贡献场景：你**没有**原仓库的写权限，所以必须 Fork。
>
> 💡 Fork 还有一个天然好处：原项目停更甚至删除时，你的 fork 仍是一份完整存档，随时可以继续独立演进。

### 1.3 Fork + PR 工作流全景

一次完整的贡献，就是你的修改从「你的电脑」一路走到「原仓库」的旅程，共 6 步：

```
第 1 步           第 2 步            第 3 步
  Fork     ──▶   clone 到本地  ──▶   修改 + add + commit
（网页点一下）    （git clone）      （git add → git commit）

第 4 步           第 5 步            第 6 步
  push     ──▶   发起 Pull Request ──▶  维护者 review
（git push）     （网页点一下）       （通过则合入原仓库）
```

按时间顺序走一遍：

1. **Fork**（网页）：在你的 GitCode 账号下得到一个云端副本
2. **clone**（命令行）：把**你的 fork**（不是原仓库！）下载到本地电脑
3. **修改 + add + commit**（命令行）：在功能分支上改代码，`git add` 暂存、`git commit` 提交——第 2、3 课的本地操作在这里上场
4. **push**（命令行）：把提交推到**你的 fork**（origin）
5. **发起 PR**（网页）：请求维护者把你的分支合入原仓库（upstream）
6. **review + 合入**（网页）：维护者审查、提意见；你按需修改后再次 push（PR 自动更新），最终合入

> 💡 注意操作场所的切换：**①⑤⑥ 在网页上点，②③④ 在终端里敲**。
>
> 三个仓库各记一个关键词：
> - **本地仓库** —— 改代码的地方（`commit`）
> - **你的 fork（origin）** —— 你的云端草稿区（`push` 的目标）
> - **原仓库（upstream）** —— 别人的地盘（只能通过 PR 请求合入）

下面我们逐步实现这个流程。

## 4. 第三步：配置 upstream 远程（关键！）

你的 fork 是原仓库的快照，但原仓库会持续更新。为了同步原仓库的更新，需要添加 **upstream** 远程。

> 这是 Fork 工作流中**最容易被新手遗漏**的一步，但它至关重要。

````python
# 添加 upstream 远程，指向原始仓库（已存在则跳过，重复执行不报错）
import subprocess
remotes = subprocess.run(["git", "remote"], capture_output=True, text=True).stdout.split()
if "upstream" in remotes:
    print("upstream 已存在，跳过添加")
else:
    !git remote add upstream https://gitcode.com/cann/cann-learning-hub.git

# 验证：现在应该有 origin 和 upstream 两个远程
!git remote -v
````

现在你有两个远程：

```
origin    → gitcode.com/your_username/cann-learning-hub  (你的 fork，可 push)
upstream  → gitcode.com/cann/cann-learning-hub            (原仓库，只 fetch)
```

### 同步上游更新

在开始开发前，先同步 upstream 的最新代码：

````python
# 定位练习仓库（已在仓库内 / 进入克隆目录 / 自动补克隆，下同）
import os, subprocess
repo = "cann-learning-hub"
if not os.path.isdir(".git"):            # 已在仓库内则直接用
    if not os.path.isdir(repo):          # 否则进入克隆目录
        print(f"未检测到 {repo}/，自动克隆原仓库用于演练……")
        print("（正式贡献请先 Fork 再克隆你自己的 fork，见 3.2 节）")
        subprocess.run(["git", "clone", "--depth", "1",
                        "https://gitcode.com/cann/cann-learning-hub.git", repo], check=True)
    os.chdir(repo)

# 1. 拉取 upstream 的最新代码（演练克隆未配置 upstream 时自动跳过）
!git fetch upstream || echo "（未配置 upstream 远程，跳过——正式流程见第 4 节）"

# 刚 clone 的仓库与远端一致，下面 2~4 步暂时不需要；
# 之后再贡献时（本地落后于远端），取消注释执行：
# 2. git checkout master
# 3. git merge upstream/master
# 4. git push origin master     # 让 fork 的 master 也保持最新

print("fetch 完成！")
````

## 5. 第四步：创建功能分支

**永远不要在 master 分支上直接开发！** 每次贡献都应创建独立的功能分支。

### 5.1 分支命名规范

```bash
feature/add-git-tutorial      # 新增功能
fix/readme-typo               # 修复 bug
docs/update-install-guide     # 文档改进
```

### 5.2 创建并切换到功能分支

````python
# 确保在最新的 master 上创建分支
# git checkout master
# git pull upstream master      # 确保最新

# 定位练习仓库，保证命令落在仓库里而不是 notebook 所在目录
import os, subprocess
repo = "cann-learning-hub"
if not os.path.isdir(".git"):            # 已在仓库内则直接用
    if not os.path.isdir(repo):          # 否则进入克隆目录
        print(f"未检测到 {repo}/，自动克隆原仓库用于演练……")
        print("（正式贡献请先 Fork 再克隆你自己的 fork，见 3.2 节）")
        subprocess.run(["git", "clone", "--depth", "1",
                        "https://gitcode.com/cann/cann-learning-hub.git", repo], check=True)
    os.chdir(repo)

# 创建并切换到功能分支
!git checkout -b feature/add-git-tutorial

# 确认当前分支
!git branch
````

## 9. 第八步：应对 Code Review

提交 PR 后，维护者会进行 Code Review。他们可能提出修改建议、问题或直接 approve。

### 9.1 Reviewer 要求修改怎么办？

不要关闭 PR 重开！在**同一个分支**上继续修改并推送，PR 会自动更新。

```bash
# 1. 在本地功能分支上修改
git checkout feature/add-git-tutorial

# 2. 根据评审意见修改代码
# ... 编辑文件 ...

# 3. 提交修改
git add <files>
git commit -m "docs: 根据评审意见调整内容"

# 4. 推送到你的 fork（PR 会自动更新！）
git push origin feature/add-git-tutorial
```

> 💡 PR 与分支是绑定的——只要往同一个分支 push 新提交，PR 就会自动包含这些新提交。

### 9.2 处理冲突：上游有新提交

如果在你提交 PR 期间，原仓库的 master 有了新提交，你的 PR 可能产生冲突。解决方法：

```bash
# 1. 拉取 upstream 最新代码
git fetch upstream

# 2. 在功能分支上 rebase 到最新的 upstream/master
git checkout feature/add-git-tutorial
git rebase upstream/master

# 3. 如果有冲突，解决后继续
#    编辑冲突文件 → git add → git rebase --continue

# 4. 强制推送到 fork（rebase 改写了历史，需要强制推送）
git push --force-with-lease origin feature/add-git-tutorial
```

> ⚠️ 使用 `--force-with-lease` 而非 `--force`，前者更安全（会检查远程是否有你不知道的更新）。

### 9.3 评审通过

当 Reviewer approve 后，通常由维护者执行合并。部分项目也允许贡献者自行合并（看仓库权限设置）。

## 10. 第九步：PR 合并与后续清理

### 10.1 合并方式

GitCode 通常提供三种合并方式：

| 方式 | 说明 | 适用场景 |
|------|------|----------|
| **Merge** | 创建合并提交，保留完整历史 | 一般情况（最常用） |
| **Squash & Merge** | 把多个提交压缩成一个 | 功能分支提交太多太杂时 |
| **Rebase & Merge** | 线性合入，无合并提交 | 追求线性历史 |

### 10.2 合并后清理

PR 合并后，功能分支已完成使命，可以清理：

````python
# 1. 同步本地 master 到最新
# git checkout master
# git pull upstream master

# 2. 删除本地功能分支
# git branch -d feature/add-git-tutorial
#   注：若 PR 以 Squash & Merge 合入，-d 会提示"未合并"（提交被压缩了），改用 -D 即可

# 3. 删除 fork 上的功能分支
# git push origin --delete feature/add-git-tutorial

print("清理完成！准备开始下一个贡献 🎉")
````

## 12. 完整流程速查表

### Linux 命令行完整流程

```bash
# 0. 网页上 Fork 仓库（只需一次）

# 1. Clone 你的 fork
git clone git@gitcode.com:<你的用户名>/repo.git
cd repo

# 2. 添加 upstream
git remote add upstream https://gitcode.com/<原组织>/repo.git

# 3. 同步上游
git fetch upstream
git checkout master
git merge upstream/master
git push origin master

# 4. 创建功能分支
git checkout -b feature/your-feature

# 5. 修改并提交
git add <files>
git commit -m "feat: 你的修改说明"

# 6. 推送到 fork
git push -u origin feature/your-feature

# 7. 网页上发起 Pull Request

# 8. 根据 Review 修改（如需要）
git add <files>
git commit -m "fix: 根据评审意见修改"
git push origin feature/your-feature

# 9. PR 合并后清理
git checkout master
git pull upstream master
git branch -d feature/your-feature
git push origin --delete feature/your-feature
```

```

5. 网页查看 Review → 网页继续编辑修改 → 提交

```

## 13. 常见问题

### Q1: push 时报错 “Permission denied”

你可能在推送到 upstream（原仓库）而非 origin（你的 fork）。检查：
```bash
git remote -v
```
确保 push 到 `origin`。

### Q2: PR 显示 “有冲突”

上游有了新提交，与你的修改冲突。在本地 rebase：
```bash
git fetch upstream
git rebase upstream/master
# 解决冲突 → git add → git rebase --continue
git push --force-with-lease origin feature/your-feature
```

### Q3: 提交了敏感信息怎么办？

如果还没 push：
```bash
git reset HEAD~1 --hard   # 撤销最后一次提交
```

如果已经 push 但 PR 还没合并：
- 用 `git rebase -i` 删除包含敏感信息的提交
- `git push --force-with-lease` 更新 PR
- **并立即在平台轮换相关密钥**

如果已经合并到 master：
- 联系维护者处理
- **必须立即轮换泄露的密钥**

### Q4: 需要签署 CLA？

部分项目（如 CANN）要求贡献者签署 CLA（贡献者许可协议）。首次提交 PR 时，会有 bot 自动检查并引导你完成签署。详见 [贡献指南](../../CONTRIBUTION.md)。

## 14. 总结

恭喜！你已经完整了解了 GitCode 平台的 Fork + PR 工作流。

**核心要点回顾：**

1. **Fork** 是贡献开源的第一步（当你没有写权限时）
2. **origin = 你的 fork**（可 push），**upstream = 原仓库**（只 fetch）
3. **永远在功能分支上开发**，不要直接改 master
4. **PR 与分支绑定**，往同一分支 push 即可更新 PR
5. **应对 Review** 在原分支继续修改 push，不要关闭重开 PR
6. **rebase** 解决冲突后用 `--force-with-lease` 推送
7. 合并后及时**清理**功能分支
