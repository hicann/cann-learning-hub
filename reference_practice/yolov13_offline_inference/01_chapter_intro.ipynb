# 1 章节介绍

YOLO（You Only Look Once）是一类单阶段目标检测模型。它把图片一次送入网络，同时预测目标的位置和类别，适合处理摄像头画面和道路图片。YOLOv13 是 YOLO 系列的一个版本，本案例使用它识别 COCO 类别中的行人和车辆。

在昇腾设备上，模型通常先由 ONNX 转成面向具体芯片的 OM，再由 ACL 加载和执行。本案例把模型转换、图片预处理、C++ 推理和结果后处理串起来，最后还会比较不同多流配置的执行时延。各章的 Code Cell 按顺序使用前一章生成的文件。

## 学习目标
- 了解 ONNX 到 OM 的离线模型转换流程。
- 使用 ACL C++ API 加载 OM 并执行推理，统计识别到的行人和车辆数量。
- 理解 letterbox 预处理、YOLO 输出解析和 NMS。
- 使用平均 `aclmdlExecute` 时延比较多流编译参数。
