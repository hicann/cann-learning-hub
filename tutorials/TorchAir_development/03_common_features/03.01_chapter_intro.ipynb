# npugraph_ex 基础功能
本章承接第二章快速入门内容，介绍 `npugraph_ex` 的常用性能优化能力，包括原地算子优化、图内存复用和 FX Pass。

---

## 前置要求

- 已完成第二章“npugraph_ex 快速入门”。
- 熟悉 `torch.compile`、`FX Graph`和 `torch_npu` 的基本概念。
- 已参照[版本配套表](https://gitcode.com/cann/cann-learning-hub/blob/test/tutorials/TorchAir_development/README.md#torchair-%E7%89%88%E6%9C%AC%E9%85%8D%E5%A5%97%E8%A1%A8)配置相互匹配的 Python、CANN、PyTorch 和 torch_npu。

> `npugraph_ex` 主要面向在线推理场景，暂不支持反向流程和随机数算子的 Capture，其 Capture-Replay 约束与 `torch.cuda.CUDAGraph` 类似。

---

## 章节目标

完成本章后，你将能够：

- 根据问题定位或性能优化目标选择合适的 `npugraph_ex` 选项；
- 区分输入原地化、输入克隆和图内存池复用的作用范围与代价；
- 通过 FX 图、Profiler 和精度基线验证优化是否真正生效。
