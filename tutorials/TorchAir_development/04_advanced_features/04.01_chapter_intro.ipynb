# npugraph_ex 深度优化
本章承接第三章基础功能内容，面向在线推理和大模型部署场景，介绍模型编译缓存、多流表达、AI Core 和 Vector Core 限核、静态 Kernel 编译以及 SuperKernel 功能。

---

## 前置要求

- 已完成第三章“npugraph_ex 基础功能”。
- 熟悉 NPU 图编译、多流异步执行和算力资源分配的基本概念。
- 已参照[版本配套表](https://gitcode.com/cann/cann-learning-hub/blob/test/tutorials/TorchAir_development/README.md#torchair-%E7%89%88%E6%9C%AC%E9%85%8D%E5%A5%97%E8%A1%A8)配置相互匹配的 CANN、PyTorch、torch_npu 和 TorchAir。

> 本章功能继承 `npugraph_ex` 的整体约束：主要面向推理场景，每个进程仅支持使用 1 张 NPU 卡，暂不支持反向流程和随机数算子的 Capture。各高级功能还可能有额外的产品型号和版本限制。

---

## 章节目标

完成本章后，你将能够：

- 使用 `cache_compile` 缩短模型冷启动时延，并判断缓存何时失效；
- 使用 Stream、Event 和 `record_stream` 正确表达计算并行与计算—通信并行；
- 使用 `npugraph_ex` 进行 Stream 级核数配置，按实际情况配置最大 AI Core 数和 Vector Core 数；
- 说明 SuperKernel 与静态 Kernel 编译的依赖关系，并完成融合结果验证。
