# npugraph_ex 实践与问题定位
本章面向已经掌握 `npugraph_ex` 基础配置和性能优化的开发者，介绍图编译 Debug 信息保存、算子 Data-Dump、多流并发死锁检测和常见问题定位方法。

## 学习目标

完成本章后，你将能够：

- 保存并归档图编译阶段的中间信息；
- 使用 Data-Dump 产物定位算子输入、输出和精度异常；
- 识别多流并发中的等待环和潜在死锁；
- 按照“现象—最小复现—分层开关—产物比对”的流程定位问题。

> 本章示例需要在已安装匹配版本 CANN、PyTorch、`torch_npu` 和 TorchAir 的昇腾 NPU 环境中运行。调试开关只建议在问题复现和分析阶段启用。
