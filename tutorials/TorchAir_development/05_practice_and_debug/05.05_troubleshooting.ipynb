# 5.5 常见问题定位方法

## 1. Plog 日志

Plog 日志获取参考：[日志参考](https://www.hiascend.com/document/detail/zh/CANNCommunityEdition/900beta2/maintenref/logreference/logreference_0008.html)

```python
#开启打屏
export ASCEND_SLOG_PRINT_TO_STDOUT=1
#日志级别改为 DEBUG
export ASCEND_GLOBAL_LOG_LEVEL=0
#提高日志上限
export ASCEND_HOST_LOG_FILE_NUM=1000
#默认日志路径
/root/ascend/log
```

## 2. TORCH_COMPILE_DEBUG

在环境变量中添加：

export TORCH_COMPILE_DEBUG=1

或者在脚本中添加：

```python
import os
os.environ["TORCH_COMPILE_DEBUG"] = "1"
```
开启后将 dump 出 npugraph_ex 每个图优化 pass 修改后的 FX 图，以及 Torch Dynamo 组件和 npugraph_ex 的 debug 日志。详细介绍：[图编译完整Debug信息保存功能](https://www.hiascend.com/document/detail/zh/Pytorch/730/modthirdparty/torchairuseguide/torchair_00014.html)
