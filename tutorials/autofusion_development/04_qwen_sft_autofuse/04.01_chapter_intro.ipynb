# Qwen3-1.7B SFT 与 AutoFuse 对比实践

本章以 Qwen3-1.7B SFT 为例，帮助你理解如何在 Ascend NPU 环境中完成一次短训练，并比较 AutoFuse 关闭和开启时的训练表现。

## 前置要求

开始本章节前，需要的环境准备：

- 至少两张可用的 Ascend NPU
- 至少30 GiB 空闲磁盘和 ；

CANNLab 上已经预安装好 CANN 包，版本可手动升级到最新。

## 章节目标
完成本章后，你将能够：
* 了解 Qwen3-1.7B SFT 基础训练流程，理解 AutoFuse 的作用以及 AutoFuse OFF/ON 两种运行模式的区别。
* 完成 AutoFuse OFF/ON 短训练实验，查看训练结果并通过耗时对比分析 AutoFuse 对训练性能的影响。
