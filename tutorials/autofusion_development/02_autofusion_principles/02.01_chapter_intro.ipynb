# AutoFuse 自动融合原理

本章在前序内容基础上，聚焦昇腾自研 GE 框架路径中 Lowering 后的融合策略，逐步介绍融合条件判断、融合策略求解，以及融合顺序和多轮融合对最终结果的影响。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 已完成前序内容学习，理解 AutoFuse 主要在图编译阶段发挥作用。
- 理解算子融合通过减少中间数据搬运和调度开销获得收益。
- 了解 AscBackend、AscGraph、Schedule 和 Lowering 的基本含义。
- 能够阅读简单的有向无环计算图，并理解节点之间的数据依赖关系。

---

## 章节目标

完成本章后，你将能够：

- 说明 CanFuse 框架与 Backend 分别判断哪些融合条件。
- 区分垂直融合与水平融合，判断两个节点是否具备减少内存读写的基础。
- 解释成环检测、最大融合个数和内存峰值评估的必要性。
- 解释如何依次比较节省的内存大小、节点临近性和 topo 序，确定可融合节点对的处理顺序。
- 分析不同融合顺序为什么可能产生不同结果，理解多轮融合的处理方式及贪心算法的局限性。
