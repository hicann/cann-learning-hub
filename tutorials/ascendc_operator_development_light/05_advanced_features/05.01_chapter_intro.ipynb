# Ascend C算子进阶特性

本章聚焦Ascend C算子在完成基础开发与接入之后的执行效率优化能力，围绕Aclgraph入图编译与运行展开。通过本章学习，你将理解Host侧重复下发开销的来源，掌握捕获、实例化与重放的执行方式，并了解自定义算子经Pybind11或Torch Library接入后被捕获和重放的过程。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 已完成Ascend C基础课程学习，理解核函数、Host侧调用、CMake编译和Add算子样例的基本结构。
- 已掌握PyTorch框架下Kernel直调的接入方式，了解Pybind11与Torch Library两种算子注册入口。
---

## 章节目标
完成本章后，你将能够：

- 理解Aclgraph减少Host侧重复下发开销的基本原因，以及捕获、实例化、重放三个阶段的作用。
- 区分Aclgraph底层能力与NPUGraph框架封装之间的关系。
- 使用NPUGraph API显式控制捕获与重放，或通过 `torch.compile` 后端从模型入口启用相关能力。
- 说明Ascend C自定义算子通过Pybind11或Torch Library接入后如何被捕获和重放。
