# 框架集成

本章介绍如何将 Ascend C Kernel 集成到 PyTorch 框架中，并从 Python 侧完成编译产物加载、算子调用与结果验证。课程以同一个 Ascend C Kernel 为基础，对比 pybind11 和 torch.library 两种调用方式，帮助你理解 PyTorch Tensor、C++ 封装层、NPU Stream 与 Device 侧 Kernel 之间的衔接关系。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 完成前四章学习，理解 Ascend C 算子的 Host 侧与 Device 侧开发流程，并能够阅读基本的 Ascend C Kernel。
- 具备基础 C/C++ 与 Python 编程能力，能够阅读函数封装、动态库加载和简单的 CMake 构建脚本。
- 了解 PyTorch Tensor、设备间数据搬运和自定义算子调用等基本概念。

在环境配置方面，本章课程建议你的环境满足以下条件：

- 系统中已部署昇腾 NPU 硬件，并能够运行 PyTorch NPU 程序。
- 已按照 [CANN 下载页面](https://www.hiascend.com/cann/download)完成对应硬件的开发环境部署。
- 已安装版本相互匹配的 PyTorch、torch_npu 和 pybind11，具体要求请参考课程中的环境准备说明。

---

## 章节目标

完成本章学习后，你将能够：

- 理解 Ascend C Kernel 接入 PyTorch 的调用链路，能够使用 pybind11 和 torch.library 完成自定义算子的封装、编译、注册、加载、调用与结果验证。
