# 训练后端切换：从 FSDP 到 TorchTitan-NPU FSDP2

前四章已经搭建好 Wordle 强化学习流程。本章沿用相同的模型、数据、算法和交互逻辑，只把 Actor 与 Reference Model 的训练后端从 FSDP 换成 TorchTitan-NPU FSDP2。切换后，训练流程便可以使用 TorchTitan-NPU 的并行、内存管理和 NPU 模型适配能力，并可根据实际负载继续调优。

---


## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 已完成前四章学习，并能够说明 Wordle 强化学习的完整训练流程。
- 理解 Actor、Reference Model 与 vLLM rollout 在强化学习训练中的职责。
- 了解 GRPO、Wordle AgentLoop 和奖励函数之间的数据关系。

---

## 章节目标

完成本章后，你将能够：

- 找出训练后端切换涉及的模块和配置。
- 区分保持不变的 RL 训练语义与发生变化的执行机制。
- 将 FSDP 配置映射到 TorchTitan-NPU 配置。
- 说明 Actor 更新后的权重如何同步给 vLLM。
