## 判断题

1. （判断题）entropy_coeff 需要通过短程实验观察训练曲线后再进行调节。

2. （判断题）entropy_coeff 越大越好，可以充分保证模型的探索能力。

3. （判断题）RL 的学习率通常比 SFT 低 1-2 个数量级。

4. （判断题）判断训练稳定性需要同时观察 entropy、KL、验证 correct 和输出长度。

5. （判断题）KL loss 和 entropy bonus 是互补机制：前者限制相对参考策略的漂移，后者降低策略过早集中的风险。

6. （判断题）训练监控时只需要看 reward 一个指标即可判断训练健康度。

7. （判断题）前 25～30 步的短程实验有助于发现 entropy、KL 和输出长度的异常趋势。
