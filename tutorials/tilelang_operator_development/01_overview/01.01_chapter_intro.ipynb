# 1.1 章节介绍

## 1. 这门课能帮你做什么

这门课介绍如何使用 **TileLang，在昇腾950（A5）上编写自定义算子**。算子是完成特定计算的程序，例如把两组数按对应位置相加。

课程从这个加法例子出发，逐步带你练习三件事：

- **编写算子**：把计算拆成适合硬件处理的数据块，用TileLang描述计算过程。
- **验证结果**：将算子输出与PyTorch给出的参考结果比较，检查计算是否正确。
- **测量耗时**：使用性能工具获取执行时间，观察代码调整带来的变化。


## 2. 开始前需要什么基础

**不要求算子开发经验，但需要具备以下基础：**

- **Python编程基础。** 熟悉变量、循环和函数，能阅读简单的装饰器和类型注解。
- **昇腾950基本架构知识。** 了解Cube与Vector的主要分工，认识GM、UB等常用存储位置。[1.3](01.03_ascend_runtime.ipynb)会结合本章示例回顾这些概念。

使用过PyTorch张量，会更容易理解示例中的输入、输出和结果校验。
