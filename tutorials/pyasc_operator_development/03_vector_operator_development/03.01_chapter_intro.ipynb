# Vector算子开发

本章以Vector（矢量）算子为主线，系统讲解一个Vector算子从计算模型到工程落地的完整知识链路。第2章我们已经学习了pyasc核函数开发的基础——编程范式、`@asc.jit`装饰器、Tensor与数据搬运、手动流水同步、核函数的Launch调用与结果验证。本章在此基础上，依次讲解Vector算子的SIMD计算模型与典型场景、数据切分（Tiling）策略，以及基于`TPipe`/`TQue`框架的设计→实现→编译→验证四阶段开发流程，最终独立开发Add、ReLU等典型Vector算子。

### 学习前置要求

学习本章前，建议已经具备以下基础：

- 完成第2章学习，掌握`@asc.jit`装饰器、`GlobalTensor`/`LocalTensor`与`asc.data_copy`数据搬运、手动流水同步，以及核函数的Launch调用与结果验证；
- 理解AI Core的存储层级（Global Memory、Unified Buffer）和SPMD并行模型；
- 系统中已安装pyasc（v1.1.0及以上）和CANN软件（8.5.0及以上）。

### 学习目标

完成本章后，开发者应能够：

1. 理解SPMD与SIMD的嵌套并行关系，理解Vector算子的SIMD计算模型和element-wise典型计算场景；
2. 掌握Tiling两级切分（多核间/核内）的术语与组合关系，理解数据切分三原则及双缓冲对切分的影响；
3. 掌握Vector算子设计→实现→编译→验证的四阶段开发流程，能基于`TPipe`/`TQue`框架实现CopyIn→Compute→CopyOut三段式核函数，并用torch验证结果；
4. 独立完成Add、ReLU等典型Vector算子的开发与验证。
