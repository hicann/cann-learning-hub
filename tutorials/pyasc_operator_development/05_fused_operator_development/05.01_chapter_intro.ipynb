# 融合算子开发

本章围绕 PyAsc CV 融合算子的设计与实现展开。课程从 AIC/AIV 分离架构入手，说明 Cube 与 Vector 计算如何在一个 Kernel 中协作，并以 MatmulLeakyRelu 样例贯穿融合算子的分片计算、LocalTensor 管理、结果写回与正确性验证。

## 前置要求

- **已具备能力**：理解 Host-Device 异构编程、GlobalTensor 与 LocalTensor，以及 TPipe/TQue 的基本用法。
- **已学前置教程**：已完成第 2 章“PyAsc 核函数开发基础”、第 3 章“Vector 算子开发”和第 4 章“Matmul 算子开发”。
- **环境要求**：已安装 PyAsc 与配套 CANN 软件，并具备算子运行环境。

## 章节目标

完成本章后，你将能够：

- 说明 AIC/AIV 分离架构与 CV 融合的关系。
- 判断 Cube 与 Vector 计算是否具备 CV 融合所需的数据依赖和分片条件。
- 使用 PyAsc Matmul 高阶 API 和 Vector API 实现 CV 融合。
- 管理 Matmul 输出分片的 LocalTensor 生命周期和分块写回。
- 使用独立参考表达式验证 Matmul + Activation 融合结果。

## 章节内容

- [5.2 CV 架构与融合算子](05.02_fusion_value_and_boundaries.ipynb)：从 AIC/AIV 分离架构理解 CV 融合的数据流、价值与适用条件。
- [5.3 基于 MatmulLeakyRelu 样例开发 CV 融合算子](05.03_common_fusion_patterns.ipynb)：掌握 Matmul 输出配置、分片激活、写回与验证流程。
- [5.4 章节实践：实现 Matmul + Activation CV 融合算子](05.04_chapter_practice.ipynb)：在 `matmul.iterate()` 内补全 Matmul 分片获取、Vector 激活和队列操作。
