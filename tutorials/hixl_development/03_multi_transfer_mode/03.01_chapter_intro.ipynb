## 传输模式选择指南

选择传输模式时，应同时考虑硬件、内存注册能力、传输接口和数据路径：

1. **FabricMem**：仅位于 Atlas A3 超节点内、满足 FabricMem 软件依赖时可用，主要面向 HCCS D2rH/rH2D、Host 内存池化或高带宽传输场景；支持同步和异步传输，但同批 `op_descs` 的传输类型必须一致。
2. **中转 Buffer**：任一端存在无法或不宜注册的 Host 内存，并且只需要同步传输时选择。未注册地址按 Host 内存处理，用户必须保证地址合法；该模式不支持异步传输，同批 `op_descs` 的传输类型必须一致，并会增加中转拷贝。
3. **直传**：不满足 FabricMem 条件时，如果本地和远端内存可注册，且业务需要同步或异步传输并希望减少额外拷贝，默认选择直传模式。

| 判断条件 | 直传模式 | 中转 Buffer 模式 | FabricMem 模式 |
| --- | --- | --- | --- |
| 硬件范围 | Atlas A2/A3 | Atlas A2/A3 | 仅 Atlas A3 超节点内 |
| 内存要求 | 本地和远端内存需满足注册要求 | 任一端可包含无法或不宜注册的 Host 内存；未注册地址必须合法 | 满足软件依赖并使用 VMM 统一编址的 Host 内存 |
| 传输接口 | 支持同步和异步 | 不支持异步，仅同步；同批 `op_descs` 传输类型必须一致 | 支持同步和异步；同批 `op_descs` 传输类型必须一致 |
| 数据路径 | 注册内存之间直接传输 | 经两端 Buffer 中转 | 通过 HCCS 执行 D2rH/rH2D 传输 |
