# PD 分离与 KV Cache 传输实战

本章聚焦于大模型推理场景下的 Prefill/Decode（PD）分离架构，讲解如何使用 HIXL llm-datadist 接口实现 P→D 的 KV Cache 传输，并介绍 llm-datadist 在 omni-infer 推理框架中的应用，帮助你掌握从底层传输 API 到上层框架调用的完整知识链路。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 完成第一章学习，理解单边通信的基本原理与 HIXL 的执行流程。
- 完成第二章学习，掌握基于 HIXL 核心 API（Initialize、RegisterMem、Connect、Transfer）的使用方法。
- 完成第三章学习，了解 HIXL 的多种传输模式（直传、中转 Buffer、FabricMem）及其适用场景。
- 了解大模型推理基本流程（Prefill 阶段和 Decode 阶段的区别）。

---

## 章节目标

完成本课程后，你将能够：

- 理解 PD 分离架构的动机、原理与核心优势。
- 掌握 HIXL llm-datadist 的 C++ 和 Python 接口体系，包括角色定义、集群链接、KV Cache 注册与传输。
- 了解 llm-datadist 在 omni-infer 推理框架中的应用。
- 能够使用 Python API 完成 P→D 的 KV Cache 传输实践。
