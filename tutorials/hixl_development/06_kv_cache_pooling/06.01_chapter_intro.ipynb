# KV Cache 池化技术介绍

本章聚焦于大模型推理场景下的 KV Cache 池化技术，讲解如何使用 Mooncake Store 实现跨节点的 KV Cache 共享与缓存，帮助你理解从单节点 KV Cache 管理到分布式 KV Cache 池化的技术演进，并掌握 Mooncake Store 在昇腾 NPU 上的安装与使用方法。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力：

- 完成第一章至第三章学习，掌握 HIXL 核心 API 和多种传输模式。
- 完成第五章学习，理解 PD 分离架构和 llm-datadist 的 KV Cache 传输机制。
- 了解分布式存储系统的基本概念（如 RDMA、内存池化）。

---

## 章节目标

完成本课程后，你将能够：

- 理解 KV Cache 池化技术的动机、原理与核心价值。
- 掌握 Mooncake Store 的架构设计与核心组件。
- 了解 Mooncake Store 在昇腾 NPU 上的三种传输链路及适用场景。
- 掌握 Mooncake Store 的安装、部署模式与 Python API 使用方法。
- 能够使用 Mooncake Store 实现跨节点 KV Cache 共享实践。
