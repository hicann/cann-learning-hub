# 图编译：怎么编译我的图

> 一级章节：进阶开发 ｜ 二级模块：图编译

本模块围绕 GE 图编译全流程展开：从 AscendIR 构图与 Parser 解析输入，到编译过程配置、编译产物形态，再到自定义算子入图与自定义融合 Pass 两类图编译扩展能力，帮助你掌握把一张图正确、高效地编译为 OM 的完整方法。

---

## 章节目标
完成本章后，你将能够：

- 使用 AscendIR API 或 Parser 构建/导入计算图
- 理解融合、精度、buffer、stream 等关键编译配置的作用与取舍
- 理解 OM 产物结构，掌握外置权重、SO in OM、模型缓存等产物配置
- 掌握自定义算子入图的完整链路
- 掌握自定义融合 Pass 的实现与验证
