## 6.1 为什么最后一公里最适合自动化

提交 PR/Issue 这类协作工作有三个特征：**规则明确**（约定式提交、模板字段——对错可判定）、**可验证**（API 回查）、**低创造性**（内容应从代码 diff **推导**，而非「创作」）。三者齐备的任务正是 Agent 的主场——cannbot-skills 仓 `infra/` 域的协作 Skill 家族（gitcode-toolkit / pr-handler / issue-gen / issue-handler）就是按这个判断设计的。

**红线是安全工程**——每条对应一类真实事故：

## 结课小结

- **最后一公里为什么适合自动化**：规则明确 + 可验证 + 低创造性——PR 从 diff 推导而非创作；
