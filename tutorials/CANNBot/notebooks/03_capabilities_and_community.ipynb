```text
我的算子跑出来全是 0，输入正常，float16，帮我看下可能是什么问题
```

## 3.2 性能优化：先测量，再动手

「我觉得这个算子慢」不是优化依据。CANNBot 的性能路线遵循工程原则——**先测量、再定位、后优化**：

```text
采集数据（ops-profiling：msprof 采集 + CSV 指标解读）
   → 定瓶颈：计算瓶颈 / 访存瓶颈 / 同步瓶颈
   → 给方案：Tiling 调整 / 指令换用 / 流水重排
   → 复测验证：优化真实生效
```

你只要说「帮我采集这个算子的性能数据，分析瓶颈并给出优化建议」。**没有数据支撑的优化建议，都不该被采纳**。

> 📖 这套方法的完整闭环（采集 → 定位 → 策略 → 实施 → 回归）将在[进阶篇第 5 册](./05_performance_optimization.ipynb)用 16 个真实用例的数据完整走一遍。

运行代码，把「需求 → 推荐路径」做成查询工具，直观感受选择逻辑：

</div>

````python
def recommend_path(need):
    """根据需求关键词推荐开发路径（演示用，实际选择还请综合评估）。"""
    rules = [
        ("入框架/正式交付/ACLNN", "Ascend C 注册调用", "标准算子工程，支持多芯片架构"),
        ("已有 Triton 代码/GPU 迁移", "Triton", "triton_ascend DSL，迁移成本低"),
        ("Python 习惯/不想写 C++", "PyPTO", "昇腾原生 Python Tile 编程"),
        ("DSL/TileLang 经验", "TileLang", "Developer/Expert 双模式"),
        ("快速验证/原型/学习", "Ascend C 直调", "`<<<>>>` 核函数直调，最直接"),
    ]
    for keys, path, why in rules:
        if any(k in need for k in keys.split("/")):
            return path, why
    return "Ascend C 直调", "默认推荐：入门友好、验证直接"

for need in ["我想快速验证一个想法", "要把算子交付到框架里用",
             "我有现成的 Triton kernel", "我只会 Python"]:
    path, why = recommend_path(need)
    print(f"需求：{need}")
    print(f"  → 推荐路径：{path}（{why}）")
    print()
````

算子之外还有更大的场景值得一提：**模型推理优化**（从 HF 模型适配到 NPU 部署的端到端优化：并行策略、KVCache、融合算子替换、量化适配等，`model-infer-*` 系列 15 款 Skill）和**训练精度诊断**（loss 异常、NaN/Inf、两次运行不一致的标准化定位）。这两类场景链路长、要素多，恰好是多 Agent + 领域 Skill 的主场——感兴趣的同学从功能清单深入。

## 3.4 cann-bench：用评测说话

「Agent 生成的算子到底行不行？」——这不能靠体感回答。cann-bench 是 CANN 社区的**算子能力公共标尺**，几个关键设计：

- **53 个算子**，按 L1–L4 难度分级，覆盖不同硬件单元的使用；
- **双集防背题**：公开用例（每任务 20 条，沿规模/对齐/维度/值域四轴设计）＋ 隐藏用例（留在榜单服务端，**不开源是刻意的**——防止针对测试集「应试」）；
- **两道布尔门 + 连续分**：先过编译门、精度门，再按性能给分；性能分以硬件理论峰值为上界（HAP）推导——**靠实现技巧刷不动分**；
- **[cannbench.com](https://cannbench.com) 排行榜**：53 个算子全覆盖，CANNBot 系列方案目前居首。

对你的两种用法：**验证自己的开发能力**（拿公开用例练手，看能不能过隐藏集的精度门）；**度量 Agent 的真实水平**（让 CANNBot 跑评测任务，用独立标尺检验生成质量）——这正好呼应第 1 册「生成结果必须独立验证」的原则。
