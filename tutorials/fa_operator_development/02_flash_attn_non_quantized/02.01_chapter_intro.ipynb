# 第2章 非量化Flash Attention算子开发

本章聚焦**非量化**Flash Attention算子，即输入Q、K、V均为FP16或BF16的高精度浮点类型。本章首先介绍非量化Flash Attention的概念、算子规格与计算流程，帮助开发者建立对非量化注意力算子的整体认识，为后续的具体实现章节打好基础。

> 本章不重复介绍Flash Attention的基础原理（注意力公式 `O = softmax(Q×K^T×scale)×V`、分块计算与在线Softmax的思想）。如需补充背景知识，请先阅读Flash Attention相关论文或前置资料。

---

## 前置要求

为了充分掌握本章内容，你应已具备以下能力:

- 完成第一章学习，理解算子的核心概念与基本原理。
- 完成第二章学习，理解Ascend C算子开发基础知识。
- 完成第三章学习，掌握如何使用算子工程方式开发算子。
- 完成第四章学习，掌握基于高阶API的矩阵乘算子开发方法。

在环境配置方面，本章课程要求你的环境满足以下条件：

- 系统中已部署昇腾NPU硬件，或已配置昇腾云服务器/仿真环境。
- 已按照[CANN下载页面](https://www.hiascend.com/cann/download)完成对应硬件的开发环境部署。

---

## 章节目标
完成本课程后，你将能够：
- 理解非量化Flash Attention的定位：区分量化与非量化Flash Attention，明确非量化场景下Q、K、V为FP16/BF16、Softmax在FP32累加、不涉及量化/反量化的特点。
- 掌握非量化Flash Attention的算子规格与计算流程：明确输入输出的dtype/shape/format与属性，理解 QK^T → 在线Softmax → PV → 累加 → 归一化 的算法流程。
- 了解Ascend C上的实现思路：从概念层面理解Cube（Matmul）与Vector（Softmax）的协同、两个Matmul对象的使用以及Tiling的切分方向，为后续实现章节做铺垫。
