# 4.2 环境搭建：三条路径任选其一

跑 FA 算子不一定要有昇腾硬件。按条件从三条路径中选一条：

| 路径 | 适用读者 | 需要 NPU 硬件 |
|--|--|--|
| A. 本地昇腾服务器 | 有 950 / Atlas 设备 | ✅ |
| B. CANNLab 在线环境 | 无硬件，想用真实远程环境 | ❌（云端提供） |
| C. CANNSim 仿真环境 | 无硬件，本课程**默认路径** | ❌ |

> 本章 4.5 节的调测流程基于 CANNSim 仿真讲解；三条路径的工具链命令基本一致，学会一条即可迁移。

## 路径 A：本地昇腾服务器

部署顺序：**驱动 → CANN 工具链 → （可选）torch_npu**。

```bash
# 1. 安装 NPU 驱动（root 权限，按硬件型号选择包）
# 详见 https://www.hiascend.com/hardware/firmware-drivers

# 2. 安装 CANN 开发套件
# 下载页：https://www.hiascend.com/cann/download

# 3. 激活环境变量（写入 ~/.bashrc）
source /usr/local/Ascend/ascend-toolkit/set_env.sh

# 4. 验证：能看到 NPU 卡信息即成功
npu-smi info
```

| 检查项 | 期望结果 |
|--|--|
| `npu-smi info` | 列出所有 AI 处理器及状态 |
| `which ascendc` / `which msopgen` | CANN 工具链命令可用 |
| `python -c "import acl"` | Python ACL 接口可导入 |

## 路径 B：CANNLab 在线环境

CANNLab 是昇腾提供的云端开发环境（[cannlab.hiascend.com](https://cannlab.hiascend.com)），预装了驱动与 CANN 工具链，开箱即用：

1. 注册 / 登录，创建工作空间（选择带 AI Core 资源的规格）；
2. 在云端终端里直接执行 `npu-smi info` 验证；
3. 所有命令与本地服务器完全一致。

> 适合不想折腾环境、又想接触真实硬件的读者。

## 路径 C：CANNSim 仿真环境（本课程默认）

CANNSim 在 **CPU 上模拟 AI Core 行为**：不需要任何 NPU 硬件，笔记本即可完成算子的编译、运行、结果比对——精度正确性验证与真实硬件一致，只是速度慢（无硬件加速）。

```bash
# 安装 CANN 仿真组件后，验证仿真模式可用
# （CANN 安装时勾选/安装 cann-toolkit + 仿真运行时）

# 仿真运行的关键：环境变量指定仿真模式
export ASCEND_LAUNCH_MODE=SIMULATE   # 以实际安装文档为准
```

> **本课程后续章节（第 5、6 章算子开发）默认在 CANNSim 上调测**：先把正确性做对，再上真机跑性能——这也是工业界算子开发的标准节奏。

无论哪条路径，进入 4.3 节前请确认以下工具可用：
