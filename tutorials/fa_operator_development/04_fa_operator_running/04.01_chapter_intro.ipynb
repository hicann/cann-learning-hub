# 4.1 章节介绍：FA 算子运行

欢迎进入 **第二部分：FA 算子开发实战**！

前三章我们打好了地基：昇腾硬件与 Ascend C（第 1 章）、FA 数学原理（第 2 章）、算子三层代码结构（第 3 章）。从本章起，一切认知都要落到**亲手运行**上。

本章的目标只有一句话：**先跑起来再说**。

> 把官方 FA 算子（`FusedInferAttentionScore`）在仿真环境里完整跑一遍：搭好环境 → 认识工程 → 读懂接口 → 调用运行 → 与 golden 对拍验证。带着「它是怎么写出来的」这个悬念，进入第 5 章从 0 到 1 的开发。

## 本章学习目标

1. **搭好三种环境之一**：本地昇腾服务器、CANNLab 在线环境、或 CANNSim 仿真环境，完成驱动 / CANN 工具链部署检查。
2. **看懂 FA 算子工程的组织方式**：每个目录、每类文件在三层结构（调用通路 / Tiling / Kernel）中的位置。
3. **逐参解读 FA 推理接口**：从 `query` 到 `inputLayout`，每个参数的语义、约束与第 2 章概念的对应关系。
4. **完成一次完整调测**：基于 CANNSim 仿真运行 FA 算子，用 DebugTools 工具走完「tiling 编译 → tiling 运行 → kernel 编译 → kernel 运行 → dump 对拍」全流程。
5. **掌握调测方法论**：为什么 tiling 与 kernel 要分开调？编译与运行为什么要分开？出错时如何按层次隔离问题。
