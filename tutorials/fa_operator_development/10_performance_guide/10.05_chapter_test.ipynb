## 全书结语：从「跑通」到「跑满」

回望全书的三段旅程：

| 部分 | 主题 | 完成的跨越 |
|--|--|--|
| 第一部分（1~3章） | 懂硬件、懂算法、懂架构 | 从「会用」到「知道每一行代码站在哪里」 |
| 第二部分（4~6章） | 跑通官方算子、从 0 到 1、全量化 | 从「看懂」到「写对」——极简 FA 一路长成带流水、mask、尾块、负载均衡的完整算子 |
| 第三部分（7~10章） | 建模、逐引擎、均衡、指南 | 从「写对」到「写快」——每一步优化都有证据、有建模、有验收 |

性能优化没有银弹，但有纪律：**先测量、再建模、后动手；从上往下排查；一次改一项；瓶颈迁移就重来**。这套纪律不只属于 FA——它是所有 Cube/Vector 算子开发通用的工程方法论。

下一步建议：

1. 用[第4章](../04_fa_operator_running/README.md)的环境跑一遍官方 FA，采集自己的画像；
2. 按 10.4 的清单给自己手头的算子做一次体检；
3. 深入开源仓库 [ops-transformer](https://gitcode.com/cann/ops-transformer/tree/master/attention/fused_infer_attention_score)，对照真实实现验证全书的每一处设计。
