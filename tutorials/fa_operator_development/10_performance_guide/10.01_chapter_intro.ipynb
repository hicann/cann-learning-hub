# 10.1 章节介绍：性能优化指南——全书的收束

第 7~9 章把性能优化的方法论逐层展开：建模、访存、均衡、CV 流水、核内流水、逐引擎优化、负载均衡。本章做一件事：**把这些方法收拢成一份可检索、可执行的指南**——拿到一个性能问题，翻开本章就知道从哪查起、按什么顺序查、每一步用什么工具、什么症状开什么药方。

| **排查路径与决策树** | 拿到问题 → 定位 → 处方的完整流程，症状对照表 | [10.3](10.03_troubleshooting.ipynb) |
| **逐项 checklist** | Cube / Vector / Scalar / 均衡 / 建模 五张可打印清单 | [10.4](10.04_checklists.ipynb) |

## 为什么需要「指南」这一章

单点方法人人能查，但**性能优化真正的难点是顺序**：

- 先抠指令还是先看均衡？——先抠指令大概率白干（[7.1](../07_performance_basics/07.01_chapter_intro.ipynb)）；
- 先切 K 还是先查 bank 冲突？——取决于 Roofline 判断你处于哪段（[7.2](../07_performance_basics/07.02_perf_modeling_roofline.ipynb)）；
- 先合轴还是先切 S2？——取决于行数与核数的关系（[9.4](../09_performance_high_level/09.04_gs1_merged.ipynb)）。

本章把这些顺序判断固化成**决策树 + 对照表**，让「老师傅的直觉」变成「新手照着走的路」。

## 一个使用场景贯穿全章

> 「我的 FA 算子在 decode 场景实测 3.2 ms，理论上应该是 1.1 ms，从哪查起？」

10.2 用方法地图定位它可能涉及的层次；10.3 走一遍完整排查；10.4 给出每一站要过的 checklist。
