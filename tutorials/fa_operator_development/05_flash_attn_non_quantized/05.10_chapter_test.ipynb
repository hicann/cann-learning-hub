# 5.10 章节测验：非量化 FA 开发

完成以下综合练习后对照 [答案](answer/05.10_answer.txt)。

1. **（路线）** 默写七阶段开发路线，每阶段一句话说明「新增概念」与「消灭的空等」。

2. **（分核）** 为什么核间切分选择 B、N、S1 轴而不是 S2 轴？FlashDecode 为什么敢切 S2？

3. **（流水）** 给定四拍时长 C1=3、V1=8、C2=3、V2=1（单位任意），串行执行 4 个基本块的总时长是多少？Preload 一轮编排下（首块预热忽略），Cube 与 Vector 各自的忙碌时长是多少？谁在断流？

4. **（buffer）** Q 常驻 + KV 乒乓的布局里，为什么 Q 不需要乒乓？`BUFFER_NUM=2` 依赖第 1 章的哪个同步机制？

5. **（mask+尾块）** QS=100、KVS=200、块大小 128、sparsemode3 因果掩码：写出 S1 块数、S2 块数、S1 尾块行数、S2 尾块行数；哪些 (S1 块, S2 块) 组合可以整块跳过？

6. **（负载均衡）** B=2、N=16、QS=256、s1Block=128、核数 20：BN1 的 blockDim 与闲置核数？BN1S1 的 totalSize、splitFactorSize 与各核任务分布？
