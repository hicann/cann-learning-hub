# 6.7 章节测验：全量化 FA 开发

完成以下综合练习后对照 [答案](answer/06.07_answer.txt)。

1. **（格式）** 给定 E4M3 的位域 `0 1001 101`（S/E/M），按 bias=7 解码出它的值；再写出 E5M2 在同样位域下的值（bias=15）。哪个格式范围大、哪个精度高？

2. **（累加）** 用 E4M3 的 max ≈ 448 证明「FP8 × FP8 的累加器必须是 FP32」；并说明 E8M0 scale 为什么让反量化退化为移位。

3. **（原理）** 写出非对称量化与反量化的公式。设 r = [0.2, 0.8, 3.0]，amax 法求对称 INT8 量化的 scale，给出量化结果与反量化误差。

4. **（粒度）** 五种量化粒度按 scale 开销从小到大排序；解释 mx 量化 = G-G 特例的两条限定，以及「离群值只污染自己的组」为什么成立。

5. **（MXFP8 方案）** 画出（或默写）MXFP8 KV Cache 的存储布局；说明 Kernel 三处 diff（Tiling / 搬入 / scale 处理），并解释 K 的 scale 为什么必须在 softmax 前生效。

6. **（HiP8 方案）** 从「位布局 → 动态范围 → scale 需求 → 三处 diff」四步，对比 HiF8 与 MXFP8 在 FA 量化方案中的数据通路差异。

完成后进入[第7章 性能优化基础](../07_performance_basics/README.md)，开始性能优化之旅。
