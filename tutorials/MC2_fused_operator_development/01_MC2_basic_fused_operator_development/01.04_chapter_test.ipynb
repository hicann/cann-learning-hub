实现MC2融合算子MatmulAllGather，这是一个典型的矩阵乘法与集合通信AllGather融合的算子，广泛应用于分布式训练场景。

相关算法：  
<div style="width: 30%; float: left; clear: both; margin: 10px 0;">

$$
\text{MatmulAllGather}(Y) = \text{AllGather}\left(\text{Matmul}(A, B)\right)
$$
</div>
<div style="clear: both;"></div>

其中：
- Matmul: 矩阵乘法运算 $C = A \times B$，输出形状为 $(M, N)$
- AllGather: 集合通信操作，将所有rank的Matmul结果C汇聚到输出$Y$，最终输出形状为 $(rankDim \times M, N)$

**与ReduceScatter的区别**：
- ReduceScatter: 输出大小 = Matmul输出 / rankDim（切分）
- AllGather: 输出大小 = Matmul输出 × rankDim（汇聚）

### **测试脚本 - 结果验证**

验证AllGather输出结果是否正确。

验证逻辑：
1. 对每个rank，计算Matmul结果
2. 模拟AllGather：将所有rank的C按顺序排列
3. 对比NPU输出与CPU预期结果（rankDim × M × N）

````python
%%writefile Sources/01.04/matmul_all_gather/scripts/verify_precision.py
import argparse
import os
import struct
import numpy as np

BF16_ATOL = 1e-2
BF16_RTOL = 1e-2

def bf16_to_float(data):
    """将 BF16 二进制数据转换为 numpy float32 数组"""
    count = len(data) // 2
    result = np.zeros(count, dtype=np.float32)
    for i in range(count):
        h = struct.unpack('<H', data[i*2:i*2+2])[0]
        result[i] = np.float32(struct.unpack('<f', struct.pack('<I', h << 16))[0])
    return result

def float_to_bf16(data):
    """将 numpy float32 数组转换为 BF16 二进制数据"""
    result = np.zeros(len(data), dtype=np.uint16)
    for i in range(len(data)):
        x = struct.unpack('<I', struct.pack('<f', np.float32(data[i])))[0]
        result[i] = (x >> 16) & 0xFFFF
    return result.tobytes()

def load_bf16_bin(filepath):
    """加载 BF16 二进制文件"""
    with open(filepath, 'rb') as f:
        data = f.read()
    return bf16_to_float(data)

def save_bf16_bin(filepath, data):
    """保存 BF16 二进制文件"""
    bf16_data = float_to_bf16(data)
    with open(filepath, 'wb') as f:
        f.write(bf16_data)

def cpu_reference_matmul_all_gather(a_inputs, b_inputs, rank_dim, m, k, n):
    """
    CPU 参考实现：Matmul + AllGather
    每个 rank 有相同的 A (M×K) 和 B (K×N)
    每个 rank 计算相同的 C = A × B (M×N)
    AllGather 将所有 rank 的 C 汇聚到输出 (rankDim × M × N)
    """
    output = np.zeros((rank_dim, m, n), dtype=np.float32)
    
    for r in range(rank_dim):
        a = a_inputs[r].reshape(m, k)
        b = b_inputs[r].reshape(k, n)
        c = np.matmul(a, b)
        output[r, :, :] = c
    
    return output.flatten()

def verify_precision(data_dir, m, k, n):
    """验证精度"""
    rank_files = sorted([f for f in os.listdir(data_dir) if f.startswith('output_rank')])
    rank_dim = len(rank_files)

    if rank_dim == 0:
        print(f"ERROR: No output files found in {data_dir}")
        return False

    a_files = sorted([f for f in os.listdir(data_dir) if f.startswith('input_a_rank')])
    b_files = sorted([f for f in os.listdir(data_dir) if f.startswith('input_b_rank')])
    
    if len(a_files) != rank_dim or len(b_files) != rank_dim:
        print(f"ERROR: Input files count mismatch: A={len(a_files)}, B={len(b_files)}, rank={rank_dim}")
        return False

    a_inputs = []
    b_inputs = []
    for i in range(rank_dim):
        a_file = os.path.join(data_dir, f'input_a_rank{i}.bin')
        b_file = os.path.join(data_dir, f'input_b_rank{i}.bin')
        a_inputs.append(load_bf16_bin(a_file))
        b_inputs.append(load_bf16_bin(b_file))

    outputs = []
    for i in range(rank_dim):
        out_file = os.path.join(data_dir, f'output_rank{i}.bin')
        outputs.append(load_bf16_bin(out_file))

    total_elements = len(outputs[0])
    expected_elements = rank_dim * m * n  # AllGather 输出大小：rankDim × M × N
    
    if total_elements != expected_elements:
        print(f"ERROR: Output length mismatch: {total_elements} vs {expected_elements}")
        return False

    reference = cpu_reference_matmul_all_gather(a_inputs, b_inputs, rank_dim, m, k, n)
    
    ref_file = os.path.join(data_dir, 'cpu_reference.bin')
    save_bf16_bin(ref_file, reference)
    print(f"CPU reference saved to: {ref_file}")

    npu_output = outputs[0]

    max_diff = np.max(np.abs(npu_output - reference))
    mean_diff = np.mean(np.abs(npu_output - reference))

    close_count = np.sum(np.isclose(npu_output, reference, atol=BF16_ATOL, rtol=BF16_RTOL))
    match_ratio = close_count / total_elements

    print(f"Rank count: {rank_dim}")
    print(f"Matrix size: M={m}, K={k}, N={n}")
    print(f"Total elements: {total_elements}")
    print(f"Max diff: {max_diff:.6f}")
    print(f"Mean diff: {mean_diff:.6f}")
    
    pass_indices = np.where(np.isclose(npu_output, reference, atol=BF16_ATOL, rtol=BF16_RTOL))[0]
    if len(pass_indices) > 0:
        print(f"\nFirst 10 pass points:")
        for idx in pass_indices[:10]:
            rank_idx = idx // (m * n)
            row_idx = (idx % (m * n)) // n
            col_idx = (idx % (m * n)) % n
            print(f"  idx={idx} (rank={rank_idx}, row={row_idx}, col={col_idx}): npu={npu_output[idx]:.6f}, ref={reference[idx]:.6f}")
    
    print(f"\nPass rate: {match_ratio:.4f} ({close_count}/{total_elements})")
    
    if match_ratio >= 0.999:
        print("PASS: Precision verification successful")
        return True
    else:
        print("FAIL: Precision mismatch")
        mismatch_indices = np.where(~np.isclose(npu_output, reference, atol=BF16_ATOL, rtol=BF16_RTOL))[0]
        if len(mismatch_indices) > 0:
            print(f"First 10 mismatch indices: {mismatch_indices[:10]}")
            for idx in mismatch_indices[:5]:
                rank_idx = idx // (m * n)
                row_idx = (idx % (m * n)) // n
                col_idx = (idx % (m * n)) % n
                a_data = a_inputs[rank_idx].reshape(m, k)
                b_data = b_inputs[rank_idx].reshape(k, n)
                a_row = a_data[row_idx, :]
                b_col = b_data[:, col_idx]
                expected_val = np.dot(a_row, b_col)
                print(f"  idx={idx} (rank={rank_idx}, row={row_idx}, col={col_idx}):")
                print(f"    npu={npu_output[idx]:.6f}, ref={reference[idx]:.6f}, cpu_calc={expected_val:.6f}")
                print(f"    A[row={row_idx}]: {a_row[:4]}... (len={len(a_row)})")
                print(f"    B[col={col_idx}]: {b_col[:4]}... (len={len(b_col)})")
        return False

def main():
    parser = argparse.ArgumentParser(description='Verify Matmul+AllGather precision')
    parser.add_argument('--data_dir', required=True, help='Directory containing output_rank*.bin files')
    parser.add_argument('--m', type=int, required=True, help='M dimension')
    parser.add_argument('--k', type=int, required=True, help='K dimension')
    parser.add_argument('--n', type=int, required=True, help='N dimension')
    args = parser.parse_args()

    if not os.path.isdir(args.data_dir):
        print(f"ERROR: Directory not found: {args.data_dir}")
        return 1

    success = verify_precision(args.data_dir, args.m, args.k, args.n)
    return 0 if success else 1

if __name__ == '__main__':
    exit(main())
````

````python
%%writefile Sources/01.04/matmul_all_gather/run_multi_data_test.sh
#!/bin/bash
set -e

SCRIPT_DIR="$(cd "$(dirname "${BASH_SOURCE[0]}")" && pwd)"
cd "$SCRIPT_DIR"

[ -n "${ASCEND_HOME_PATH:-}" ] || { echo "ERROR: ASCEND_HOME_PATH 未设置，请先配置 CANN 环境" >&2; exit 1; }
source "${ASCEND_HOME_PATH}/set_env.sh" || { echo "ERROR: set_env.sh 执行失败" >&2; exit 1; }

# ---------- 可调参数 ----------
RANK_SIZE=4
BLOCK_NUM=64
DATA_BASE_DIR="${SCRIPT_DIR}/test_data"

# 测试用例: M K N (M 需要能被 RANK_SIZE 整除)
TEST_CASES=(
    "1024 128 1024"
)

# TEST_CASES=(
#     "1024 1024 1024"
#     "2048 1024 512"
#     "512 2048 1024"
#     "4096 4096 4096"
#     "256 256 256"
# )
# ---------- 可调参数结束 ----------

BUILD_DIR="${SCRIPT_DIR}/build"
export LD_LIBRARY_PATH=${BUILD_DIR}/lib:${BUILD_DIR}:$LD_LIBRARY_PATH

run_single_test() {
    local m=$1
    local k=$2
    local n=$3
    local out_dir="${DATA_BASE_DIR}/m${m}_k${k}_n${n}"

    echo "Testing M=${m}, K=${k}, N=${n} (m_aligned=$([ $((m % RANK_SIZE)) -eq 0 ] && echo YES || echo NO))"

    if [ $((m % RANK_SIZE)) -ne 0 ]; then
        echo "  SKIP: M=${m} not divisible by ${RANK_SIZE}"
        return 0
    fi

    mkdir -p "$out_dir"

    local pids=()
    for rank in $(seq 0 $((RANK_SIZE - 1))); do
        ${BUILD_DIR}/test_matmul_all_gather \
            --rank_id ${rank} \
            --rank_size ${RANK_SIZE} \
            --m ${m} \
            --k ${k} \
            --n ${n} \
            --output_dir "$out_dir" \
            &
        pids+=($!)
    done

    local all_ok=1
    for pid in "${pids[@]}"; do
        wait $pid 2>/dev/null || all_ok=0
    done

    if [ $all_ok -ne 1 ]; then
        echo "  FAIL: Some ranks failed"
        tail -n5 /tmp/matmul_all_gather_rank*_m${m}_k${k}_n${n}.log 2>/dev/null
        return 1
    fi

    ls ${out_dir}/output_rank*.bin >/dev/null 2>&1 || { echo "  FAIL: missing output files"; return 1; }

    python3 ${SCRIPT_DIR}/scripts/verify_precision.py --data_dir "$out_dir" --m ${m} --k ${k} --n ${n} 2>&1 || {
        echo "  FAIL: Precision verification failed"
        return 1
    }

    echo "  PASS (${out_dir}/)"
    return 0
}

# echo "=== Building ==="
# bash build.sh

# echo "=== cp ==="
cp build/lib/libascendc_kernels.so build/

echo ""
echo "Tests: ${#TEST_CASES[@]} cases, output=${DATA_BASE_DIR}/"

PASSED=0
FAILED=0
FAILED_TESTS=()

for case in "${TEST_CASES[@]}"; do
    read -r m k n <<< "$case"
    
    if run_single_test $m $k $n; then
        PASSED=$((PASSED + 1))
    else
        FAILED=$((FAILED + 1))
        FAILED_TESTS+=("M=${m},K=${k},N=${n}")
    fi
done

echo ""
echo "Summary: Passed=${PASSED}, Failed=${FAILED}${FAILED_TESTS:+ (cases: ${FAILED_TESTS[*]})}"
[ $FAILED -eq 0 ] && echo "ALL PASSED" || echo "SOME FAILED"
````

## 实践总结

通过本次实践，您已经掌握了MC2融合算子MatmulAllGather的完整开发流程：

1. **算子工程创建**：使用CMake工程结构，适合多卡分布式场景

2. **Host侧开发**：
   - Tiling参数计算（M/K/N、baseM/baseK/baseN）
   - HCCL Buffer大小动态计算
   - HCCL通信上下文创建

3. **Kernel侧开发**：
   - SimpleBlockScheduler切分调度
   - SimpleBlockMmad Matmul计算
   - Win Area状态同步
   - AllGather数据汇聚

4. **测试验证**：
   - 多卡数据生成
   - CPU模拟AllGather验证
   - 结果对比分析

**关键知识点**：
- MC2融合算子将计算与通信融合，减少内存拷贝
- AllGather输出大小 = Matmul输出 × rankDim（汇聚）
- SimpleBlockScheduler相比ReduceScatter的复杂调度更简单
- 状态同步机制：确保多rank数据就绪后再执行AllGather

**与ReduceScatter对比**：
- **ReduceScatter**：切分+Reduce，输出大小 = 输入 / rankDim
- **AllGather**：汇聚，输出大小 = 输入 × rankDim
- **通信复杂度**：ReduceScatter需要ReduceSum，AllGather仅拷贝

**扩展练习**：
- 尝试优化Matmul结果直接写入Win Area（避免GM中间拷贝）
- 扩展支持更大的M/N/K维度（如2048、4096）
- 对比不同rankSize（2卡 vs 4卡）的性能差异
