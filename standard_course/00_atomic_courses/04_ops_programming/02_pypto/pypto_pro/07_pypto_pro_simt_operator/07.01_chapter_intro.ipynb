# PyPTO Pro SIMT 算子编程

本章把线程模型用于离散地址：Gather 根据索引读取，Scatter 根据索引写入，直方图让线程竞争同一计数桶。重点是为重复索引和边界写出明确语义，再建立可复查的实现与 Golden。

---


## 前置要求

- 理解第 6 章线程索引、块内屏障与原子操作。
- 能区分静态 Tensor 容量、运行时 count 与实际索引值域。

实际执行需按 [01.06 环境配置](../01_pypto_pro_introduction/01.06_env_and_first_kernel.ipynb) 准备 Ascend 950；本章设备结果仍待实测。

---


## 章节目标

完成本章后，你将能够：

- 用 grid-stride 覆盖超过线程总数的数据且不重复。
- 实现 Gather 与数值分桶，检查负索引和上界。
- 把确定性 Scatter 拆成最大分数、同分 ID 仲裁、唯一写者三个阶段。
- 根据失败位置区分覆盖遗漏、索引越界与写入竞争。
