# PyPTO Pro 算子调试与性能优化

本章把一个可以校验的算子变成可复现的调试与优化实验。沿失败用例、地址检查、精度回归、预热计时和 profiling 逐步收集证据，再比较单槽、手动双槽与自动流水版本。

---


## 前置要求

- 已阅读前章完整 Kernel，能说明其输入输出和数据路径。
- 理解正确性断言、同步、缓冲区间和跨核事件。

实际执行需按 [01.06 环境配置](../01_pypto_pro_introduction/01.06_env_and_first_kernel.ipynb) 准备 Ascend 950；本章设备结果仍待实测。

---


## 章节目标

完成本章后，你将能够：

- 保存最小失败条件，定位首个不匹配位置与相关 Tile。
- 解释 sanitizer 报告覆盖范围及地址复用的人工证明。
- 用统一输入、精度与计时口径建立可重复基线。
- 比较手动/自动流水产物，并按实际瓶颈判断 Phase 等优化是否值得。
