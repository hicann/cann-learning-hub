# PyPTO Pro SIMD 融合算子编程

本章把第 4 章的 GEMM 结果交给 Vector 完成 Bias 与 GELU 后处理。重点是定义两个执行域的数据交接：谁负责哪几行，中间结果放在哪个槽位，什么事件允许读取或覆写。先建立单槽版本，再扩为两槽并准备同口径性能比较。

---


## 前置要求

- 能实现或读懂第 4 章单 Tile GEMM。
- 理解 TileGroup 核内同步与 READY/RELEASE 跨核事件。

实际执行需按 [01.06 环境配置](../01_pypto_pro_introduction/01.06_env_and_first_kernel.ipynb) 准备 Ascend 950；

---


## 章节目标

完成本章后，你将能够：

- 为 AIC 与两个 AIV 划分互不重叠的输出区域。
- 按中间数据生命周期选择 GM 或片上交接。
- 写出完整 GEMM+Bias/GELU 与两槽复用协议。
- 在数值一致的前提下比较单槽与两槽版本。
