# PyPTO Pro 快速入门

本章从 Host 与 Device 的分工出发，认识 950 的计算和存储路径，并完成第一个向量加法的定义、启动与结果校验。目标是读懂一次算子调用的数据流，为后续 Tile、同步和多核编程建立共同起点。

---


## 前置要求

- 能阅读 Python 函数、类型标注和简单循环。
- 理解数组 shape、dtype 与逐元素加法；能使用 Linux 命令行。

实际执行需按 [01.06 环境配置](../01_pypto_pro_introduction/01.06_env_and_first_kernel.ipynb) 准备 Ascend 950；本章设备结果仍待实测。

---


## 章节目标

完成本章后，你将能够：

- 根据输入输出写出算子规格，区分 Host 准备与 Device 执行。
- 沿数据路径识别 AIC/AIV、GM、UB 和 L0 的作用。
- 区分核间 SPMD 与核内 SIMD/SIMT，并解释一个索引覆盖示例。
- 完成首个 Kernel 的代码阅读与验收流程；具备 950 时执行并记录结果。
