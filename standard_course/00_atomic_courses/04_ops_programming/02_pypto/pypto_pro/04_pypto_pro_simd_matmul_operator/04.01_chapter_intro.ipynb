# PyPTO Pro SIMD 矩阵算子编程

前两章已经介绍 Tile 的搬运、缓冲与 Vector 计算。本章转向 Cube，围绕 `C = A @ B` 完成一条矩阵算子的开发路径：先实现单 Tile 矩阵乘，再处理 K 维累加，最后扩展到多核与非整除尺寸。

学习过程中，我们会把矩阵形状、片上地址、流水依赖和校验结果逐一对应起来，解释每一步设计的原因。

---


## 前置要求

- 能区分 GM、L1、L0A/L0B/L0C，并读懂矩阵乘的 M、K、N 三个维度。
- 已完成第 2～3 章，能够使用 TileGroup、`current()/next()` 和 `auto_mutex` 组织搬运与计算。
- 运行代码需具备 [01.06](../01_pypto_pro_introduction/01.06_env_and_first_kernel.ipynb) 说明的 Ascend 950 环境。
---


## 章节目标

完成本章后，你将能够：

- 按 Cube 的数据流规划 5 个 TileGroup，说明各段搬运、计算和同步依赖。
- 从小矩阵推导 NZ/ZN 的元素排列，区分格式转换、逻辑转置和物理存储占用。
- 实现单 Tile 矩阵乘与 K 维分块累加，并用随机输入和可手算输入检查结果。
- 将输出划分为互不重叠的多核任务，处理动态 M/N 与尾块。
