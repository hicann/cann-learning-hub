# PyPTO Pro SIMD&SIMT 协同编程与调优

本章按数据访问和输出语义划分 SIMD/SIMT 的职责。先完成同一 UB Tile 上的最小协同，再处理变长紧凑输出与稳定 TopK，明确计数、索引和数据何时全部就绪，为后两章的稀疏注意力交接做准备。

---


## 前置要求

- 能阅读第 3 章归约广播、第 6～7 章线程与原子代码。
- 理解第 5、8 章的数据生命周期及性能测量边界。

实际执行需按 [01.06 环境配置](../01_pypto_pro_introduction/01.06_env_and_first_kernel.ipynb) 准备 Ascend 950；本章设备结果仍待实测。

---


## 章节目标

完成本章后，你将能够：

- 根据访问规律、操作一致性和输出位置拆分计算段。
- 实现同一 Kernel 的 UB 交接，并解释 GM 跨阶段的额外要求。
- 验收紧凑输出的计数、来源索引与数值，而非依赖输出顺序。
- 实现有限候选的稳定 TopK，并区分命中率与硬件 occupancy。
