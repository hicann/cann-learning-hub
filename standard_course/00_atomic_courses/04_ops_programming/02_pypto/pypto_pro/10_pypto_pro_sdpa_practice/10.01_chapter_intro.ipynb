# SDPA 融合算子实战

本章把两次矩阵乘、online softmax 和跨核交接串成一个 SDPA 算子。从两块数据的手算开始，建立分块状态、地址和事件的对应关系，再用完整的 GM 中转教学基线验证无 mask 和 causal 两种情况，为后续缩短数据路径提供起点。

---


## 前置要求

- 已完成第 3～5 章的归约广播、矩阵乘和 Cube/Vector 交接。
- 能使用第 8 章的精度先行测量方法，并阅读第 9 章的 SIMD/SIMT 协同。

实际执行需按 [01.06 环境配置](../01_pypto_pro_introduction/01.06_env_and_first_kernel.ipynb) 准备 Ascend 950。

---


## 章节目标

完成本章后，你将能够：

- 推导 m、l、acc 的联合更新，解释旧贡献为何必须同时缩放。
- 按 batch/Q 块分配任务，并计算每种内存空间的真实占用。
- 读懂完整 QK→online→PV→累加链路及四个跨核事件。
- 按课程规格（RFC）验证多种形状、两种 mask 和极端输入，区分数学规格与性能实现。
