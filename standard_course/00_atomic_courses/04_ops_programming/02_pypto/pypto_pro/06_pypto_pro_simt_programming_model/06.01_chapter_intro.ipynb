# PyPTO Pro SIMT 编程模型

本章把逐线程编程落实为三个可检查的问题：每个元素由谁处理，线程之间通过什么位置交换数据，哪些线程必须一起到达屏障。通过向量加法、块内计数和分支对照实验，建立 SIMT 的索引与同步基础。

---


## 前置要求

- 理解外层 JIT Kernel、Vector 执行域和 TileGroup。
- 会计算数组下标与边界；有 CUDA 经验也需核对本课程的内嵌启动路径。

实际执行需按 [01.06 环境配置](../01_pypto_pro_introduction/01.06_env_and_first_kernel.ipynb) 准备 Ascend 950；本章设备结果仍待实测。

---


## 章节目标

完成本章后，你将能够：

- 根据实际核数与线程块尺寸推导全局索引。
- 区分 GM Tensor 与块内共享 Tile 的坐标和生命周期。
- 实现初始化、屏障、原子累加、屏障、读取的协作流程。
- 使用相同工作量比较分支版与无显式分支版，记录测量边界。
