# PyPTO Pro SIMD 编程模型

本章把第一个加法算子拆开：输入怎样成为设备侧视图，片上 Tile 如何占用地址，缓冲槽位怎样轮转，哪些依赖由自动同步管理。最后用带计算的双缓冲例子把这些概念连起来，并观察 JIT 编译与执行的边界。

---


## 前置要求

- 能读懂第 1 章加法主例的 Kernel 与 Host 调用。
- 理解 shape、dtype、字节数和 Python 类型标注。

实际执行需按 [01.06 环境配置](../01_pypto_pro_introduction/01.06_env_and_first_kernel.ipynb) 准备 Ascend 950；本章设备结果仍待实测。

---


## 章节目标

完成本章后，你将能够：

- 按规格选择 Tensor、Ptr、标量与 TilingData 参数。
- 计算视图地址与 Tile 容量，区分元素坐标、块编号和字节偏移。
- 写出单/双缓冲搬运计算，说明每个槽位何时可覆盖。
- 区分核内与跨核同步，并解释 JIT 签名改变的影响。
