# PyPTO Pro SIMD 矢量算子编程

本章围绕同一条归一化计算链比较 Tile 与 Reg 两种表达。先用块级接口完成运算，再逐段解释寄存器加载、掩码、归约与广播，最终实现不带 gamma 的行 RMSNorm，并以相同输入和 Golden 比较两个版本。

---


## 前置要求

- 理解 TileGroup、UB 地址预算与自动同步。
- 能阅读第 2 章完整 Kernel 的搬入、计算、写回与 Host 校验。

实际执行需按 [01.06 环境配置](../01_pypto_pro_introduction/01.06_env_and_first_kernel.ipynb) 准备 Ascend 950；本章设备结果仍待实测。

---


## 章节目标

完成本章后，你将能够：

- 按 64 个 FP32 lane 划分一行数据，完成 VF 的 load/compute/store。
- 组合归约、逐元素与广播操作，说明统计量布局的作用。
- 在相同规格下比较 Tile 与 VF 的数据访问及资源开销。
- 完成两种 RMSNorm 实现并准备零值、常量、随机输入验收。
