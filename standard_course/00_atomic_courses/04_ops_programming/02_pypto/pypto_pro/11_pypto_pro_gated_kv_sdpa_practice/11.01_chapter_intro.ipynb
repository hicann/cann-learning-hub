# Gated-KV SDPA 综合实战

本章在 SDPA 上加入按分数选 KV 块的教学模型。先定义稳定排名、原始 token 位置和空可见行的语义，再把独立选择 Kernel 接到第 10 章的完整注意力基线，分别验证索引和输出，并测量耗时。最后通过 gym 源码辨认教学模型与生产算子的契约差异。

---


## 前置要求

- 已完成第 9 章稳定 TopK，以及第 10 章 online softmax 和四事件基线。
- 理解 batch/Q 行共享一组门控分数，与逐 query 选择的区别。

实际执行需按 [01.06 环境配置](../01_pypto_pro_introduction/01.06_env_and_first_kernel.ipynb) 准备 Ascend 950。

---


## 章节目标

完成本章后，你将能够：

- 写出按分数降序、同分块号升序的确定性选择规则。
- 保留原始块号构造 causal，正确处理没有可见 key 的行。
- 完成选择 Kernel→注意力 Kernel 的阶段交接及分层验收。
- 区分减少计算量、实际加速比与算法输出变化，并据源码阅读 gym 契约。
